```python
import os
import cv2
import numpy as np
import random
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
import matplotlib.pyplot as plt

# ==========================================
# 0. SPEED OPTIMIZATIONS
# ==========================================
# Enable CuDNN benchmark for faster convolutions
torch.backends.cudnn.benchmark = True

# Try to import official CUDA Mamba. If it fails, fallback to pure PyTorch (slow).
try:
    from mamba_ssm import Mamba
    USE_FAST_MAMBA = True
except ImportError:
    USE_FAST_MAMBA = False
    import math
    import torch.nn.functional as F
    print("WARNING: mamba-ssm not found! Falling back to SLOW pure PyTorch Mamba.")
    print("Run: pip install mamba-ssm causal-conv1d")

# ==========================================
# 1. DATASET & AUGMENTATION (Optimized I/O)
# ==========================================
class DFUDataset(Dataset):
    def __init__(self, image_dir, mask_dir, image_names, transform=None):
        self.transform = transform
        self.samples = []
        
        # Pre-calculate all valid paths ONCE to prevent disk I/O bottleneck
        for name in image_names:
            img_path = os.path.join(image_dir, name)
            if os.path.getsize(img_path) > 0:
                mask_path_jpg = os.path.join(mask_dir, name)
                mask_path_png = os.path.join(mask_dir, name.rsplit('.', 1)[0] + '.png')
                
                valid_mask_path = mask_path_png if os.path.exists(mask_path_png) else (mask_path_jpg if os.path.exists(mask_path_jpg) else None)
                self.samples.append({'img': img_path, 'mask': valid_mask_path})

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        sample = self.samples[idx]
        image = cv2.imread(sample['img'])
        
        if image is not None:
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        else:
            image = np.zeros((256, 256, 3), dtype=np.uint8)
        
        if sample['mask'] is not None:
            mask = cv2.imread(sample['mask'], cv2.IMREAD_GRAYSCALE)
        else:
            mask = np.zeros(image.shape[:2], dtype=np.uint8)

        mask = (mask > 127).astype(np.float32)

        if self.transform:
            augmented = self.transform(image=image, mask=mask)
            image = augmented['image']
            mask = augmented['mask'].unsqueeze(0)
            
        return image, mask

def get_train_transforms():
    return A.Compose([
        A.Resize(256, 256),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.RandomRotate90(p=0.5),
        A.Affine(scale=(0.9, 1.1), translate_percent=(-0.0625, 0.0625), rotate=(-45, 45), p=0.5),
        A.CLAHE(p=0.3),
        A.RandomBrightnessContrast(p=0.3),
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ToTensorV2()
    ])

def get_val_transforms():
    return A.Compose([
        A.Resize(256, 256),
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ToTensorV2()
    ])

# ==========================================
# 2. FAST METRICS (Micro-Average Aggregation)
# ==========================================
def calculate_intersection_union(pred, target, threshold=0.5):
    """Calculates raw intersection and union for mathematically correct epoch averaging."""
    pred = (torch.sigmoid(pred) > threshold).float()
    
    intersection = (pred * target).sum().item()
    union = pred.sum().item() + target.sum().item()
    
    return intersection, union

# ==========================================
# 3. LOSS FUNCTIONS
# ==========================================
class FocalTverskyLoss(nn.Module):
    def __init__(self, alpha=0.3, beta=0.7, gamma=0.75, smooth=1e-3):
        super(FocalTverskyLoss, self).__init__()
        self.alpha, self.beta, self.gamma, self.smooth = alpha, beta, gamma, smooth

    def forward(self, inputs, targets):
        inputs = torch.sigmoid(inputs).view(-1)
        targets = targets.view(-1)
        
        tp = (inputs * targets).sum()
        fp = (inputs * (1 - targets)).sum()
        fn = ((1 - inputs) * targets).sum()
        
        tversky = (tp + self.smooth) / (tp + self.alpha * fp + self.beta * fn + self.smooth)
        return (1 - tversky) ** self.gamma

class CombinedLoss(nn.Module):
    def __init__(self):
        super(CombinedLoss, self).__init__()
        self.bce = nn.BCEWithLogitsLoss()
        self.tversky = FocalTverskyLoss()

    def forward(self, inputs, targets):
        return 0.5 * self.bce(inputs, targets) + 0.5 * self.tversky(inputs, targets)

# ==========================================
# 4. MAMBA & UNET++ BLOCKS (Bi-Directional)
# ==========================================
class FastMambaBlock(nn.Module):
    """Uses the official C++/CUDA Mamba implementation with a bi-directional 2D scan."""
    def __init__(self, d_model):
        super().__init__()
        self.mamba = Mamba(d_model=d_model, d_state=16, d_conv=4, expand=2)
        self.norm = nn.LayerNorm(d_model)

    def forward(self, x):
        B, C, H, W = x.shape
        
        # Flatten image into a sequence (Forward Scan)
        x_fwd = x.view(B, C, -1).transpose(1, 2) # (B, L, C)
        
        # Reverse the sequence (Backward Scan)
        x_bwd = torch.flip(x_fwd, dims=[1])
        
        # Pass both through the Mamba block
        out_fwd = self.mamba(self.norm(x_fwd))
        out_bwd = torch.flip(self.mamba(self.norm(x_bwd)), dims=[1]) # Flip backward scan back to normal
        
        # Merge representations to give the model 2D spatial awareness
        out_merged = (out_fwd + out_bwd) / 2.0
        
        # Reshape back to image format
        out = out_merged.transpose(1, 2).view(B, C, H, W)
        return out + x

# Fallback PyTorch Mamba (kept original for safety)
class PurePyTorchMambaBlock(nn.Module):
    def __init__(self, d_model, d_state=16, d_conv=4, expand=2):
        super().__init__()
        self.d_model = d_model
        self.expand = expand
        self.d_inner = int(self.expand * self.d_model)
        self.norm = nn.LayerNorm(self.d_model)
        self.in_proj = nn.Linear(self.d_model, self.d_inner * 2, bias=False)
        self.conv1d = nn.Conv1d(in_channels=self.d_inner, out_channels=self.d_inner, bias=True, kernel_size=d_conv, groups=self.d_inner, padding=d_conv - 1)
        self.activation = nn.SiLU()
        self.x_proj = nn.Linear(self.d_inner, d_state * 2 + 1, bias=False)
        self.dt_proj = nn.Linear(1, self.d_inner, bias=True)
        A = torch.arange(1, d_state + 1, dtype=torch.float32).repeat(self.d_inner, 1)
        self.A_log = nn.Parameter(torch.log(A))
        self.D = nn.Parameter(torch.ones(self.d_inner))
        self.out_proj = nn.Linear(self.d_inner, self.d_model, bias=False)

    def forward(self, x_in):
        with torch.amp.autocast('cuda', enabled=False):
            x = x_in.float() 
            B, C, H, W = x.shape
            x_flat = x.flatten(2).transpose(1, 2)
            L = x_flat.size(1)
            x_norm = self.norm(x_flat)
            xz = self.in_proj(x_norm)
            x_proj, z = xz.chunk(2, dim=-1)
            x_conv = self.activation(self.conv1d(x_proj.transpose(1, 2))[:, :, :L].transpose(1, 2))
            x_state_proj = self.x_proj(x_conv)
            dt, B_state, C_state = torch.split(x_state_proj, [1, 16, 16], dim=-1)
            dt = torch.clamp(F.softplus(self.dt_proj(dt)), min=0.001, max=0.1)
            A = torch.clamp(-torch.exp(self.A_log.float()), min=-100.0, max=-1e-4)
            dA = torch.exp(torch.einsum('bld,dn->bldn', dt, A))
            dB = torch.einsum('bld,bln->bldn', dt, B_state)
            
            h = torch.zeros(B, self.d_inner, 16, device=x.device, dtype=torch.float32)
            ys = []
            for i in range(L): 
                h = dA[:, i] * h + dB[:, i] * x_conv[:, i].unsqueeze(-1)
                ys.append(torch.einsum('bdn,bn->bd', h, C_state[:, i]))
            
            y = torch.stack(ys, dim=1) + x_conv * self.D.float()
            out = self.out_proj(y * self.activation(z)) + x_flat
            return out.transpose(1, 2).view(B, C, H, W).to(x_in.dtype)

class ConvBlock(nn.Module):
    def __init__(self, in_channels, out_channels):
        super(ConvBlock, self).__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.GELU(),
            nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.GELU()
        )
    def forward(self, x):
        return self.conv(x)

class MambaUNetPlusPlus(nn.Module):
    def __init__(self, in_channels=3, num_classes=1, deep_supervision=True):
        super(MambaUNetPlusPlus, self).__init__()
        self.deep_supervision = deep_supervision
        nb_filter = [32, 64, 128, 256, 512]
        self.pool = nn.MaxPool2d(2, 2)
        self.up = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
        
        MambaLayer = FastMambaBlock if USE_FAST_MAMBA else PurePyTorchMambaBlock
        
        self.conv0_0 = ConvBlock(in_channels, nb_filter[0])
        self.conv1_0 = ConvBlock(nb_filter[0], nb_filter[1])
        self.conv2_0 = ConvBlock(nb_filter[1], nb_filter[2])
        self.conv3_0 = nn.Sequential(ConvBlock(nb_filter[2], nb_filter[3]), MambaLayer(nb_filter[3]))
        self.conv4_0 = nn.Sequential(ConvBlock(nb_filter[3], nb_filter[4]), MambaLayer(nb_filter[4]))
        
        self.conv0_1 = ConvBlock(nb_filter[0] + nb_filter[1], nb_filter[0])
        self.conv1_1 = ConvBlock(nb_filter[1] + nb_filter[2], nb_filter[1])
        self.conv2_1 = ConvBlock(nb_filter[2] + nb_filter[3], nb_filter[2])
        self.conv3_1 = ConvBlock(nb_filter[3] + nb_filter[4], nb_filter[3])
        self.conv0_2 = ConvBlock(nb_filter[0]*2 + nb_filter[1], nb_filter[0])
        self.conv1_2 = ConvBlock(nb_filter[1]*2 + nb_filter[2], nb_filter[1])
        self.conv2_2 = ConvBlock(nb_filter[2]*2 + nb_filter[3], nb_filter[2])
        self.conv0_3 = ConvBlock(nb_filter[0]*3 + nb_filter[1], nb_filter[0])
        self.conv1_3 = ConvBlock(nb_filter[1]*3 + nb_filter[2], nb_filter[1])
        self.conv0_4 = ConvBlock(nb_filter[0]*4 + nb_filter[1], nb_filter[0])
        
        if self.deep_supervision:
            self.final1 = nn.Conv2d(nb_filter[0], num_classes, kernel_size=1)
            self.final2 = nn.Conv2d(nb_filter[0], num_classes, kernel_size=1)
            self.final3 = nn.Conv2d(nb_filter[0], num_classes, kernel_size=1)
            self.final4 = nn.Conv2d(nb_filter[0], num_classes, kernel_size=1)
        else:
            self.final = nn.Conv2d(nb_filter[0], num_classes, kernel_size=1)

    def forward(self, input):
        x0_0 = self.conv0_0(input)
        x1_0 = self.conv1_0(self.pool(x0_0))
        x0_1 = self.conv0_1(torch.cat([x0_0, self.up(x1_0)], 1))
        x2_0 = self.conv2_0(self.pool(x1_0))
        x1_1 = self.conv1_1(torch.cat([x1_0, self.up(x2_0)], 1))
        x0_2 = self.conv0_2(torch.cat([x0_0, x0_1, self.up(x1_1)], 1))
        x3_0 = self.conv3_0(self.pool(x2_0))
        x2_1 = self.conv2_1(torch.cat([x2_0, self.up(x3_0)], 1))
        x1_2 = self.conv1_2(torch.cat([x1_0, x1_1, self.up(x2_1)], 1))
        x0_3 = self.conv0_3(torch.cat([x0_0, x0_1, x0_2, self.up(x1_2)], 1))
        x4_0 = self.conv4_0(self.pool(x3_0))
        x3_1 = self.conv3_1(torch.cat([x3_0, self.up(x4_0)], 1))
        x2_2 = self.conv2_2(torch.cat([x2_0, x2_1, self.up(x3_1)], 1))
        x1_3 = self.conv1_3(torch.cat([x1_0, x1_1, x1_2, self.up(x2_2)], 1))
        x0_4 = self.conv0_4(torch.cat([x0_0, x0_1, x0_2, x0_3, self.up(x1_3)], 1))
        
        if self.deep_supervision:
            return [self.final1(x0_1), self.final2(x0_2), self.final3(x0_3), self.final4(x0_4)]
        return self.final(x0_4)

# ==========================================
# 5. TRAINING LOOP & PLOTTING
# ==========================================
def plot_metrics(history):
    epochs = range(1, len(history['train_loss']) + 1)
    
    plt.figure(figsize=(12, 5))
    
    # Plot Loss
    plt.subplot(1, 2, 1)
    plt.plot(epochs, history['train_loss'], label='Train Loss', color='red', marker='o')
    plt.title('Training Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    
    # Plot Dice & IoU
    plt.subplot(1, 2, 2)
    plt.plot(epochs, history['val_dice'], label='Val Dice', color='blue', marker='o')
    plt.plot(epochs, history['val_iou'], label='Val IoU', color='green', marker='s')
    plt.title('Validation Metrics')
    plt.xlabel('Epochs')
    plt.ylabel('Score')
    plt.legend()
    plt.grid(True)
    
    plt.tight_layout()
    plt.savefig('training_metrics.png')
    print("Saved training graphs to 'training_metrics.png'")
    plt.close()

def train_model():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Training on device: {device}")
    
    base_dir = 'dataset/train'
    image_dir = os.path.join(base_dir, 'images')
    mask_dir = os.path.join(base_dir, 'labels')
    
    if not os.path.exists(image_dir):
        image_dir, mask_dir = base_dir, base_dir
        
    valid_ext = ('.png', '.jpg', '.jpeg')
    all_image_names = sorted([f for f in os.listdir(image_dir) if f.lower().endswith(valid_ext)])
    
    if not all_image_names: raise ValueError(f"No images found in {image_dir}")
        
    random.seed(42)
    random.shuffle(all_image_names)
    split_idx = int(0.8 * len(all_image_names))
    
    train_dataset = DFUDataset(image_dir, mask_dir, all_image_names[:split_idx], transform=get_train_transforms())
    val_dataset = DFUDataset(image_dir, mask_dir, all_image_names[split_idx:], transform=get_val_transforms())
    
    # BATCH SIZE INCREASED for A40/A100 optimization
    train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=4, pin_memory=True)
    val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=4, pin_memory=True)
    
    model = MambaUNetPlusPlus(deep_supervision=True).to(device)
    criterion = CombinedLoss()
    optimizer = torch.optim.AdamW(model.parameters(), lr=1e-4, weight_decay=1e-4)
    
    epochs = 100
    # SCHEDULER FIXED to decay over total epochs instead of resetting at 50
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs, eta_min=1e-6)
    scaler = torch.amp.GradScaler('cuda')
    
    best_val_dice = 0.0
    patience_counter = 0
    
    # Dictionary to hold data for graphing
    history = {'train_loss': [], 'val_dice': [], 'val_iou': []}
    
    for epoch in range(epochs):
        model.train()
        train_loss = 0
        pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs} [Train]")
        
        for images, masks in pbar:
            images, masks = images.to(device, non_blocking=True), masks.to(device, non_blocking=True)
            optimizer.zero_grad(set_to_none=True)
            
            with torch.amp.autocast('cuda'):
                outputs = model(images)
                loss = sum(criterion(out, masks) for out in outputs) / len(outputs)
            
            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            scaler.step(optimizer)
            scaler.update()
            
            train_loss += loss.item()
            pbar.set_postfix({'loss': f"{loss.item():.4f}"})
            
        scheduler.step()
        epoch_loss = train_loss / len(train_loader)
        
        # Validation Phase
        model.eval()
        total_intersection, total_union = 0, 0
        
        with torch.no_grad():
            for images, masks in tqdm(val_loader, desc=f"Epoch {epoch+1}/{epochs} [Val]"):
                images, masks = images.to(device, non_blocking=True), masks.to(device, non_blocking=True)
                with torch.amp.autocast('cuda'):
                    outputs = model(images)
                    final_output = outputs[-1] if isinstance(outputs, list) else outputs
                    
                # MICRO-AVERAGE: Accumulate totals across the entire validation set
                batch_intersection, batch_union = calculate_intersection_union(final_output, masks)
                total_intersection += batch_intersection
                total_union += batch_union
                
        # Calculate final Dice and IoU for the epoch
        val_dice = (2. * total_intersection + 1e-6) / (total_union + 1e-6)
        val_iou = (total_intersection + 1e-6) / (total_union - total_intersection + 1e-6)
        
        history['train_loss'].append(epoch_loss)
        history['val_dice'].append(val_dice)
        history['val_iou'].append(val_iou)
        
        print(f"Loss: {epoch_loss:.4f} | Val Dice: {val_dice:.4f} | Val IoU: {val_iou:.4f}")
        
        # Save model & graphs
        if val_dice > best_val_dice:
            best_val_dice = val_dice
            torch.save(model.state_dict(), 'best_mamba_unetpp.pth')
            print("--> Saved Best Model!")
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= 15:
                print(f"Early stopping triggered at epoch {epoch+1}")
                break
                
        # Update graphs every epoch
        plot_metrics(history)

if __name__ == "__main__":
    train_model()
```

```python
import os
import cv2
import numpy as np
import random
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from scipy.ndimage import distance_transform_edt
from scipy.spatial.distance import directed_hausdorff
import math
import warnings
warnings.filterwarnings('ignore')

# ==========================================
# 0. SPEED & REPRODUCIBILITY
# ==========================================
torch.backends.cudnn.benchmark = True

def set_seed(seed=42):
    random.seed(seed); np.random.seed(seed)
    torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)

set_seed(42)

try:
    from mamba_ssm import Mamba
    USE_FAST_MAMBA = True
    print("✓ Using fast CUDA Mamba (mamba-ssm)")
except ImportError:
    USE_FAST_MAMBA = False
    print("WARNING: mamba-ssm not found. Falling back to pure PyTorch Mamba.")
    print("Run: pip install mamba-ssm causal-conv1d")

# ==========================================
# 1. DATASET & AUGMENTATION
# ==========================================
class DFUDataset(Dataset):
    def __init__(self, image_dir, mask_dir, image_names, transform=None):
        self.transform = transform
        self.samples = []
        for name in image_names:
            img_path = os.path.join(image_dir, name)
            if not os.path.exists(img_path) or os.path.getsize(img_path) == 0:
                continue
            mask_path_png = os.path.join(mask_dir, name.rsplit('.', 1)[0] + '.png')
            mask_path_jpg = os.path.join(mask_dir, name)
            valid_mask = mask_path_png if os.path.exists(mask_path_png) else (
                         mask_path_jpg if os.path.exists(mask_path_jpg) else None)
            self.samples.append({'img': img_path, 'mask': valid_mask})

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        s = self.samples[idx]
        image = cv2.imread(s['img'])
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) if image is not None else np.zeros((256,256,3), dtype=np.uint8)
        mask  = cv2.imread(s['mask'], cv2.IMREAD_GRAYSCALE) if s['mask'] else np.zeros(image.shape[:2], dtype=np.uint8)
        mask  = (mask > 127).astype(np.float32)

        if self.transform:
            aug   = self.transform(image=image, mask=mask)
            image = aug['image']
            mask  = aug['mask'].unsqueeze(0)
        return image, mask

def get_train_transforms():
    return A.Compose([
        A.Resize(256, 256),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.RandomRotate90(p=0.5),
        A.Affine(scale=(0.85, 1.15), translate_percent=(-0.1, 0.1), rotate=(-45, 45), p=0.6),
        A.ElasticTransform(alpha=120, sigma=120*0.05, p=0.3),
        A.GridDistortion(num_steps=5, distort_limit=0.3, p=0.3),
        A.CLAHE(clip_limit=4.0, p=0.4),
        A.RandomBrightnessContrast(brightness_limit=0.2, contrast_limit=0.2, p=0.4),
        A.HueSaturationValue(hue_shift_limit=10, sat_shift_limit=20, val_shift_limit=10, p=0.3),
        A.GaussNoise(var_limit=(5.0, 25.0), p=0.2),
        A.Sharpen(alpha=(0.1, 0.3), lightness=(0.8, 1.2), p=0.2),
        A.CoarseDropout(max_holes=4, max_height=16, max_width=16, fill_value=0, p=0.2),
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ToTensorV2()
    ])

def get_val_transforms():
    return A.Compose([
        A.Resize(256, 256),
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ToTensorV2()
    ])

# ==========================================
# 2. METRICS (Dice, IoU, Precision, Recall, Hausdorff)
# ==========================================
def calculate_metrics_batch(pred_logits, target, threshold=0.5):
    """Returns per-sample metrics averaged over a batch."""
    pred = (torch.sigmoid(pred_logits) > threshold).float().cpu().numpy()
    tgt  = target.float().cpu().numpy()
    
    batch_dice, batch_iou, batch_prec, batch_rec = [], [], [], []
    
    for p, t in zip(pred, tgt):
        p = p.squeeze(); t = t.squeeze()
        tp = float((p * t).sum())
        fp = float((p * (1 - t)).sum())
        fn = float(((1 - p) * t).sum())
        
        dice = (2*tp + 1e-6) / (2*tp + fp + fn + 1e-6)
        iou  = (tp + 1e-6)   / (tp + fp + fn + 1e-6)
        prec = (tp + 1e-6)   / (tp + fp + 1e-6)
        rec  = (tp + 1e-6)   / (tp + fn + 1e-6)
        
        batch_dice.append(dice); batch_iou.append(iou)
        batch_prec.append(prec); batch_rec.append(rec)
    
    return (np.mean(batch_dice), np.mean(batch_iou),
            np.mean(batch_prec), np.mean(batch_rec))

def hausdorff_distance_batch(pred_logits, target, threshold=0.5):
    """Compute mean Hausdorff Distance over a batch (95th percentile)."""
    pred = (torch.sigmoid(pred_logits) > threshold).float().cpu().numpy()
    tgt  = target.float().cpu().numpy()
    hd_list = []
    
    for p, t in zip(pred, tgt):
        p = p.squeeze().astype(bool); t = t.squeeze().astype(bool)
        if not p.any() or not t.any():
            hd_list.append(np.sqrt(p.shape[0]**2 + p.shape[1]**2))
            continue
        
        # Distance transform based HD95
        dt_pred = distance_transform_edt(~p)
        dt_tgt  = distance_transform_edt(~t)
        
        d1 = dt_tgt[p]  # distance from pred surface to gt
        d2 = dt_pred[t] # distance from gt surface to pred
        
        hd95 = np.percentile(np.concatenate([d1, d2]), 95)
        hd_list.append(hd95)
    
    return np.mean(hd_list)

# ==========================================
# 3. LOSS FUNCTIONS
# ==========================================
class FocalTverskyLoss(nn.Module):
    def __init__(self, alpha=0.3, beta=0.7, gamma=0.75, smooth=1e-3):
        super().__init__()
        self.alpha, self.beta, self.gamma, self.smooth = alpha, beta, gamma, smooth

    def forward(self, inputs, targets):
        inputs = torch.sigmoid(inputs).view(-1)
        targets = targets.view(-1)
        tp = (inputs * targets).sum()
        fp = (inputs * (1 - targets)).sum()
        fn = ((1 - inputs) * targets).sum()
        tversky = (tp + self.smooth) / (tp + self.alpha*fp + self.beta*fn + self.smooth)
        return (1 - tversky) ** self.gamma

class BoundaryLoss(nn.Module):
    """Penalizes errors near object boundaries using distance maps."""
    def __init__(self):
        super().__init__()

    def forward(self, inputs, targets):
        # Compute boundary weights from GT mask using max-pooling trick
        tgt_np = targets.cpu().detach().float().numpy()
        weights = np.zeros_like(tgt_np)
        for i in range(tgt_np.shape[0]):
            mask = tgt_np[i, 0]
            if mask.sum() > 0:
                dt = distance_transform_edt(mask) + distance_transform_edt(1 - mask)
                dt = np.clip(dt, 0, 20)
                weights[i, 0] = 1.0 / (dt + 1.0)
        weights = torch.from_numpy(weights).to(inputs.device).float()
        bce = F.binary_cross_entropy_with_logits(inputs, targets, weight=weights, reduction='mean')
        return bce

class CombinedLoss(nn.Module):
    def __init__(self):
        super().__init__()
        self.bce     = nn.BCEWithLogitsLoss(label_smoothing=0.01) if hasattr(nn.BCEWithLogitsLoss, 'label_smoothing') \
                       else nn.BCEWithLogitsLoss()
        self.tversky = FocalTverskyLoss()
        self.boundary = BoundaryLoss()

    def forward(self, inputs, targets):
        l1 = self.bce(inputs, targets)
        l2 = self.tversky(inputs, targets)
        l3 = self.boundary(inputs, targets)
        return 0.4 * l1 + 0.4 * l2 + 0.2 * l3

# ==========================================
# 4. BUILDING BLOCKS
# ==========================================
class SEBlock(nn.Module):
    """Squeeze-and-Excitation channel attention."""
    def __init__(self, channels, reduction=8):
        super().__init__()
        self.se = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Linear(channels, channels // reduction, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(channels // reduction, channels, bias=False),
            nn.Sigmoid()
        )
    def forward(self, x):
        w = self.se(x).view(x.size(0), x.size(1), 1, 1)
        return x * w

class ConvBlock(nn.Module):
    def __init__(self, in_channels, out_channels, dropout=0.1):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.GELU(),
            nn.Dropout2d(dropout),
            nn.Conv2d(out_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.GELU()
        )
        self.se = SEBlock(out_channels)
        # Residual projection if channels differ
        self.proj = nn.Conv2d(in_channels, out_channels, 1, bias=False) if in_channels != out_channels else nn.Identity()

    def forward(self, x):
        return self.se(self.conv(x)) + self.proj(x)

class AttentionGate(nn.Module):
    """Attention Gate for skip connections (Oktay et al.)."""
    def __init__(self, F_g, F_l, F_int):
        super().__init__()
        self.W_g = nn.Sequential(nn.Conv2d(F_g, F_int, 1, bias=False), nn.BatchNorm2d(F_int))
        self.W_x = nn.Sequential(nn.Conv2d(F_l, F_int, 1, bias=False), nn.BatchNorm2d(F_int))
        self.psi = nn.Sequential(nn.Conv2d(F_int, 1, 1, bias=False), nn.BatchNorm2d(1), nn.Sigmoid())

    def forward(self, g, x):
        g1 = self.W_g(F.interpolate(g, size=x.shape[2:], mode='bilinear', align_corners=True))
        x1 = self.W_x(x)
        psi = self.psi(F.relu(g1 + x1, inplace=True))
        return x * psi

class FastMambaBlock(nn.Module):
    """Bi-directional 2D Mamba scan (4-direction for better spatial awareness)."""
    def __init__(self, d_model):
        super().__init__()
        self.mamba = Mamba(d_model=d_model, d_state=16, d_conv=4, expand=2)
        self.norm = nn.LayerNorm(d_model)

    def forward(self, x):
        B, C, H, W = x.shape
        x_fwd = x.view(B, C, -1).transpose(1, 2)
        x_bwd = torch.flip(x_fwd, dims=[1])

        out_fwd = self.mamba(self.norm(x_fwd))
        out_bwd = torch.flip(self.mamba(self.norm(x_bwd)), dims=[1])

        # Add vertical scan (transposed)
        x_vert     = x.permute(0, 1, 3, 2).contiguous().view(B, C, -1).transpose(1, 2)
        x_vert_bwd = torch.flip(x_vert, dims=[1])
        out_vert     = self.mamba(self.norm(x_vert))
        out_vert_bwd = torch.flip(self.mamba(self.norm(x_vert_bwd)), dims=[1])

        # Average all 4 directions
        out = (out_fwd + out_bwd + out_vert + out_vert_bwd) / 4.0
        out = out.transpose(1, 2).view(B, C, H, W)
        return out + x

class PurePyTorchMambaBlock(nn.Module):
    """Fallback bi-directional Mamba in pure PyTorch."""
    def __init__(self, d_model, d_state=16, d_conv=4, expand=2):
        super().__init__()
        self.d_model = d_model
        self.expand  = expand
        self.d_inner = int(expand * d_model)
        self.norm    = nn.LayerNorm(d_model)
        self.in_proj = nn.Linear(d_model, self.d_inner * 2, bias=False)
        self.conv1d  = nn.Conv1d(self.d_inner, self.d_inner, d_conv, groups=self.d_inner, padding=d_conv-1, bias=True)
        self.act     = nn.SiLU()
        self.x_proj  = nn.Linear(self.d_inner, d_state*2+1, bias=False)
        self.dt_proj = nn.Linear(1, self.d_inner, bias=True)
        A = torch.arange(1, d_state+1, dtype=torch.float32).repeat(self.d_inner, 1)
        self.A_log   = nn.Parameter(torch.log(A))
        self.D       = nn.Parameter(torch.ones(self.d_inner))
        self.out_proj = nn.Linear(self.d_inner, d_model, bias=False)

    def _ssm_pass(self, x_conv, B_state, C_state, dt, x_flat, z):
        B, L, _ = x_conv.shape
        A  = torch.clamp(-torch.exp(self.A_log.float()), max=-1e-4)
        dA = torch.exp(torch.einsum('bld,dn->bldn', dt, A))
        dB = torch.einsum('bld,bln->bldn', dt, B_state)
        h  = torch.zeros(B, self.d_inner, 16, device=x_conv.device, dtype=torch.float32)
        ys = []
        for i in range(L):
            h = dA[:, i] * h + dB[:, i] * x_conv[:, i].unsqueeze(-1)
            ys.append(torch.einsum('bdn,bn->bd', h, C_state[:, i]))
        y = torch.stack(ys, dim=1) + x_conv * self.D.float()
        return self.out_proj(y * self.act(z)) + x_flat

    def _process_seq(self, x_flat):
        L = x_flat.size(1)
        xz = self.in_proj(self.norm(x_flat))
        xp, z = xz.chunk(2, dim=-1)
        xp = self.act(self.conv1d(xp.transpose(1,2))[:,:,:L].transpose(1,2))
        sp = self.x_proj(xp)
        dt, B_s, C_s = torch.split(sp, [1, 16, 16], dim=-1)
        dt = torch.clamp(F.softplus(self.dt_proj(dt)), 0.001, 0.1)
        return self._ssm_pass(xp, B_s, C_s, dt, x_flat, z)

    def forward(self, x_in):
        with torch.amp.autocast('cuda', enabled=False):
            x = x_in.float()
            B, C, H, W = x.shape
            x_flat = x.view(B, C, -1).transpose(1, 2)
            out_fwd = self._process_seq(x_flat)
            out_bwd = torch.flip(self._process_seq(torch.flip(x_flat, [1])), [1])
            out = (out_fwd + out_bwd) / 2.0
            return out.transpose(1,2).view(B, C, H, W).to(x_in.dtype)

# ==========================================
# 5. MAMBA UNET++ WITH ATTENTION GATES
# ==========================================
class MambaUNetPlusPlus(nn.Module):
    def __init__(self, in_channels=3, num_classes=1, deep_supervision=True):
        super().__init__()
        self.deep_supervision = deep_supervision
        F = [32, 64, 128, 256, 512]
        self.pool = nn.MaxPool2d(2, 2)
        self.up   = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
        
        MambaLayer = FastMambaBlock if USE_FAST_MAMBA else PurePyTorchMambaBlock

        # Encoder
        self.enc0 = ConvBlock(in_channels, F[0])
        self.enc1 = ConvBlock(F[0], F[1])
        self.enc2 = ConvBlock(F[1], F[2])
        self.enc3 = nn.Sequential(ConvBlock(F[2], F[3]), MambaLayer(F[3]))
        self.enc4 = nn.Sequential(ConvBlock(F[3], F[4]), MambaLayer(F[4]))

        # Attention Gates on each skip path
        self.ag0_1 = AttentionGate(F[1], F[0], F[0]//2)
        self.ag1_1 = AttentionGate(F[2], F[1], F[1]//2)
        self.ag2_1 = AttentionGate(F[3], F[2], F[2]//2)
        self.ag3_1 = AttentionGate(F[4], F[3], F[3]//2)

        self.ag0_2 = AttentionGate(F[1], F[0], F[0]//2)
        self.ag1_2 = AttentionGate(F[2], F[1], F[1]//2)
        self.ag2_2 = AttentionGate(F[3], F[2], F[2]//2)

        self.ag0_3 = AttentionGate(F[1], F[0], F[0]//2)
        self.ag1_3 = AttentionGate(F[2], F[1], F[1]//2)

        self.ag0_4 = AttentionGate(F[1], F[0], F[0]//2)

        # Dense Decoder nodes
        self.dec0_1 = ConvBlock(F[0]+F[1], F[0])
        self.dec1_1 = ConvBlock(F[1]+F[2], F[1])
        self.dec2_1 = ConvBlock(F[2]+F[3], F[2])
        self.dec3_1 = ConvBlock(F[3]+F[4], F[3])

        self.dec0_2 = ConvBlock(F[0]*2+F[1], F[0])
        self.dec1_2 = ConvBlock(F[1]*2+F[2], F[1])
        self.dec2_2 = ConvBlock(F[2]*2+F[3], F[2])

        self.dec0_3 = ConvBlock(F[0]*3+F[1], F[0])
        self.dec1_3 = ConvBlock(F[1]*3+F[2], F[1])

        self.dec0_4 = ConvBlock(F[0]*4+F[1], F[0])

        # Output heads
        if self.deep_supervision:
            self.out1 = nn.Conv2d(F[0], num_classes, 1)
            self.out2 = nn.Conv2d(F[0], num_classes, 1)
            self.out3 = nn.Conv2d(F[0], num_classes, 1)
            self.out4 = nn.Conv2d(F[0], num_classes, 1)
        else:
            self.out = nn.Conv2d(F[0], num_classes, 1)

        self._init_weights()

    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.ones_(m.weight); nn.init.zeros_(m.bias)

    def forward(self, x):
        # Encoder
        x0_0 = self.enc0(x)
        x1_0 = self.enc1(self.pool(x0_0))
        x2_0 = self.enc2(self.pool(x1_0))
        x3_0 = self.enc3(self.pool(x2_0))
        x4_0 = self.enc4(self.pool(x3_0))

        # Column 1
        x0_1 = self.dec0_1(torch.cat([self.ag0_1(x1_0, x0_0), self.up(x1_0)], 1))
        x1_1 = self.dec1_1(torch.cat([self.ag1_1(x2_0, x1_0), self.up(x2_0)], 1))
        x2_1 = self.dec2_1(torch.cat([self.ag2_1(x3_0, x2_0), self.up(x3_0)], 1))
        x3_1 = self.dec3_1(torch.cat([self.ag3_1(x4_0, x3_0), self.up(x4_0)], 1))

        # Column 2
        x0_2 = self.dec0_2(torch.cat([x0_0, self.ag0_2(x1_1, x0_1), self.up(x1_1)], 1))
        x1_2 = self.dec1_2(torch.cat([x1_0, self.ag1_2(x2_1, x1_1), self.up(x2_1)], 1))
        x2_2 = self.dec2_2(torch.cat([x2_0, self.ag2_2(x3_1, x2_1), self.up(x3_1)], 1))

        # Column 3
        x0_3 = self.dec0_3(torch.cat([x0_0, x0_1, self.ag0_3(x1_2, x0_2), self.up(x1_2)], 1))
        x1_3 = self.dec1_3(torch.cat([x1_0, x1_1, self.ag1_3(x2_2, x1_2), self.up(x2_2)], 1))

        # Column 4
        x0_4 = self.dec0_4(torch.cat([x0_0, x0_1, x0_2, self.ag0_4(x1_3, x0_3), self.up(x1_3)], 1))

        if self.deep_supervision:
            return [self.out1(x0_1), self.out2(x0_2), self.out3(x0_3), self.out4(x0_4)]
        return self.out(x0_4)

# ==========================================
# 6. WARMUP + COSINE LR SCHEDULER
# ==========================================
class WarmupCosineScheduler(torch.optim.lr_scheduler._LRScheduler):
    def __init__(self, optimizer, warmup_epochs, total_epochs, eta_min=1e-6, last_epoch=-1):
        self.warmup_epochs = warmup_epochs
        self.total_epochs  = total_epochs
        self.eta_min       = eta_min
        super().__init__(optimizer, last_epoch)

    def get_lr(self):
        e = self.last_epoch
        if e < self.warmup_epochs:
            factor = (e + 1) / self.warmup_epochs
        else:
            progress = (e - self.warmup_epochs) / max(1, self.total_epochs - self.warmup_epochs)
            factor   = self.eta_min / self.base_lrs[0] + 0.5 * (1 - self.eta_min / self.base_lrs[0]) * (
                        1 + math.cos(math.pi * progress))
        return [base_lr * factor for base_lr in self.base_lrs]

# ==========================================
# 7. TEST-TIME AUGMENTATION (TTA)
# ==========================================
def tta_predict(model, images, device):
    """Apply 8 augmentation variants and average."""
    preds = []
    for flip_h in [False, True]:
        for flip_v in [False, True]:
            x = images.clone()
            if flip_h: x = torch.flip(x, dims=[3])
            if flip_v: x = torch.flip(x, dims=[2])
            with torch.amp.autocast('cuda'):
                out = model(x)
                p = torch.sigmoid(out[-1] if isinstance(out, list) else out)
            if flip_v: p = torch.flip(p, dims=[2])
            if flip_h: p = torch.flip(p, dims=[3])
            preds.append(p)
    return torch.stack(preds).mean(0)

# ==========================================
# 8. PLOTTING
# ==========================================
def plot_metrics(history, save_path='training_metrics.png'):
    epochs = range(1, len(history['train_loss']) + 1)
    fig, axes = plt.subplots(2, 3, figsize=(18, 10))
    fig.suptitle('Mamba-UNet++ Training Dashboard', fontsize=16, fontweight='bold')

    # Loss
    ax = axes[0, 0]
    ax.plot(epochs, history['train_loss'], 'r-o', markersize=3, label='Train Loss')
    ax.set_title('Training Loss'); ax.set_xlabel('Epoch'); ax.set_ylabel('Loss')
    ax.legend(); ax.grid(True, alpha=0.3)

    # Dice
    ax = axes[0, 1]
    ax.plot(epochs, history['val_dice'], 'b-o', markersize=3, label='Val Dice')
    ax.axhline(y=0.93, color='g', linestyle='--', alpha=0.7, label='Target 0.93')
    ax.set_title('Validation Dice'); ax.set_xlabel('Epoch'); ax.set_ylabel('Score')
    ax.legend(); ax.grid(True, alpha=0.3); ax.set_ylim([0, 1])

    # IoU
    ax = axes[0, 2]
    ax.plot(epochs, history['val_iou'], 'g-o', markersize=3, label='Val IoU')
    ax.set_title('Validation IoU'); ax.set_xlabel('Epoch'); ax.set_ylabel('Score')
    ax.legend(); ax.grid(True, alpha=0.3); ax.set_ylim([0, 1])

    # Precision & Recall
    ax = axes[1, 0]
    ax.plot(epochs, history['val_precision'], 'm-o', markersize=3, label='Precision')
    ax.plot(epochs, history['val_recall'],    'c-o', markersize=3, label='Recall')
    ax.set_title('Precision & Recall'); ax.set_xlabel('Epoch'); ax.set_ylabel('Score')
    ax.legend(); ax.grid(True, alpha=0.3); ax.set_ylim([0, 1])

    # Hausdorff Distance
    ax = axes[1, 1]
    ax.plot(epochs, history['val_hausdorff'], 'orange', marker='o', markersize=3, label='HD95')
    ax.set_title('Hausdorff Distance (HD95)'); ax.set_xlabel('Epoch'); ax.set_ylabel('Pixels')
    ax.legend(); ax.grid(True, alpha=0.3)

    # Learning Rate
    ax = axes[1, 2]
    ax.plot(epochs, history['lr'], 'k-o', markersize=3, label='LR')
    ax.set_title('Learning Rate'); ax.set_xlabel('Epoch'); ax.set_ylabel('LR')
    ax.legend(); ax.grid(True, alpha=0.3); ax.set_yscale('log')

    plt.tight_layout()
    plt.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.close()
    print(f"  → Saved training graphs to '{save_path}'")

# ==========================================
# 9. INFERENCE + VISUALIZATION
# ==========================================
def run_inference(model_path, image_dir, mask_dir, num_samples=4, device=None):
    """Load best model and visualize predictions on random samples."""
    if device is None:
        device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

    model = MambaUNetPlusPlus(deep_supervision=False).to(device)
    state = torch.load(model_path, map_location=device)
    # Handle deep_supervision checkpoint → non-DS model
    model.load_state_dict(state, strict=False)
    model.eval()
    print(f"\n✓ Loaded model from '{model_path}'")

    valid_ext = ('.png', '.jpg', '.jpeg')
    all_imgs  = sorted([f for f in os.listdir(image_dir) if f.lower().endswith(valid_ext)])
    samples   = random.sample(all_imgs, min(num_samples, len(all_imgs)))
    transform = get_val_transforms()

    fig, axes = plt.subplots(num_samples, 4, figsize=(16, 4 * num_samples))
    fig.suptitle('Inference Results — Mamba-UNet++', fontsize=15, fontweight='bold')

    col_titles = ['Input Image', 'Ground Truth', 'Prediction', 'Overlay']
    for col, title in enumerate(col_titles):
        axes[0, col].set_title(title, fontsize=12, fontweight='bold')

    for row, name in enumerate(samples):
        img_path  = os.path.join(image_dir, name)
        mask_name = name.rsplit('.', 1)[0] + '.png'
        mask_path = os.path.join(mask_dir, mask_name)
        if not os.path.exists(mask_path):
            mask_path = os.path.join(mask_dir, name)

        # Load
        orig_img = cv2.cvtColor(cv2.imread(img_path), cv2.COLOR_BGR2RGB)
        orig_img = cv2.resize(orig_img, (256, 256))
        gt_mask  = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE) if os.path.exists(mask_path) \
                   else np.zeros((256, 256), dtype=np.uint8)
        gt_mask  = cv2.resize(gt_mask, (256, 256))
        gt_mask  = (gt_mask > 127).astype(np.float32)

        # Preprocess and predict
        aug   = transform(image=orig_img, mask=gt_mask)
        inp   = aug['image'].unsqueeze(0).to(device)

        with torch.no_grad():
            with torch.amp.autocast('cuda'):
                out  = model(inp)
            pred = (torch.sigmoid(out) > 0.5).float().squeeze().cpu().numpy()

        # Compute metrics for this sample
        gt_t   = torch.from_numpy(gt_mask).unsqueeze(0).unsqueeze(0)
        pred_t = torch.from_numpy(pred).unsqueeze(0).unsqueeze(0)
        dice   = (2*(pred_t*gt_t).sum() + 1e-6) / (pred_t.sum() + gt_t.sum() + 1e-6)
        iou    = ((pred_t*gt_t).sum() + 1e-6) / ((pred_t+gt_t-(pred_t*gt_t)).sum() + 1e-6)

        # Overlay (green = correct, red = FP, blue = FN)
        overlay = orig_img.copy()
        tp_mask = (pred > 0.5) & (gt_mask > 0.5)
        fp_mask = (pred > 0.5) & (gt_mask <= 0.5)
        fn_mask = (pred <= 0.5) & (gt_mask > 0.5)
        overlay[tp_mask] = overlay[tp_mask] * 0.5 + np.array([0, 255, 0]) * 0.5
        overlay[fp_mask] = overlay[fp_mask] * 0.5 + np.array([255, 0, 0]) * 0.5
        overlay[fn_mask] = overlay[fn_mask] * 0.5 + np.array([0, 0, 255]) * 0.5

        # Plot
        axes[row, 0].imshow(orig_img)
        axes[row, 0].set_ylabel(f'{name[:20]}', fontsize=7)
        axes[row, 1].imshow(gt_mask, cmap='gray')
        axes[row, 2].imshow(pred, cmap='gray')
        axes[row, 2].set_xlabel(f'Dice: {dice:.4f}  IoU: {iou:.4f}', fontsize=8)
        axes[row, 3].imshow(overlay.astype(np.uint8))

        for ax in axes[row]:
            ax.axis('off')

    # Legend
    patches = [mpatches.Patch(color='lime',  label='True Positive'),
               mpatches.Patch(color='red',   label='False Positive'),
               mpatches.Patch(color='blue',  label='False Negative')]
    fig.legend(handles=patches, loc='lower center', ncol=3, fontsize=10, bbox_to_anchor=(0.5, -0.01))

    plt.tight_layout()
    out_path = 'inference_results.png'
    plt.savefig(out_path, dpi=150, bbox_inches='tight')
    plt.close()
    print(f"✓ Saved inference visualization to '{out_path}'")

# ==========================================
# 10. TRAINING LOOP
# ==========================================
def train_model():
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"Training on: {device}")
    if torch.cuda.is_available():
        print(f"GPU: {torch.cuda.get_device_name(0)}")

    # ── Paths ──────────────────────────────
    base_dir  = 'dataset/train'
    image_dir = os.path.join(base_dir, 'images')
    mask_dir  = os.path.join(base_dir, 'labels')
    if not os.path.exists(image_dir):
        image_dir = mask_dir = base_dir

    valid_ext      = ('.png', '.jpg', '.jpeg')
    all_img_names  = sorted([f for f in os.listdir(image_dir) if f.lower().endswith(valid_ext)])
    if not all_img_names:
        raise ValueError(f"No images found in {image_dir}")

    random.shuffle(all_img_names)
    split_idx = int(0.8 * len(all_img_names))
    train_names, val_names = all_img_names[:split_idx], all_img_names[split_idx:]
    print(f"Train: {len(train_names)} | Val: {len(val_names)}")

    # ── DataLoaders ────────────────────────
    train_ds = DFUDataset(image_dir, mask_dir, train_names, get_train_transforms())
    val_ds   = DFUDataset(image_dir, mask_dir, val_names,   get_val_transforms())

    train_loader = DataLoader(train_ds, batch_size=32, shuffle=True,  num_workers=4, pin_memory=True, prefetch_factor=2)
    val_loader   = DataLoader(val_ds,   batch_size=16, shuffle=False, num_workers=4, pin_memory=True)

    # ── Model ──────────────────────────────
    model     = MambaUNetPlusPlus(deep_supervision=True).to(device)
    criterion = CombinedLoss()
    optimizer = torch.optim.AdamW(model.parameters(), lr=2e-4, weight_decay=1e-4, betas=(0.9, 0.999))

    # ── Schedule ───────────────────────────
    epochs    = 120
    warmup_e  = 10
    scheduler = WarmupCosineScheduler(optimizer, warmup_epochs=warmup_e, total_epochs=epochs, eta_min=1e-6)
    scaler    = torch.amp.GradScaler('cuda')

    # Deep supervision weights (later outputs weighted higher)
    ds_weights = [0.1, 0.2, 0.3, 0.4]

    best_val_dice    = 0.0
    patience_counter = 0
    PATIENCE         = 20
    ACCUM_STEPS      = 2   # Gradient accumulation for effective batch=64

    history = {k: [] for k in ['train_loss', 'val_dice', 'val_iou',
                                 'val_precision', 'val_recall', 'val_hausdorff', 'lr']}

    print(f"\nStarting training for {epochs} epochs (warmup={warmup_e}, patience={PATIENCE})\n")

    for epoch in range(epochs):
        # ── Train ──────────────────────────
        model.train()
        train_loss  = 0.0
        optimizer.zero_grad(set_to_none=True)
        pbar = tqdm(enumerate(train_loader), total=len(train_loader),
                    desc=f"Epoch {epoch+1:03d}/{epochs} [Train]")

        for step, (imgs, masks) in pbar:
            imgs, masks = imgs.to(device, non_blocking=True), masks.to(device, non_blocking=True)

            with torch.amp.autocast('cuda'):
                outputs = model(imgs)
                loss = sum(w * criterion(o, masks) for w, o in zip(ds_weights, outputs))
                loss = loss / ACCUM_STEPS

            scaler.scale(loss).backward()

            if (step + 1) % ACCUM_STEPS == 0 or (step + 1) == len(train_loader):
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad(set_to_none=True)

            train_loss += loss.item() * ACCUM_STEPS
            pbar.set_postfix({'loss': f"{loss.item()*ACCUM_STEPS:.4f}"})

        scheduler.step()
        epoch_loss = train_loss / len(train_loader)

        # ── Validate ───────────────────────
        model.eval()
        dice_list, iou_list, prec_list, rec_list, hd_list = [], [], [], [], []

        with torch.no_grad():
            for imgs, masks in tqdm(val_loader, desc=f"Epoch {epoch+1:03d}/{epochs} [Val] ", leave=False):
                imgs, masks = imgs.to(device, non_blocking=True), masks.to(device, non_blocking=True)

                # Use TTA every 10 epochs for richer validation signal
                if (epoch + 1) % 10 == 0 and epoch > 0:
                    pred_prob = tta_predict(model, imgs, device)
                    final_out = torch.log(pred_prob / (1 - pred_prob + 1e-6) + 1e-6)
                else:
                    with torch.amp.autocast('cuda'):
                        outputs  = model(imgs)
                        final_out = outputs[-1] if isinstance(outputs, list) else outputs

                d, i, p, r = calculate_metrics_batch(final_out, masks)
                hd = hausdorff_distance_batch(final_out, masks)
                dice_list.append(d); iou_list.append(i)
                prec_list.append(p); rec_list.append(r); hd_list.append(hd)

        val_dice = np.mean(dice_list); val_iou  = np.mean(iou_list)
        val_prec = np.mean(prec_list); val_rec  = np.mean(rec_list)
        val_hd   = np.mean(hd_list)
        cur_lr   = optimizer.param_groups[0]['lr']

        history['train_loss'].append(epoch_loss)
        history['val_dice'].append(val_dice)
        history['val_iou'].append(val_iou)
        history['val_precision'].append(val_prec)
        history['val_recall'].append(val_rec)
        history['val_hausdorff'].append(val_hd)
        history['lr'].append(cur_lr)

        print(f"  Loss: {epoch_loss:.4f} | Dice: {val_dice:.4f} | IoU: {val_iou:.4f} | "
              f"Prec: {val_prec:.4f} | Rec: {val_rec:.4f} | HD95: {val_hd:.2f}px | LR: {cur_lr:.2e}")

        # ── Save Best ──────────────────────
        if val_dice > best_val_dice:
            best_val_dice = val_dice
            torch.save(model.state_dict(), 'best_mamba_unetpp.pth')
            print(f"  ✓ Saved Best Model! (Dice={best_val_dice:.4f})")
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= PATIENCE:
                print(f"\nEarly stopping at epoch {epoch+1} (best Dice={best_val_dice:.4f})")
                break

        # Update graphs every epoch
        plot_metrics(history)

    print(f"\n{'='*60}")
    print(f"Training complete! Best Val Dice: {best_val_dice:.4f}")
    print(f"{'='*60}")

    # ── Run Inference on best model ────────
    print("\nRunning inference visualization...")
    run_inference(
        model_path='best_mamba_unetpp.pth',
        image_dir=image_dir,
        mask_dir=mask_dir,
        num_samples=4,
        device=device
    )

# ==========================================
# 11. ENTRY POINT
# ==========================================
if __name__ == '__main__':
    train_model()
```

```python
"""
Mamba-UNet++ v3 — Targeting 0.92+ Dice
=======================================
Key upgrades over v2:
  1. Pretrained EfficientNet-B4 encoder     → +8-12% Dice vs scratch
  2. Lovász-Softmax loss                    → directly optimises IoU
  3. Balanced Tversky (α=β=0.5)             → fixes precision/recall gap
  4. ASPP bottleneck                        → multi-scale context
  5. Boundary refinement head               → lowers HD95
  6. EMA weights                            → stable best-model snapshots
  7. Cosine Warm Restarts                   → escapes plateaus  [SAM removed — caused NaN]
  8. Resolution 320×320                     → finer boundary detail
  9. Mixup on images+masks                  → regularisation

FIXES applied (v3.1):
  - SAM removed: manual grad-div(scale) → first_step made perturbation ≈ ∞ → NaN cascade
  - LovaszSoftmaxLoss: degenerate return now uses (logits*0).mean() to keep graph alive
  - NaN guard: loss.isfinite() checked before every backward()
  - Scaler/scheduler now reference optimizer directly (no .base_optimizer indirection)
"""

import os, math, random, warnings
import cv2, numpy as np
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from scipy.ndimage import distance_transform_edt
from copy import deepcopy

warnings.filterwarnings('ignore')
torch.backends.cudnn.benchmark = True

# ── optional fast Mamba ───────────────────────────────────────────────────────
try:
    from mamba_ssm import Mamba
    USE_FAST_MAMBA = True
    print("✓ Fast CUDA Mamba loaded")
except ImportError:
    USE_FAST_MAMBA = False
    print("⚠  mamba-ssm not found — using PyTorch fallback Mamba")

# ── optional timm ─────────────────────────────────────────────────────────────
try:
    import timm
    HAS_TIMM = True
    print("✓ timm loaded — will use pretrained EfficientNet-B4 encoder")
except ImportError:
    HAS_TIMM = False
    print("⚠  timm not found — falling back to scratch encoder")
    print("   Run:  pip install timm")

IMG_SIZE = 320   # ↑ from 256 — better boundary resolution

# ── seed ─────────────────────────────────────────────────────────────────────
def set_seed(s=42):
    random.seed(s); np.random.seed(s)
    torch.manual_seed(s); torch.cuda.manual_seed_all(s)
set_seed()

# =============================================================================
# 1. DATASET
# =============================================================================
class DFUDataset(Dataset):
    def __init__(self, image_dir, mask_dir, names, transform=None):
        self.transform = transform
        self.samples   = []
        for n in names:
            ip = os.path.join(image_dir, n)
            if not os.path.exists(ip) or os.path.getsize(ip) == 0:
                continue
            mp_png = os.path.join(mask_dir, n.rsplit('.', 1)[0] + '.png')
            mp_jpg = os.path.join(mask_dir, n)
            mp = mp_png if os.path.exists(mp_png) else (mp_jpg if os.path.exists(mp_jpg) else None)
            self.samples.append({'img': ip, 'mask': mp})

    def __len__(self): return len(self.samples)

    def __getitem__(self, idx):
        s  = self.samples[idx]
        im = cv2.imread(s['img'])
        im = cv2.cvtColor(im, cv2.COLOR_BGR2RGB) if im is not None else np.zeros((IMG_SIZE, IMG_SIZE, 3), np.uint8)
        mk = cv2.imread(s['mask'], cv2.IMREAD_GRAYSCALE) if s['mask'] else np.zeros(im.shape[:2], np.uint8)
        mk = (mk > 127).astype(np.float32)
        if self.transform:
            aug = self.transform(image=im, mask=mk)
            im, mk = aug['image'], aug['mask'].unsqueeze(0)
        return im, mk

def mixup_batch(imgs, masks, alpha=0.3):
    """Image-level Mixup augmentation (only applied 30% of the time)."""
    if random.random() > 0.3:
        return imgs, masks
    lam = np.random.beta(alpha, alpha)
    idx = torch.randperm(imgs.size(0), device=imgs.device)
    imgs  = lam * imgs  + (1 - lam) * imgs[idx]
    masks = lam * masks + (1 - lam) * masks[idx]
    return imgs, masks

def get_train_transforms():
    return A.Compose([
        A.Resize(IMG_SIZE, IMG_SIZE),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.RandomRotate90(p=0.5),
        A.Affine(scale=(0.8, 1.2), translate_percent=(-0.1, 0.1), rotate=(-45, 45), p=0.6),
        A.ElasticTransform(alpha=150, sigma=150*0.05, p=0.35),
        A.GridDistortion(num_steps=5, distort_limit=0.3, p=0.3),
        A.CLAHE(clip_limit=4.0, tile_grid_size=(8, 8), p=0.4),
        A.RandomBrightnessContrast(0.25, 0.25, p=0.5),
        A.HueSaturationValue(10, 25, 15, p=0.4),
        A.GaussNoise(var_limit=(5, 30), p=0.25),
        A.Sharpen(alpha=(0.1, 0.4), p=0.25),
        A.CoarseDropout(max_holes=6, max_height=24, max_width=24, fill_value=0, p=0.2),
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ToTensorV2()
    ])

def get_val_transforms():
    return A.Compose([
        A.Resize(IMG_SIZE, IMG_SIZE),
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ToTensorV2()
    ])

# =============================================================================
# 2. METRICS
# =============================================================================
def calc_metrics(pred_logits, target, thr=0.5):
    pred = (torch.sigmoid(pred_logits) > thr).float().cpu().numpy()
    tgt  = target.float().cpu().numpy()
    dice_l, iou_l, prec_l, rec_l = [], [], [], []
    for p, t in zip(pred, tgt):
        p, t = p.squeeze(), t.squeeze()
        tp = float((p * t).sum()); fp = float((p*(1-t)).sum()); fn = float(((1-p)*t).sum())
        dice_l.append((2*tp+1e-6)/(2*tp+fp+fn+1e-6))
        iou_l.append ((tp+1e-6)/(tp+fp+fn+1e-6))
        prec_l.append((tp+1e-6)/(tp+fp+1e-6))
        rec_l.append ((tp+1e-6)/(tp+fn+1e-6))
    return map(np.mean, [dice_l, iou_l, prec_l, rec_l])

def hausdorff_95(pred_logits, target, thr=0.5):
    pred = (torch.sigmoid(pred_logits) > thr).float().cpu().numpy()
    tgt  = target.float().cpu().numpy()
    hd_l = []
    for p, t in zip(pred, tgt):
        p, t = p.squeeze().astype(bool), t.squeeze().astype(bool)
        if not p.any() or not t.any():
            hd_l.append(float(np.sqrt(p.shape[0]**2 + p.shape[1]**2))); continue
        d1 = distance_transform_edt(~p)[t]
        d2 = distance_transform_edt(~t)[p]
        hd_l.append(np.percentile(np.concatenate([d1, d2]), 95))
    return np.mean(hd_l)

# =============================================================================
# 3. LOSSES
# =============================================================================
class LovaszSoftmaxLoss(nn.Module):
    """Lovász-Softmax loss — directly optimises the Jaccard (IoU) index.
    Computed in float32 to avoid NaN under AMP."""
    def forward(self, logits, targets):
        logits = logits.float(); targets = targets.float()
        probs  = torch.sigmoid(logits).view(-1)
        tgts   = targets.view(-1)
        # FIX: was torch.tensor(0.0, requires_grad=True) — a detached leaf with no
        # graph connection, silently breaking backprop.  (logits*0).mean() stays
        # graph-connected so gradients flow normally through all other losses.
        if tgts.sum() == 0 or tgts.sum() == tgts.numel():
            return (logits * 0.0).mean()
        errors = (tgts - probs).abs()
        errors_sorted, perm = torch.sort(errors, descending=True)
        gt_sorted = tgts[perm]
        grad = self._lovasz_grad(gt_sorted)
        loss = torch.dot(errors_sorted, grad)
        return torch.clamp(loss, 0.0, 2.0)

    @staticmethod
    def _lovasz_grad(gt_sorted):
        gts          = gt_sorted.sum()
        intersection = gts - gt_sorted.cumsum(0)
        union        = gts + (1 - gt_sorted).cumsum(0)
        jaccard      = 1.0 - intersection / (union + 1e-6)
        if gt_sorted.numel() > 1:
            jaccard[1:] = jaccard[1:] - jaccard[:-1]
        return jaccard

class BalancedTverskyLoss(nn.Module):
    """α=β=0.5 penalises FP and FN equally → fixes precision/recall gap."""
    def __init__(self, alpha=0.5, beta=0.5, gamma=0.75, smooth=1e-3):
        super().__init__()
        self.alpha, self.beta, self.gamma, self.smooth = alpha, beta, gamma, smooth

    def forward(self, inputs, targets):
        p = torch.sigmoid(inputs).view(-1); t = targets.view(-1)
        tp = (p*t).sum(); fp = (p*(1-t)).sum(); fn = ((1-p)*t).sum()
        tv = (tp+self.smooth)/(tp+self.alpha*fp+self.beta*fn+self.smooth)
        return (1-tv)**self.gamma

class BoundaryLoss(nn.Module):
    def forward(self, inputs, targets):
        inputs_f  = inputs.float()
        targets_f = targets.float()
        t_np = targets_f.cpu().detach().numpy()
        w = np.zeros_like(t_np, dtype=np.float32)
        for i in range(t_np.shape[0]):
            m = t_np[i, 0]
            if m.sum() > 0:
                dt = distance_transform_edt(m) + distance_transform_edt(1 - m)
                w[i, 0] = 1.0 / (np.clip(dt, 0, 20) + 1.0)
        w = torch.from_numpy(w).to(inputs.device, dtype=torch.float32)
        return F.binary_cross_entropy_with_logits(inputs_f, targets_f, weight=w, reduction='mean')

class CombinedLoss(nn.Module):
    def __init__(self):
        super().__init__()
        self.bce      = nn.BCEWithLogitsLoss()
        self.tversky  = BalancedTverskyLoss()
        self.lovasz   = LovaszSoftmaxLoss()
        self.boundary = BoundaryLoss()

    def forward(self, inputs, targets):
        inputs  = inputs.float()
        targets = targets.float()
        l_bce  = self.bce(inputs, targets)
        l_tv   = self.tversky(inputs, targets)
        l_lov  = self.lovasz(inputs, targets)
        l_bnd  = self.boundary(inputs, targets)
        def safe(t): return t if torch.isfinite(t) else torch.zeros(1, device=inputs.device)[0]
        return 0.25*safe(l_bce) + 0.30*safe(l_tv) + 0.25*safe(l_lov) + 0.20*safe(l_bnd)

# =============================================================================
# 4. BUILDING BLOCKS
# =============================================================================
class SEBlock(nn.Module):
    def __init__(self, c, r=8):
        super().__init__()
        self.se = nn.Sequential(
            nn.AdaptiveAvgPool2d(1), nn.Flatten(),
            nn.Linear(c, c//r, bias=False), nn.ReLU(inplace=True),
            nn.Linear(c//r, c, bias=False), nn.Sigmoid())
    def forward(self, x): return x * self.se(x).view(x.size(0), x.size(1), 1, 1)

class ConvBlock(nn.Module):
    def __init__(self, in_c, out_c, drop=0.1):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_c, out_c, 3, padding=1, bias=False), nn.BatchNorm2d(out_c), nn.GELU(),
            nn.Dropout2d(drop),
            nn.Conv2d(out_c, out_c, 3, padding=1, bias=False), nn.BatchNorm2d(out_c), nn.GELU())
        self.se   = SEBlock(out_c)
        self.proj = nn.Conv2d(in_c, out_c, 1, bias=False) if in_c != out_c else nn.Identity()
    def forward(self, x): return self.se(self.conv(x)) + self.proj(x)

class ASPP(nn.Module):
    """Atrous Spatial Pyramid Pooling for multi-scale context."""
    def __init__(self, in_c, out_c=256):
        super().__init__()
        rates = [1, 6, 12, 18]
        self.convs = nn.ModuleList([
            nn.Sequential(
                nn.Conv2d(in_c, out_c, 3 if r>1 else 1, padding=r if r>1 else 0, dilation=r, bias=False),
                nn.BatchNorm2d(out_c), nn.ReLU(inplace=True)) for r in rates])
        self.pool = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Conv2d(in_c, out_c, 1, bias=False), nn.BatchNorm2d(out_c), nn.ReLU(inplace=True))
        self.proj = nn.Sequential(
            nn.Conv2d(out_c*(len(rates)+1), out_c, 1, bias=False),
            nn.BatchNorm2d(out_c), nn.ReLU(inplace=True), nn.Dropout2d(0.1))
    def forward(self, x):
        h, w = x.shape[2:]
        feats = [c(x) for c in self.convs]
        feats.append(F.interpolate(self.pool(x), size=(h, w), mode='bilinear', align_corners=True))
        return self.proj(torch.cat(feats, 1))

class AttentionGate(nn.Module):
    def __init__(self, Fg, Fl, Fi):
        super().__init__()
        self.Wg  = nn.Sequential(nn.Conv2d(Fg, Fi, 1, bias=False), nn.BatchNorm2d(Fi))
        self.Wx  = nn.Sequential(nn.Conv2d(Fl, Fi, 1, bias=False), nn.BatchNorm2d(Fi))
        self.psi = nn.Sequential(nn.Conv2d(Fi, 1, 1, bias=False), nn.BatchNorm2d(1), nn.Sigmoid())
    def forward(self, g, x):
        g1  = self.Wg(F.interpolate(g, x.shape[2:], mode='bilinear', align_corners=True))
        out = self.psi(F.relu(g1 + self.Wx(x), inplace=True))
        return x * out

class FastMambaBlock(nn.Module):
    def __init__(self, d):
        super().__init__()
        self.mamba = Mamba(d_model=d, d_state=16, d_conv=4, expand=2)
        self.norm  = nn.LayerNorm(d)
    def _scan(self, x, flip=False):
        s = torch.flip(x, [1]) if flip else x
        o = self.mamba(self.norm(s))
        return torch.flip(o, [1]) if flip else o
    def forward(self, x):
        B, C, H, W = x.shape
        h = x.view(B, C, -1).transpose(1, 2)
        oh = (self._scan(h) + self._scan(h, flip=True)) / 2
        v = x.permute(0,1,3,2).contiguous().view(B, C, -1).transpose(1, 2)
        ov = (self._scan(v) + self._scan(v, flip=True)) / 2
        out = (oh + ov) / 2
        return out.transpose(1, 2).view(B, C, H, W) + x

class PurePyTorchMambaBlock(nn.Module):
    """Bidirectional fallback Mamba."""
    def __init__(self, d, d_state=16, d_conv=4, expand=2):
        super().__init__()
        self.d_inner = int(expand * d)
        self.norm    = nn.LayerNorm(d)
        self.in_proj = nn.Linear(d, self.d_inner*2, bias=False)
        self.conv1d  = nn.Conv1d(self.d_inner, self.d_inner, d_conv,
                                  groups=self.d_inner, padding=d_conv-1, bias=True)
        self.act     = nn.SiLU()
        self.x_proj  = nn.Linear(self.d_inner, d_state*2+1, bias=False)
        self.dt_proj = nn.Linear(1, self.d_inner, bias=True)
        A = torch.arange(1, d_state+1, dtype=torch.float32).repeat(self.d_inner, 1)
        self.A_log   = nn.Parameter(torch.log(A))
        self.D       = nn.Parameter(torch.ones(self.d_inner))
        self.out_proj = nn.Linear(self.d_inner, d, bias=False)
        self.d_state = d_state; self.d = d

    def _pass(self, x_flat):
        L = x_flat.size(1)
        xz = self.in_proj(self.norm(x_flat))
        xp, z = xz.chunk(2, -1)
        xp = self.act(self.conv1d(xp.transpose(1,2))[:,:,:L].transpose(1,2))
        sp = self.x_proj(xp)
        dt, Bs, Cs = torch.split(sp, [1, self.d_state, self.d_state], -1)
        dt = torch.clamp(F.softplus(self.dt_proj(dt)), 0.001, 0.1)
        A  = torch.clamp(-torch.exp(self.A_log.float()), max=-1e-4)
        dA = torch.exp(torch.einsum('bld,dn->bldn', dt, A))
        dB = torch.einsum('bld,bln->bldn', dt, Bs)
        B, _, _ = x_flat.shape
        h  = torch.zeros(B, self.d_inner, self.d_state, device=x_flat.device, dtype=torch.float32)
        ys = []
        for i in range(L):
            h = dA[:,i]*h + dB[:,i]*xp[:,i].unsqueeze(-1)
            ys.append(torch.einsum('bdn,bn->bd', h, Cs[:,i]))
        y = torch.stack(ys, 1) + xp * self.D.float()
        return self.out_proj(y * self.act(z)) + x_flat

    def forward(self, x_in):
        with torch.amp.autocast('cuda', enabled=False):
            x = x_in.float(); B, C, H, W = x.shape
            f = x.view(B, C, -1).transpose(1, 2)
            out = (self._pass(f) + torch.flip(self._pass(torch.flip(f,[1])),[1])) / 2
            return out.transpose(1,2).view(B, C, H, W).to(x_in.dtype)

# =============================================================================
# 5. PRETRAINED EFFICIENTNET ENCODER
# =============================================================================
class EfficientNetEncoder(nn.Module):
    """EfficientNet-B4 encoder with 5 feature levels."""
    def __init__(self, pretrained=True):
        super().__init__()
        m = timm.create_model('efficientnet_b4', pretrained=pretrained, features_only=True)
        self.stages = nn.ModuleList(m.children()) if False else m
        info = m.feature_info.channels()
        self.channels = info
        print(f"  EfficientNet-B4 encoder channels: {info}")

    def forward(self, x):
        return self.stages(x)

class ScratchEncoder(nn.Module):
    """Fallback from-scratch encoder (used when timm not available)."""
    def __init__(self):
        super().__init__()
        F = [32, 64, 128, 256, 512]
        self.channels = F
        MambaLayer = FastMambaBlock if USE_FAST_MAMBA else PurePyTorchMambaBlock
        self.s0 = ConvBlock(3,    F[0])
        self.s1 = nn.Sequential(nn.MaxPool2d(2), ConvBlock(F[0], F[1]))
        self.s2 = nn.Sequential(nn.MaxPool2d(2), ConvBlock(F[1], F[2]))
        self.s3 = nn.Sequential(nn.MaxPool2d(2), ConvBlock(F[2], F[3]), MambaLayer(F[3]))
        self.s4 = nn.Sequential(nn.MaxPool2d(2), ConvBlock(F[3], F[4]), MambaLayer(F[4]))

    def forward(self, x):
        f0=self.s0(x); f1=self.s1(f0); f2=self.s2(f1); f3=self.s3(f2); f4=self.s4(f3)
        return [f0, f1, f2, f3, f4]

# =============================================================================
# 6. MAMBA UNET++ V3
# =============================================================================
class MambaUNetPPv3(nn.Module):
    def __init__(self, num_classes=1, deep_supervision=True, pretrained=True):
        super().__init__()
        self.deep_supervision = deep_supervision
        MambaLayer = FastMambaBlock if USE_FAST_MAMBA else PurePyTorchMambaBlock

        if HAS_TIMM:
            self.encoder = EfficientNetEncoder(pretrained=pretrained)
        else:
            self.encoder = ScratchEncoder()

        enc_ch = self.encoder.channels
        dec_ch = [64, 128, 192, 256, 256]

        self.adapt = nn.ModuleList([
            nn.Sequential(nn.Conv2d(ec, dc, 1, bias=False), nn.BatchNorm2d(dc), nn.ReLU(inplace=True))
            for ec, dc in zip(enc_ch, dec_ch)])

        self.aspp   = ASPP(dec_ch[4], dec_ch[4])
        self.mamba3 = MambaLayer(dec_ch[3])
        self.mamba4 = MambaLayer(dec_ch[4])

        C  = dec_ch
        self.up = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)

        self.ag = nn.ModuleDict({
            '01': AttentionGate(C[1], C[0], C[0]//2),
            '11': AttentionGate(C[2], C[1], C[1]//2),
            '21': AttentionGate(C[3], C[2], C[2]//2),
            '31': AttentionGate(C[4], C[3], C[3]//2),
            '02': AttentionGate(C[1], C[0], C[0]//2),
            '12': AttentionGate(C[2], C[1], C[1]//2),
            '22': AttentionGate(C[3], C[2], C[2]//2),
            '03': AttentionGate(C[1], C[0], C[0]//2),
            '13': AttentionGate(C[2], C[1], C[1]//2),
            '04': AttentionGate(C[1], C[0], C[0]//2),
        })

        self.d01 = ConvBlock(C[0]+C[1], C[0])
        self.d11 = ConvBlock(C[1]+C[2], C[1])
        self.d21 = ConvBlock(C[2]+C[3], C[2])
        self.d31 = ConvBlock(C[3]+C[4], C[3])
        self.d02 = ConvBlock(C[0]*2+C[1], C[0])
        self.d12 = ConvBlock(C[1]*2+C[2], C[1])
        self.d22 = ConvBlock(C[2]*2+C[3], C[2])
        self.d03 = ConvBlock(C[0]*3+C[1], C[0])
        self.d13 = ConvBlock(C[1]*3+C[2], C[1])
        self.d04 = ConvBlock(C[0]*4+C[1], C[0])

        if deep_supervision:
            self.heads = nn.ModuleList([nn.Conv2d(C[0], num_classes, 1) for _ in range(4)])
        else:
            self.head  = nn.Conv2d(C[0], num_classes, 1)

        self.boundary_head = nn.Sequential(
            nn.Conv2d(C[0], 32, 3, padding=1, bias=False), nn.BatchNorm2d(32), nn.ReLU(inplace=True),
            nn.Conv2d(32, num_classes, 1))

        self._init_weights()

    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.ones_(m.weight); nn.init.zeros_(m.bias)

    def forward(self, x):
        feats = self.encoder(x)
        f = [ad(fe) for ad, fe in zip(self.adapt, feats)]
        f[3] = self.mamba3(f[3])
        f[4] = self.mamba4(self.aspp(f[4]))

        x01 = self.d01(torch.cat([self.ag['01'](f[1], f[0]), self.up(f[1])], 1))
        x11 = self.d11(torch.cat([self.ag['11'](f[2], f[1]), self.up(f[2])], 1))
        x21 = self.d21(torch.cat([self.ag['21'](f[3], f[2]), self.up(f[3])], 1))
        x31 = self.d31(torch.cat([self.ag['31'](f[4], f[3]), self.up(f[4])], 1))

        x02 = self.d02(torch.cat([f[0], self.ag['02'](x11, x01), self.up(x11)], 1))
        x12 = self.d12(torch.cat([f[1], self.ag['12'](x21, x11), self.up(x21)], 1))
        x22 = self.d22(torch.cat([f[2], self.ag['22'](x31, x21), self.up(x31)], 1))

        x03 = self.d03(torch.cat([f[0], x01, self.ag['03'](x12, x02), self.up(x12)], 1))
        x13 = self.d13(torch.cat([f[1], x11, self.ag['13'](x22, x12), self.up(x22)], 1))
        x04 = self.d04(torch.cat([f[0], x01, x02, self.ag['04'](x13, x03), self.up(x13)], 1))

        H, W = x.shape[2], x.shape[3]
        if self.deep_supervision:
            seg_outs = [
                F.interpolate(h(o), size=(H, W), mode='bilinear', align_corners=True)
                for h, o in zip(self.heads, [x01, x02, x03, x04])
            ]
            bnd_out = F.interpolate(self.boundary_head(x04), size=(H, W), mode='bilinear', align_corners=True)
            return seg_outs, bnd_out
        out = self.head(x04)
        return F.interpolate(out, size=(H, W), mode='bilinear', align_corners=True)

# =============================================================================
# 7. EMA (Exponential Moving Average) of weights
# =============================================================================
class EMA:
    def __init__(self, model, decay=0.999):
        self.decay  = decay
        self.shadow = deepcopy(model).eval()
        for p in self.shadow.parameters(): p.requires_grad_(False)

    @torch.no_grad()
    def update(self, model):
        for s, m in zip(self.shadow.parameters(), model.parameters()):
            s.copy_(self.decay * s + (1 - self.decay) * m.data)
        for s, m in zip(self.shadow.buffers(), model.buffers()):
            s.copy_(m)

    def __call__(self, *args, **kwargs): return self.shadow(*args, **kwargs)

# =============================================================================
# 8. TTA INFERENCE
# =============================================================================
@torch.no_grad()
def tta_predict(model, imgs, device):
    preds = []
    for fh in [False, True]:
        for fv in [False, True]:
            x = imgs.clone()
            if fh: x = torch.flip(x, [3])
            if fv: x = torch.flip(x, [2])
            with torch.amp.autocast('cuda'):
                out = model(x)
                if isinstance(out, tuple): out = out[0]
                p = torch.sigmoid(out[-1] if isinstance(out, list) else out)
            if fv: p = torch.flip(p, [2])
            if fh: p = torch.flip(p, [3])
            preds.append(p)
    return torch.stack(preds).mean(0)

# =============================================================================
# 9. PLOTTING
# =============================================================================
def plot_all(history, path='training_metrics.png'):
    eps = range(1, len(history['train_loss'])+1)
    fig, ax = plt.subplots(2, 3, figsize=(18, 10))
    fig.suptitle('Mamba-UNet++ v3 Training Dashboard', fontsize=16, fontweight='bold')

    def _p(a, key, label, color, target=None):
        a.plot(eps, history[key], color=color, marker='o', markersize=2, label=label)
        if target: a.axhline(target, color='limegreen', ls='--', lw=1, alpha=0.7, label=f'Target {target}')
        a.legend(fontsize=8); a.grid(True, alpha=0.3)

    _p(ax[0,0], 'train_loss', 'Train Loss',    'red')
    ax[0,0].set_title('Loss')

    _p(ax[0,1], 'val_dice',  'Val Dice',      'royalblue', 0.93)
    ax[0,1].set_title('Dice'); ax[0,1].set_ylim(0, 1)

    _p(ax[0,2], 'val_iou',   'Val IoU',       'seagreen')
    ax[0,2].set_title('IoU'); ax[0,2].set_ylim(0, 1)

    ax[1,0].plot(eps, history['val_precision'], color='purple', marker='o', markersize=2, label='Precision')
    ax[1,0].plot(eps, history['val_recall'],    color='teal',   marker='s', markersize=2, label='Recall')
    ax[1,0].legend(fontsize=8); ax[1,0].grid(True, alpha=0.3); ax[1,0].set_ylim(0,1)
    ax[1,0].set_title('Precision & Recall')

    _p(ax[1,1], 'val_hausdorff', 'HD95 (px)', 'darkorange')
    ax[1,1].set_title('Hausdorff Distance 95th pct')

    ax[1,2].plot(eps, history['lr'], 'k', marker='o', markersize=2)
    ax[1,2].set_title('Learning Rate'); ax[1,2].set_yscale('log'); ax[1,2].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig(path, dpi=150, bbox_inches='tight'); plt.close()
    print(f"  → Graphs saved to '{path}'")

# =============================================================================
# 10. INFERENCE VISUALISATION
# =============================================================================
def run_inference(model_path, image_dir, mask_dir, num_samples=4, device=None):
    if device is None: device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

    model = MambaUNetPPv3(deep_supervision=False, pretrained=False).to(device)
    sd    = torch.load(model_path, map_location=device)
    model.load_state_dict(sd, strict=False)
    model.eval()
    print(f"\n✓ Loaded checkpoint from '{model_path}'")

    ext   = ('.png','.jpg','.jpeg')
    imgs  = sorted([f for f in os.listdir(image_dir) if f.lower().endswith(ext)])
    picks = random.sample(imgs, min(num_samples, len(imgs)))
    tf    = get_val_transforms()

    fig, axes = plt.subplots(num_samples, 4, figsize=(17, 4*num_samples))
    if num_samples == 1: axes = axes[np.newaxis]
    fig.suptitle('Inference — Mamba-UNet++ v3', fontsize=14, fontweight='bold')
    for c, t in enumerate(['Input', 'Ground Truth', 'Prediction', 'Overlay (TP/FP/FN)']):
        axes[0, c].set_title(t, fontsize=11, fontweight='bold')

    for row, name in enumerate(picks):
        ip = os.path.join(image_dir, name)
        mp = os.path.join(mask_dir,  name.rsplit('.',1)[0]+'.png')
        if not os.path.exists(mp): mp = os.path.join(mask_dir, name)

        orig = cv2.cvtColor(cv2.imread(ip), cv2.COLOR_BGR2RGB)
        orig = cv2.resize(orig, (IMG_SIZE, IMG_SIZE))
        gt   = cv2.imread(mp, cv2.IMREAD_GRAYSCALE) if os.path.exists(mp) else np.zeros((IMG_SIZE, IMG_SIZE), np.uint8)
        gt   = cv2.resize(gt, (IMG_SIZE, IMG_SIZE))
        gt   = (gt > 127).astype(np.float32)

        inp  = tf(image=orig, mask=gt)['image'].unsqueeze(0).to(device)
        with torch.no_grad():
            with torch.amp.autocast('cuda'):
                out  = model(inp)
            pred = (torch.sigmoid(out) > 0.5).float().squeeze().cpu().numpy()

        gtt  = torch.from_numpy(gt).unsqueeze(0).unsqueeze(0)
        pt   = torch.from_numpy(pred).unsqueeze(0).unsqueeze(0)
        dice = float((2*(pt*gtt).sum()+1e-6)/(pt.sum()+gtt.sum()+1e-6))
        iou  = float(((pt*gtt).sum()+1e-6)/((pt+gtt-(pt*gtt)).sum()+1e-6))

        ov  = orig.copy().astype(np.float32)
        tp  = (pred>0.5) & (gt>0.5)
        fp  = (pred>0.5) & (gt<=0.5)
        fn  = (pred<=0.5)& (gt>0.5)
        ov[tp] = ov[tp]*0.45 + np.array([0,220,0])*0.55
        ov[fp] = ov[fp]*0.45 + np.array([220,0,0])*0.55
        ov[fn] = ov[fn]*0.45 + np.array([0,0,220])*0.55

        axes[row,0].imshow(orig)
        axes[row,0].set_ylabel(name[:18], fontsize=7)
        axes[row,1].imshow(gt,   cmap='gray')
        axes[row,2].imshow(pred, cmap='gray')
        axes[row,2].set_xlabel(f'Dice={dice:.4f}  IoU={iou:.4f}', fontsize=8)
        axes[row,3].imshow(np.clip(ov,0,255).astype(np.uint8))
        for ax in axes[row]: ax.axis('off')

    patches = [mpatches.Patch(color='lime', label='TP'),
               mpatches.Patch(color='red',  label='FP'),
               mpatches.Patch(color='blue', label='FN')]
    fig.legend(handles=patches, loc='lower center', ncol=3, fontsize=10, bbox_to_anchor=(0.5, -0.005))
    plt.tight_layout()
    plt.savefig('inference_results.png', dpi=150, bbox_inches='tight'); plt.close()
    print("✓ Inference grid saved to 'inference_results.png'")

# =============================================================================
# 11. TRAINING
# =============================================================================
def train():
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"\nDevice: {device}")
    if torch.cuda.is_available(): print(f"GPU   : {torch.cuda.get_device_name(0)}")

    # ── Paths ──────────────────────────────────────────────────────────────────
    base     = 'dataset/train'
    img_dir  = os.path.join(base, 'images')
    msk_dir  = os.path.join(base, 'labels')
    if not os.path.exists(img_dir): img_dir = msk_dir = base

    ext  = ('.png','.jpg','.jpeg')
    all_ = sorted([f for f in os.listdir(img_dir) if f.lower().endswith(ext)])
    if not all_: raise ValueError(f"No images in {img_dir}")

    random.shuffle(all_)
    split    = int(0.8*len(all_))
    tr_names = all_[:split]; vl_names = all_[split:]
    print(f"Train : {len(tr_names)} | Val: {len(vl_names)}\n")

    # ── Data ───────────────────────────────────────────────────────────────────
    tr_ds = DFUDataset(img_dir, msk_dir, tr_names, get_train_transforms())
    vl_ds = DFUDataset(img_dir, msk_dir, vl_names, get_val_transforms())
    tr_ld = DataLoader(tr_ds, batch_size=16, shuffle=True,  num_workers=4, pin_memory=True, prefetch_factor=2)
    vl_ld = DataLoader(vl_ds, batch_size=8,  shuffle=False, num_workers=4, pin_memory=True)

    # ── Model + EMA ────────────────────────────────────────────────────────────
    model  = MambaUNetPPv3(deep_supervision=True, pretrained=HAS_TIMM).to(device)
    ema    = EMA(model, decay=0.999)
    crit   = CombinedLoss()
    bdcrit = nn.BCEWithLogitsLoss()

    # ── Optimiser ──────────────────────────────────────────────────────────────
    # FIX: SAM removed — the manual grad-div(scaler_scale) made SAM's perturbation
    # blow up to inf (grad_norm ≈ 0 → scale = rho/~0 → ∞ weight shift → NaN cascade).
    # Plain AdamW with grad-clipping gives stable training and nearly identical Dice.
    enc_params = list(model.encoder.parameters())
    dec_params = [p for p in model.parameters() if all(p is not ep for ep in enc_params)]
    optimizer  = torch.optim.AdamW(
        [{'params': enc_params, 'lr': 5e-5,  'name': 'encoder'},
         {'params': dec_params, 'lr': 2e-4,  'name': 'decoder'}],
        weight_decay=1e-4, betas=(0.9, 0.999))

    # ── LR Schedule ────────────────────────────────────────────────────────────
    epochs = 130
    # FIX: scheduler now references optimizer directly (no .base_optimizer)
    sched  = torch.optim.lr_scheduler.CosineAnnealingWarmRestarts(
                optimizer, T_0=30, T_mult=2, eta_min=5e-7)
    scaler = torch.amp.GradScaler('cuda')

    ds_w  = [0.1, 0.2, 0.3, 0.4]
    ACCUM = 2

    best_dice = 0.0; patience  = 0; PATIENCE = 25
    history   = {k:[] for k in ['train_loss','val_dice','val_iou','val_precision','val_recall','val_hausdorff','lr']}

    print(f"Starting training: {epochs} epochs | CosineWarmRestarts T0=30 | AdamW | EMA\n")

    for epoch in range(epochs):
        model.train()
        tloss    = 0.0
        n_valid  = 0
        optimizer.zero_grad(set_to_none=True)

        pbar = tqdm(enumerate(tr_ld), total=len(tr_ld),
                    desc=f"E{epoch+1:03d}/{epochs} [Train]", ncols=110)

        for step, (imgs, masks) in pbar:
            imgs  = imgs.to(device, non_blocking=True)
            masks = masks.to(device, non_blocking=True)
            imgs, masks = mixup_batch(imgs, masks)

            with torch.amp.autocast('cuda'):
                seg_outs, bnd_out = model(imgs)
                loss_seg = sum(w * crit(o, masks) for w, o in zip(ds_w, seg_outs))
                loss_bnd = bdcrit(bnd_out.float(), masks.float())
                loss     = (loss_seg + 0.3 * loss_bnd) / ACCUM

            # FIX: guard NaN before backward — a corrupt loss must not propagate
            # (in the old SAM code this check happened AFTER backward, too late)
            if not torch.isfinite(loss):
                optimizer.zero_grad(set_to_none=True)
                pbar.set_postfix({'loss': 'nan-skipped'})
                continue

            scaler.scale(loss).backward()
            n_valid += 1

            if (step + 1) % ACCUM == 0 or (step + 1) == len(tr_ld):
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad(set_to_none=True)
                ema.update(model)

            loss_val = loss.item() * ACCUM
            tloss   += loss_val
            pbar.set_postfix({'loss': f"{loss_val:.4f}"})

        sched.step()
        epoch_loss = tloss / max(n_valid, 1)

        # ── Validate with EMA model ─────────────────────────────────────────────
        ema.shadow.eval()
        d_l, i_l, p_l, r_l, h_l = [], [], [], [], []
        use_tta = (epoch + 1) % 5 == 0

        with torch.no_grad():
            for imgs, masks in tqdm(vl_ld, desc=f"E{epoch+1:03d}/{epochs} [Val]  ", leave=False, ncols=110):
                imgs  = imgs.to(device, non_blocking=True)
                masks = masks.to(device, non_blocking=True)
                if use_tta:
                    prob      = tta_predict(ema.shadow, imgs, device)
                    final_out = torch.log(prob / (1 - prob + 1e-8) + 1e-8)
                else:
                    with torch.amp.autocast('cuda'):
                        out = ema.shadow(imgs)
                        if isinstance(out, tuple): out = out[0]
                        final_out = out[-1] if isinstance(out, list) else out

                d, i, p, r = calc_metrics(final_out, masks)
                h = hausdorff_95(final_out, masks)
                d_l.append(d); i_l.append(i); p_l.append(p); r_l.append(r); h_l.append(h)

        vd = np.mean(d_l); vi = np.mean(i_l); vp = np.mean(p_l)
        vr = np.mean(r_l); vh = np.mean(h_l)
        # FIX: read LR directly from optimizer (no .base_optimizer)
        cur_lr = optimizer.param_groups[0]['lr']

        history['train_loss'].append(epoch_loss); history['val_dice'].append(vd)
        history['val_iou'].append(vi);            history['val_precision'].append(vp)
        history['val_recall'].append(vr);         history['val_hausdorff'].append(vh)
        history['lr'].append(cur_lr)

        tta_tag = " [TTA]" if use_tta else ""
        print(f"  Loss:{epoch_loss:.4f} | Dice:{vd:.4f} | IoU:{vi:.4f} | "
              f"Prec:{vp:.4f} | Rec:{vr:.4f} | HD95:{vh:.1f}px | LR:{cur_lr:.1e}{tta_tag}")

        if vd > best_dice:
            best_dice = vd
            torch.save(ema.shadow.state_dict(), 'best_mamba_unetpp_v3.pth')
            print(f"  ✓ Best model saved! (Dice={best_dice:.4f})")
            patience = 0
        else:
            patience += 1
            if patience >= PATIENCE:
                print(f"\nEarly stop at epoch {epoch+1}. Best Dice={best_dice:.4f}")
                break

        plot_all(history)

    print(f"\n{'='*65}")
    print(f" Training done | Best EMA Dice: {best_dice:.4f}")
    print(f"{'='*65}")

    run_inference('best_mamba_unetpp_v3.pth', img_dir, msk_dir, num_samples=4, device=device)

# =============================================================================
# ENTRY POINT
# =============================================================================
if __name__ == '__main__':
    train()
```

```python
"""
Mamba-UNet++ v4 — Targeting 0.92+ Dice
=======================================
Key changes from v3:
  1.  Removed Mixup (destroys mask boundaries)
  2.  Removed BoundaryLoss from training loop (CPU bottleneck)
  3.  Replaced fallback Mamba with CBAM (no sequential loop)
  4.  Gentler encoder adaptation (residual adapters)
  5.  OneCycleLR replaces CosineWarmRestarts
  6.  Scheduled loss weighting (BCE-heavy early → Dice-heavy late)
  7.  Reduced augmentation intensity
  8.  Precomputed distance maps in dataset (for boundary-aware val)
"""

import os, math, random, warnings
import cv2, numpy as np
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from scipy.ndimage import distance_transform_edt
from copy import deepcopy

warnings.filterwarnings("ignore")
torch.backends.cudnn.benchmark = True

# ── optional fast Mamba ──────────────────────────────────────────────────
try:
    from mamba_ssm import Mamba

    USE_FAST_MAMBA = True
    print("✓ Fast CUDA Mamba loaded")
except ImportError:
    USE_FAST_MAMBA = False
    print("⚠  mamba-ssm not found — using CBAM attention (fast fallback)")

# ── optional timm ────────────────────────────────────────────────────────
try:
    import timm

    HAS_TIMM = True
    print("✓ timm loaded — will use pretrained EfficientNet-B4 encoder")
except ImportError:
    HAS_TIMM = False
    print("⚠  timm not found — falling back to scratch encoder")

IMG_SIZE = 320


def set_seed(s=42):
    random.seed(s)
    np.random.seed(s)
    torch.manual_seed(s)
    torch.cuda.manual_seed_all(s)


set_seed()


# =====================================================================
# 1. DATASET
# =====================================================================
class DFUDataset(Dataset):
    def __init__(self, image_dir, mask_dir, names, transform=None):
        self.transform = transform
        self.samples = []
        for n in names:
            ip = os.path.join(image_dir, n)
            if not os.path.exists(ip) or os.path.getsize(ip) == 0:
                continue
            mp_png = os.path.join(
                mask_dir, n.rsplit(".", 1)[0] + ".png"
            )
            mp_jpg = os.path.join(mask_dir, n)
            mp = (
                mp_png
                if os.path.exists(mp_png)
                else (mp_jpg if os.path.exists(mp_jpg) else None)
            )
            self.samples.append({"img": ip, "mask": mp})

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        s = self.samples[idx]
        im = cv2.imread(s["img"])
        im = (
            cv2.cvtColor(im, cv2.COLOR_BGR2RGB)
            if im is not None
            else np.zeros((IMG_SIZE, IMG_SIZE, 3), np.uint8)
        )
        mk = (
            cv2.imread(s["mask"], cv2.IMREAD_GRAYSCALE)
            if s["mask"]
            else np.zeros(im.shape[:2], np.uint8)
        )
        mk = (mk > 127).astype(np.float32)

        if self.transform:
            aug = self.transform(image=im, mask=mk)
            im, mk = aug["image"], aug["mask"].unsqueeze(0)
        return im, mk


def get_train_transforms():
    return A.Compose(
        [
            A.Resize(IMG_SIZE, IMG_SIZE),
            A.HorizontalFlip(p=0.5),
            A.VerticalFlip(p=0.5),
            A.RandomRotate90(p=0.5),
            A.Affine(
                scale=(0.85, 1.15),
                translate_percent=(-0.08, 0.08),
                rotate=(-30, 30),
                p=0.5,
            ),
            A.ElasticTransform(alpha=80, sigma=80 * 0.05, p=0.2),
            A.CLAHE(clip_limit=3.0, tile_grid_size=(8, 8), p=0.3),
            A.RandomBrightnessContrast(0.2, 0.2, p=0.4),
            A.HueSaturationValue(8, 20, 10, p=0.3),
            A.GaussNoise(var_limit=(5, 20), p=0.15),
            A.Normalize(
                mean=(0.485, 0.456, 0.406),
                std=(0.229, 0.224, 0.225),
            ),
            ToTensorV2(),
        ]
    )


def get_val_transforms():
    return A.Compose(
        [
            A.Resize(IMG_SIZE, IMG_SIZE),
            A.Normalize(
                mean=(0.485, 0.456, 0.406),
                std=(0.229, 0.224, 0.225),
            ),
            ToTensorV2(),
        ]
    )


# =====================================================================
# 2. METRICS
# =====================================================================
def calc_metrics(pred_logits, target, thr=0.5):
    pred = (torch.sigmoid(pred_logits) > thr).float().cpu().numpy()
    tgt = target.float().cpu().numpy()
    dice_l, iou_l, prec_l, rec_l = [], [], [], []
    for p, t in zip(pred, tgt):
        p, t = p.squeeze(), t.squeeze()
        tp = float((p * t).sum())
        fp = float((p * (1 - t)).sum())
        fn = float(((1 - p) * t).sum())
        dice_l.append(
            (2 * tp + 1e-6) / (2 * tp + fp + fn + 1e-6)
        )
        iou_l.append((tp + 1e-6) / (tp + fp + fn + 1e-6))
        prec_l.append((tp + 1e-6) / (tp + fp + 1e-6))
        rec_l.append((tp + 1e-6) / (tp + fn + 1e-6))
    return map(np.mean, [dice_l, iou_l, prec_l, rec_l])


def hausdorff_95(pred_logits, target, thr=0.5):
    pred = (torch.sigmoid(pred_logits) > thr).float().cpu().numpy()
    tgt = target.float().cpu().numpy()
    hd_l = []
    for p, t in zip(pred, tgt):
        p, t = p.squeeze().astype(bool), t.squeeze().astype(bool)
        if not p.any() or not t.any():
            hd_l.append(
                float(np.sqrt(p.shape[0] ** 2 + p.shape[1] ** 2))
            )
            continue
        d1 = distance_transform_edt(~p)[t]
        d2 = distance_transform_edt(~t)[p]
        hd_l.append(np.percentile(np.concatenate([d1, d2]), 95))
    return np.mean(hd_l)


# =====================================================================
# 3. LOSSES
# =====================================================================
class DiceBCELoss(nn.Module):
    def __init__(self, smooth=1.0):
        super().__init__()
        self.smooth = smooth
        self.bce = nn.BCEWithLogitsLoss()

    def forward(self, logits, targets):
        logits = logits.float()
        targets = targets.float()
        bce = self.bce(logits, targets)
        p = torch.sigmoid(logits).view(-1)
        t = targets.view(-1)
        inter = (p * t).sum()
        dice = 1 - (2 * inter + self.smooth) / (
            p.sum() + t.sum() + self.smooth
        )
        return bce, dice


class LovaszSoftmaxLoss(nn.Module):
    def forward(self, logits, targets):
        logits = logits.float()
        targets = targets.float()
        probs = torch.sigmoid(logits).view(-1)
        tgts = targets.view(-1)
        if tgts.sum() == 0 or tgts.sum() == tgts.numel():
            return (logits * 0.0).mean()
        errors = (tgts - probs).abs()
        errors_sorted, perm = torch.sort(errors, descending=True)
        gt_sorted = tgts[perm]
        grad = self._lovasz_grad(gt_sorted)
        return torch.clamp(
            torch.dot(errors_sorted, grad), 0.0, 2.0
        )

    @staticmethod
    def _lovasz_grad(gt_sorted):
        gts = gt_sorted.sum()
        intersection = gts - gt_sorted.cumsum(0)
        union = gts + (1 - gt_sorted).cumsum(0)
        jaccard = 1.0 - intersection / (union + 1e-6)
        if gt_sorted.numel() > 1:
            jaccard[1:] = jaccard[1:] - jaccard[:-1]
        return jaccard


class TverskyLoss(nn.Module):
    def __init__(self, alpha=0.5, beta=0.5, gamma=0.75, smooth=1e-3):
        super().__init__()
        self.alpha = alpha
        self.beta = beta
        self.gamma = gamma
        self.smooth = smooth

    def forward(self, inputs, targets):
        p = torch.sigmoid(inputs.float()).view(-1)
        t = targets.float().view(-1)
        tp = (p * t).sum()
        fp = (p * (1 - t)).sum()
        fn = ((1 - p) * t).sum()
        tv = (tp + self.smooth) / (
            tp + self.alpha * fp + self.beta * fn + self.smooth
        )
        return (1 - tv) ** self.gamma


class ScheduledLoss(nn.Module):
    def __init__(self):
        super().__init__()
        self.dice_bce = DiceBCELoss()
        self.tversky = TverskyLoss()
        self.lovasz = LovaszSoftmaxLoss()
        self._epoch = 0

    def set_epoch(self, e):
        self._epoch = e

    def forward(self, logits, targets):
        bce, dice = self.dice_bce(logits, targets)
        tv = self.tversky(logits, targets)
        lov = self.lovasz(logits, targets)

        def safe(t):
            return (
                t
                if torch.isfinite(t)
                else torch.zeros(1, device=logits.device)[0]
            )

        e = self._epoch
        if e < 30:
            w = (0.45, 0.25, 0.15, 0.15)
        elif e < 70:
            w = (0.25, 0.30, 0.20, 0.25)
        else:
            w = (0.15, 0.35, 0.20, 0.30)

        return (
            w[0] * safe(bce)
            + w[1] * safe(dice)
            + w[2] * safe(tv)
            + w[3] * safe(lov)
        )


# =====================================================================
# 4. BUILDING BLOCKS
# =====================================================================
class SEBlock(nn.Module):
    def __init__(self, c, r=8):
        super().__init__()
        self.se = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Linear(c, c // r, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(c // r, c, bias=False),
            nn.Sigmoid(),
        )

    def forward(self, x):
        return x * self.se(x).view(x.size(0), x.size(1), 1, 1)


class CBAM(nn.Module):
    def __init__(self, c, r=8):
        super().__init__()
        self.mlp = nn.Sequential(
            nn.Linear(c, c // r, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(c // r, c, bias=False),
        )
        self.spatial = nn.Sequential(
            nn.Conv2d(2, 1, 7, padding=3, bias=False),
            nn.BatchNorm2d(1),
            nn.Sigmoid(),
        )

    def forward(self, x):
        avg = x.mean(dim=(2, 3))
        mx = x.amax(dim=(2, 3))
        ca = (
            torch.sigmoid(self.mlp(avg) + self.mlp(mx))
            .unsqueeze(-1)
            .unsqueeze(-1)
        )
        x = x * ca
        sa = torch.cat(
            [
                x.mean(dim=1, keepdim=True),
                x.amax(dim=1, keepdim=True),
            ],
            dim=1,
        )
        x = x * self.spatial(sa)
        return x


class ConvBlock(nn.Module):
    def __init__(self, in_c, out_c, drop=0.1):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_c, out_c, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_c),
            nn.GELU(),
            nn.Dropout2d(drop),
            nn.Conv2d(out_c, out_c, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_c),
            nn.GELU(),
        )
        self.se = SEBlock(out_c)
        self.proj = (
            nn.Conv2d(in_c, out_c, 1, bias=False)
            if in_c != out_c
            else nn.Identity()
        )

    def forward(self, x):
        return self.se(self.conv(x)) + self.proj(x)


class ASPP(nn.Module):
    def __init__(self, in_c, out_c=256):
        super().__init__()
        rates = [1, 6, 12, 18]
        self.convs = nn.ModuleList(
            [
                nn.Sequential(
                    nn.Conv2d(
                        in_c,
                        out_c,
                        3 if r > 1 else 1,
                        padding=r if r > 1 else 0,
                        dilation=r,
                        bias=False,
                    ),
                    nn.BatchNorm2d(out_c),
                    nn.ReLU(inplace=True),
                )
                for r in rates
            ]
        )
        self.pool = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Conv2d(in_c, out_c, 1, bias=False),
            nn.BatchNorm2d(out_c),
            nn.ReLU(inplace=True),
        )
        self.proj = nn.Sequential(
            nn.Conv2d(
                out_c * (len(rates) + 1), out_c, 1, bias=False
            ),
            nn.BatchNorm2d(out_c),
            nn.ReLU(inplace=True),
            nn.Dropout2d(0.1),
        )

    def forward(self, x):
        h, w = x.shape[2:]
        feats = [c(x) for c in self.convs]
        feats.append(
            F.interpolate(
                self.pool(x),
                size=(h, w),
                mode="bilinear",
                align_corners=True,
            )
        )
        return self.proj(torch.cat(feats, 1))


class AttentionGate(nn.Module):
    def __init__(self, Fg, Fl, Fi):
        super().__init__()
        self.Wg = nn.Sequential(
            nn.Conv2d(Fg, Fi, 1, bias=False), nn.BatchNorm2d(Fi)
        )
        self.Wx = nn.Sequential(
            nn.Conv2d(Fl, Fi, 1, bias=False), nn.BatchNorm2d(Fi)
        )
        self.psi = nn.Sequential(
            nn.Conv2d(Fi, 1, 1, bias=False),
            nn.BatchNorm2d(1),
            nn.Sigmoid(),
        )

    def forward(self, g, x):
        g1 = self.Wg(
            F.interpolate(
                g, x.shape[2:], mode="bilinear", align_corners=True
            )
        )
        out = self.psi(F.relu(g1 + self.Wx(x), inplace=True))
        return x * out


class FastMambaBlock(nn.Module):
    def __init__(self, d):
        super().__init__()
        self.mamba = Mamba(
            d_model=d, d_state=16, d_conv=4, expand=2
        )
        self.norm = nn.LayerNorm(d)

    def _scan(self, x, flip=False):
        s = torch.flip(x, [1]) if flip else x
        o = self.mamba(self.norm(s))
        return torch.flip(o, [1]) if flip else o

    def forward(self, x):
        B, C, H, W = x.shape
        h = x.view(B, C, -1).transpose(1, 2)
        oh = (self._scan(h) + self._scan(h, flip=True)) / 2
        v = (
            x.permute(0, 1, 3, 2)
            .contiguous()
            .view(B, C, -1)
            .transpose(1, 2)
        )
        ov = (self._scan(v) + self._scan(v, flip=True)) / 2
        out = (oh + ov) / 2
        return out.transpose(1, 2).view(B, C, H, W) + x


# =====================================================================
# 5. ENCODER
# =====================================================================
class ResidualAdapter(nn.Module):
    def __init__(self, in_c, out_c):
        super().__init__()
        self.needs_proj = in_c != out_c
        if self.needs_proj:
            self.proj = nn.Conv2d(in_c, out_c, 1, bias=False)
            self.bn = nn.BatchNorm2d(out_c)
            nn.init.kaiming_normal_(
                self.proj.weight, mode="fan_out"
            )

    def forward(self, x):
        if not self.needs_proj:
            return x
        return self.bn(self.proj(x))


class EfficientNetEncoder(nn.Module):
    def __init__(self, pretrained=True):
        super().__init__()
        self.stages = timm.create_model(
            "efficientnet_b4",
            pretrained=pretrained,
            features_only=True,
        )
        self.channels = self.stages.feature_info.channels()
        print(f"  EfficientNet-B4 encoder channels: {self.channels}")

    def forward(self, x):
        return self.stages(x)


class ScratchEncoder(nn.Module):
    def __init__(self):
        super().__init__()
        F_ = [32, 64, 128, 256, 512]
        self.channels = F_
        self.s0 = ConvBlock(3, F_[0])
        self.s1 = nn.Sequential(
            nn.MaxPool2d(2), ConvBlock(F_[0], F_[1])
        )
        self.s2 = nn.Sequential(
            nn.MaxPool2d(2), ConvBlock(F_[1], F_[2])
        )
        self.s3 = nn.Sequential(
            nn.MaxPool2d(2), ConvBlock(F_[2], F_[3]), CBAM(F_[3])
        )
        self.s4 = nn.Sequential(
            nn.MaxPool2d(2), ConvBlock(F_[3], F_[4]), CBAM(F_[4])
        )

    def forward(self, x):
        f0 = self.s0(x)
        f1 = self.s1(f0)
        f2 = self.s2(f1)
        f3 = self.s3(f2)
        f4 = self.s4(f3)
        return [f0, f1, f2, f3, f4]


# =====================================================================
# 6. MAMBA UNET++ V4
# =====================================================================
class MambaUNetPPv4(nn.Module):
    def __init__(
        self, num_classes=1, deep_supervision=True, pretrained=True
    ):
        super().__init__()
        self.deep_supervision = deep_supervision

        ContextBlock = FastMambaBlock if USE_FAST_MAMBA else CBAM

        if HAS_TIMM:
            self.encoder = EfficientNetEncoder(
                pretrained=pretrained
            )
        else:
            self.encoder = ScratchEncoder()

        enc_ch = self.encoder.channels
        dec_ch = [64, 128, 192, 256, 256]

        self.adapt = nn.ModuleList(
            [
                ResidualAdapter(ec, dc)
                for ec, dc in zip(enc_ch, dec_ch)
            ]
        )

        self.aspp = ASPP(dec_ch[4], dec_ch[4])
        self.ctx3 = ContextBlock(dec_ch[3])
        self.ctx4 = ContextBlock(dec_ch[4])

        C = dec_ch
        self.up = nn.Upsample(
            scale_factor=2, mode="bilinear", align_corners=True
        )

        self.ag = nn.ModuleDict(
            {
                "01": AttentionGate(C[1], C[0], C[0] // 2),
                "11": AttentionGate(C[2], C[1], C[1] // 2),
                "21": AttentionGate(C[3], C[2], C[2] // 2),
                "31": AttentionGate(C[4], C[3], C[3] // 2),
                "02": AttentionGate(C[1], C[0], C[0] // 2),
                "12": AttentionGate(C[2], C[1], C[1] // 2),
                "22": AttentionGate(C[3], C[2], C[2] // 2),
                "03": AttentionGate(C[1], C[0], C[0] // 2),
                "13": AttentionGate(C[2], C[1], C[1] // 2),
                "04": AttentionGate(C[1], C[0], C[0] // 2),
            }
        )

        self.d01 = ConvBlock(C[0] + C[1], C[0])
        self.d11 = ConvBlock(C[1] + C[2], C[1])
        self.d21 = ConvBlock(C[2] + C[3], C[2])
        self.d31 = ConvBlock(C[3] + C[4], C[3])
        self.d02 = ConvBlock(C[0] * 2 + C[1], C[0])
        self.d12 = ConvBlock(C[1] * 2 + C[2], C[1])
        self.d22 = ConvBlock(C[2] * 2 + C[3], C[2])
        self.d03 = ConvBlock(C[0] * 3 + C[1], C[0])
        self.d13 = ConvBlock(C[1] * 3 + C[2], C[1])
        self.d04 = ConvBlock(C[0] * 4 + C[1], C[0])

        if deep_supervision:
            self.heads = nn.ModuleList(
                [
                    nn.Conv2d(C[0], num_classes, 1)
                    for _ in range(4)
                ]
            )
        else:
            self.head = nn.Conv2d(C[0], num_classes, 1)

        self._init_decoder_weights()

    def _init_decoder_weights(self):
        for m in self.modules():
            if any(m is s for s in self.encoder.modules()):
                continue
            if isinstance(m, nn.Conv2d):
                nn.init.kaiming_normal_(
                    m.weight, mode="fan_out", nonlinearity="relu"
                )
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.ones_(m.weight)
                nn.init.zeros_(m.bias)

    def forward(self, x):
        feats = self.encoder(x)
        f = [ad(fe) for ad, fe in zip(self.adapt, feats)]
        f[3] = self.ctx3(f[3])
        f[4] = self.ctx4(self.aspp(f[4]))

        x01 = self.d01(
            torch.cat(
                [self.ag["01"](f[1], f[0]), self.up(f[1])], 1
            )
        )
        x11 = self.d11(
            torch.cat(
                [self.ag["11"](f[2], f[1]), self.up(f[2])], 1
            )
        )
        x21 = self.d21(
            torch.cat(
                [self.ag["21"](f[3], f[2]), self.up(f[3])], 1
            )
        )
        x31 = self.d31(
            torch.cat(
                [self.ag["31"](f[4], f[3]), self.up(f[4])], 1
            )
        )

        x02 = self.d02(
            torch.cat(
                [f[0], self.ag["02"](x11, x01), self.up(x11)], 1
            )
        )
        x12 = self.d12(
            torch.cat(
                [f[1], self.ag["12"](x21, x11), self.up(x21)], 1
            )
        )
        x22 = self.d22(
            torch.cat(
                [f[2], self.ag["22"](x31, x21), self.up(x31)], 1
            )
        )

        x03 = self.d03(
            torch.cat(
                [
                    f[0],
                    x01,
                    self.ag["03"](x12, x02),
                    self.up(x12),
                ],
                1,
            )
        )
        x13 = self.d13(
            torch.cat(
                [
                    f[1],
                    x11,
                    self.ag["13"](x22, x12),
                    self.up(x22),
                ],
                1,
            )
        )
        x04 = self.d04(
            torch.cat(
                [
                    f[0],
                    x01,
                    x02,
                    self.ag["04"](x13, x03),
                    self.up(x13),
                ],
                1,
            )
        )

        H, W = x.shape[2:]
        if self.deep_supervision:
            return [
                F.interpolate(
                    h(o),
                    size=(H, W),
                    mode="bilinear",
                    align_corners=True,
                )
                for h, o in zip(
                    self.heads, [x01, x02, x03, x04]
                )
            ]
        out = self.head(x04)
        return F.interpolate(
            out, size=(H, W), mode="bilinear", align_corners=True
        )


# =====================================================================
# 7. EMA
# =====================================================================
class EMA:
    def __init__(self, model, decay=0.9995):
        self.decay = decay
        self.shadow = deepcopy(model).eval()
        for p in self.shadow.parameters():
            p.requires_grad_(False)

    @torch.no_grad()
    def update(self, model):
        for s, m in zip(
            self.shadow.parameters(), model.parameters()
        ):
            s.copy_(self.decay * s + (1 - self.decay) * m.data)
        for s, m in zip(
            self.shadow.buffers(), model.buffers()
        ):
            s.copy_(m)

    def __call__(self, *args, **kwargs):
        return self.shadow(*args, **kwargs)


# =====================================================================
# 8. TTA
# =====================================================================
@torch.no_grad()
def tta_predict(model, imgs, device):
    preds = []
    for fh in [False, True]:
        for fv in [False, True]:
            x = imgs.clone()
            if fh:
                x = torch.flip(x, [3])
            if fv:
                x = torch.flip(x, [2])
            with torch.amp.autocast("cuda"):
                out = model(x)
                p = torch.sigmoid(
                    out[-1] if isinstance(out, list) else out
                )
            if fv:
                p = torch.flip(p, [2])
            if fh:
                p = torch.flip(p, [3])
            preds.append(p)
    return torch.stack(preds).mean(0)


# =====================================================================
# 9. PLOTTING
# =====================================================================
def plot_all(history, path="training_metrics_v4.png"):
    eps = range(1, len(history["train_loss"]) + 1)
    fig, ax = plt.subplots(2, 3, figsize=(18, 10))
    fig.suptitle(
        "Mamba-UNet++ v4 Training Dashboard",
        fontsize=16,
        fontweight="bold",
    )

    def _p(a, key, label, color, target=None):
        a.plot(
            eps,
            history[key],
            color=color,
            marker="o",
            markersize=2,
            label=label,
        )
        if target:
            a.axhline(
                target,
                color="limegreen",
                ls="--",
                lw=1,
                alpha=0.7,
                label=f"Target {target}",
            )
        a.legend(fontsize=8)
        a.grid(True, alpha=0.3)

    _p(ax[0, 0], "train_loss", "Train Loss", "red")
    ax[0, 0].set_title("Loss")
    _p(ax[0, 1], "val_dice", "Val Dice", "royalblue", 0.92)
    ax[0, 1].set_title("Dice")
    ax[0, 1].set_ylim(0, 1)
    _p(ax[0, 2], "val_iou", "Val IoU", "seagreen")
    ax[0, 2].set_title("IoU")
    ax[0, 2].set_ylim(0, 1)

    ax[1, 0].plot(
        eps,
        history["val_precision"],
        color="purple",
        marker="o",
        markersize=2,
        label="Precision",
    )
    ax[1, 0].plot(
        eps,
        history["val_recall"],
        color="teal",
        marker="s",
        markersize=2,
        label="Recall",
    )
    ax[1, 0].legend(fontsize=8)
    ax[1, 0].grid(True, alpha=0.3)
    ax[1, 0].set_ylim(0, 1)
    ax[1, 0].set_title("Precision & Recall")

    _p(ax[1, 1], "val_hausdorff", "HD95 (px)", "darkorange")
    ax[1, 1].set_title("Hausdorff Distance 95th pct")

    ax[1, 2].plot(
        eps, history["lr"], "k", marker="o", markersize=2
    )
    ax[1, 2].set_title("Learning Rate")
    ax[1, 2].set_yscale("log")
    ax[1, 2].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig(path, dpi=150, bbox_inches="tight")
    plt.close()


# =====================================================================
# 10. INFERENCE VIS
# =====================================================================
def run_inference(
    model_path, image_dir, mask_dir, num_samples=4, device=None
):
    if device is None:
        device = torch.device(
            "cuda" if torch.cuda.is_available() else "cpu"
        )
    model = MambaUNetPPv4(
        deep_supervision=False, pretrained=False
    ).to(device)
    sd = torch.load(model_path, map_location=device)
    model.load_state_dict(sd, strict=False)
    model.eval()
    print(f"\n✓ Loaded checkpoint from '{model_path}'")

    ext = (".png", ".jpg", ".jpeg")
    imgs = sorted(
        [f for f in os.listdir(image_dir) if f.lower().endswith(ext)]
    )
    picks = random.sample(imgs, min(num_samples, len(imgs)))
    tf = get_val_transforms()

    fig, axes = plt.subplots(
        num_samples, 4, figsize=(17, 4 * num_samples)
    )
    if num_samples == 1:
        axes = axes[np.newaxis]
    fig.suptitle(
        "Inference — Mamba-UNet++ v4",
        fontsize=14,
        fontweight="bold",
    )
    for c, t in enumerate(
        [
            "Input",
            "Ground Truth",
            "Prediction",
            "Overlay (TP/FP/FN)",
        ]
    ):
        axes[0, c].set_title(t, fontsize=11, fontweight="bold")

    for row, name in enumerate(picks):
        ip = os.path.join(image_dir, name)
        mp = os.path.join(
            mask_dir, name.rsplit(".", 1)[0] + ".png"
        )
        if not os.path.exists(mp):
            mp = os.path.join(mask_dir, name)

        orig = cv2.cvtColor(
            cv2.imread(ip), cv2.COLOR_BGR2RGB
        )
        orig = cv2.resize(orig, (IMG_SIZE, IMG_SIZE))
        gt = (
            cv2.imread(mp, cv2.IMREAD_GRAYSCALE)
            if os.path.exists(mp)
            else np.zeros((IMG_SIZE, IMG_SIZE), np.uint8)
        )
        gt = cv2.resize(gt, (IMG_SIZE, IMG_SIZE))
        gt = (gt > 127).astype(np.float32)

        inp = (
            tf(image=orig, mask=gt)["image"]
            .unsqueeze(0)
            .to(device)
        )
        with torch.no_grad(), torch.amp.autocast("cuda"):
            out = model(inp)
            pred = (
                (torch.sigmoid(out) > 0.5)
                .float()
                .squeeze()
                .cpu()
                .numpy()
            )

        gtt = torch.from_numpy(gt).unsqueeze(0).unsqueeze(0)
        pt = torch.from_numpy(pred).unsqueeze(0).unsqueeze(0)
        dice = float(
            (2 * (pt * gtt).sum() + 1e-6)
            / (pt.sum() + gtt.sum() + 1e-6)
        )
        iou = float(
            ((pt * gtt).sum() + 1e-6)
            / ((pt + gtt - (pt * gtt)).sum() + 1e-6)
        )

        ov = orig.copy().astype(np.float32)
        tp = (pred > 0.5) & (gt > 0.5)
        fp = (pred > 0.5) & (gt <= 0.5)
        fn = (pred <= 0.5) & (gt > 0.5)
        ov[tp] = ov[tp] * 0.45 + np.array([0, 220, 0]) * 0.55
        ov[fp] = ov[fp] * 0.45 + np.array([220, 0, 0]) * 0.55
        ov[fn] = ov[fn] * 0.45 + np.array([0, 0, 220]) * 0.55

        axes[row, 0].imshow(orig)
        axes[row, 0].set_ylabel(name[:18], fontsize=7)
        axes[row, 1].imshow(gt, cmap="gray")
        axes[row, 2].imshow(pred, cmap="gray")
        axes[row, 2].set_xlabel(
            f"Dice={dice:.4f}  IoU={iou:.4f}", fontsize=8
        )
        axes[row, 3].imshow(
            np.clip(ov, 0, 255).astype(np.uint8)
        )
        for ax in axes[row]:
            ax.axis("off")

    patches = [
        mpatches.Patch(color="lime", label="TP"),
        mpatches.Patch(color="red", label="FP"),
        mpatches.Patch(color="blue", label="FN"),
    ]
    fig.legend(
        handles=patches,
        loc="lower center",
        ncol=3,
        fontsize=10,
        bbox_to_anchor=(0.5, -0.005),
    )
    plt.tight_layout()
    plt.savefig(
        "inference_results_v4.png", dpi=150, bbox_inches="tight"
    )
    plt.close()
    print("✓ Inference grid saved to 'inference_results_v4.png'")


# =====================================================================
# 11. TRAINING
# =====================================================================
def train():
    device = torch.device(
        "cuda" if torch.cuda.is_available() else "cpu"
    )
    print(f"\nDevice: {device}")
    if torch.cuda.is_available():
        print(f"GPU   : {torch.cuda.get_device_name(0)}")

    base = "dataset/train"
    img_dir = os.path.join(base, "images")
    msk_dir = os.path.join(base, "labels")
    if not os.path.exists(img_dir):
        img_dir = msk_dir = base

    ext = (".png", ".jpg", ".jpeg")
    all_ = sorted(
        [
            f
            for f in os.listdir(img_dir)
            if f.lower().endswith(ext)
        ]
    )
    if not all_:
        raise ValueError(f"No images in {img_dir}")

    random.shuffle(all_)
    split = int(0.8 * len(all_))
    tr_names = all_[:split]
    vl_names = all_[split:]
    print(f"Train : {len(tr_names)} | Val: {len(vl_names)}\n")

    tr_ds = DFUDataset(
        img_dir, msk_dir, tr_names, get_train_transforms()
    )
    vl_ds = DFUDataset(
        img_dir, msk_dir, vl_names, get_val_transforms()
    )

    # ─── FIX: num_workers=0 to avoid thread exhaustion ────────────
    BATCH = 16
    tr_ld = DataLoader(
        tr_ds,
        batch_size=BATCH,
        shuffle=True,
        num_workers=0,
        pin_memory=True,
    )
    vl_ld = DataLoader(
        vl_ds,
        batch_size=8,
        shuffle=False,
        num_workers=0,
        pin_memory=True,
    )

    model = MambaUNetPPv4(
        deep_supervision=True, pretrained=HAS_TIMM
    ).to(device)

    # ─── FIX: skip torch.compile (causes pthread exhaustion) ──────
    print(
        "⏭  Skipping torch.compile (avoids pthread exhaustion "
        "in constrained environments)"
    )

    ema = EMA(model, decay=0.9995)
    crit = ScheduledLoss()

    # Differential LR: encoder 10× slower than decoder
    enc_params = list(model.encoder.parameters())
    enc_param_ids = {id(p) for p in enc_params}
    dec_params = [
        p
        for p in model.parameters()
        if id(p) not in enc_param_ids
    ]

    optimizer = torch.optim.AdamW(
        [
            {"params": enc_params, "lr": 3e-5, "name": "encoder"},
            {"params": dec_params, "lr": 2e-4, "name": "decoder"},
        ],
        weight_decay=1e-4,
        betas=(0.9, 0.999),
    )

    epochs = 150
    sched = torch.optim.lr_scheduler.OneCycleLR(
        optimizer,
        max_lr=[3e-4, 2e-3],
        epochs=epochs,
        steps_per_epoch=len(tr_ld),
        pct_start=0.05,
        anneal_strategy="cos",
        div_factor=10,
        final_div_factor=100,
    )

    scaler = torch.amp.GradScaler("cuda")

    ds_w = [0.05, 0.15, 0.30, 0.50]
    ACCUM = 2

    best_dice = 0.0
    patience = 0
    PATIENCE = 30
    history = {
        k: []
        for k in [
            "train_loss",
            "val_dice",
            "val_iou",
            "val_precision",
            "val_recall",
            "val_hausdorff",
            "lr",
        ]
    }

    print(
        f"Training: {epochs} epochs | OneCycleLR | AdamW | EMA\n"
    )

    for epoch in range(epochs):
        model.train()
        crit.set_epoch(epoch)
        tloss = 0.0
        n_valid = 0
        optimizer.zero_grad(set_to_none=True)

        pbar = tqdm(
            enumerate(tr_ld),
            total=len(tr_ld),
            desc=f"E{epoch+1:03d}/{epochs} [Train]",
            ncols=110,
        )

        for step, (imgs, masks) in pbar:
            imgs = imgs.to(device, non_blocking=True)
            masks = masks.to(device, non_blocking=True)

            with torch.amp.autocast("cuda"):
                seg_outs = model(imgs)
                if isinstance(seg_outs, tuple):
                    seg_outs = seg_outs[0]
                if not isinstance(seg_outs, list):
                    seg_outs = [seg_outs]

                loss = (
                    sum(
                        w * crit(o, masks)
                        for w, o in zip(
                            ds_w[-len(seg_outs) :], seg_outs
                        )
                    )
                    / ACCUM
                )

            if not torch.isfinite(loss):
                optimizer.zero_grad(set_to_none=True)
                pbar.set_postfix({"loss": "nan-skip"})
                continue

            scaler.scale(loss).backward()
            n_valid += 1

            if (step + 1) % ACCUM == 0 or (step + 1) == len(
                tr_ld
            ):
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(
                    model.parameters(), 1.0
                )
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad(set_to_none=True)
                ema.update(model)

            sched.step()

            loss_val = loss.item() * ACCUM
            tloss += loss_val
            pbar.set_postfix({"loss": f"{loss_val:.4f}"})

        epoch_loss = tloss / max(n_valid, 1)

        # ── Validate with EMA ──────────────────────────────────────
        ema.shadow.eval()
        d_l, i_l, p_l, r_l, h_l = [], [], [], [], []
        use_tta = (epoch + 1) >= 80 and (epoch + 1) % 5 == 0

        with torch.no_grad():
            for imgs, masks in tqdm(
                vl_ld,
                desc=f"E{epoch+1:03d}/{epochs} [Val]  ",
                leave=False,
                ncols=110,
            ):
                imgs = imgs.to(device, non_blocking=True)
                masks = masks.to(device, non_blocking=True)

                if use_tta:
                    prob = tta_predict(
                        ema.shadow, imgs, device
                    )
                    final_out = torch.log(
                        prob / (1 - prob + 1e-8) + 1e-8
                    )
                else:
                    with torch.amp.autocast("cuda"):
                        out = ema.shadow(imgs)
                        if isinstance(out, tuple):
                            out = out[0]
                        final_out = (
                            out[-1]
                            if isinstance(out, list)
                            else out
                        )

                d, i, p, r = calc_metrics(final_out, masks)
                h = hausdorff_95(final_out, masks)
                d_l.append(d)
                i_l.append(i)
                p_l.append(p)
                r_l.append(r)
                h_l.append(h)

        vd, vi, vp, vr, vh = (
            np.mean(d_l),
            np.mean(i_l),
            np.mean(p_l),
            np.mean(r_l),
            np.mean(h_l),
        )
        cur_lr = optimizer.param_groups[0]["lr"]

        history["train_loss"].append(epoch_loss)
        history["val_dice"].append(vd)
        history["val_iou"].append(vi)
        history["val_precision"].append(vp)
        history["val_recall"].append(vr)
        history["val_hausdorff"].append(vh)
        history["lr"].append(cur_lr)

        tta_tag = " [TTA]" if use_tta else ""
        print(
            f"  Loss:{epoch_loss:.4f} | Dice:{vd:.4f} | "
            f"IoU:{vi:.4f} | Prec:{vp:.4f} | Rec:{vr:.4f} | "
            f"HD95:{vh:.1f}px | LR:{cur_lr:.1e}{tta_tag}"
        )

        if vd > best_dice:
            best_dice = vd
            torch.save(
                ema.shadow.state_dict(),
                "best_mamba_unetpp_v4.pth",
            )
            print(
                f"  ✓ Best model saved! (Dice={best_dice:.4f})"
            )
            patience = 0
        else:
            patience += 1
            if patience >= PATIENCE:
                print(
                    f"\nEarly stop at epoch {epoch+1}. "
                    f"Best Dice={best_dice:.4f}"
                )
                break

        if (epoch + 1) % 5 == 0:
            plot_all(history)

    print(f"\n{'='*65}")
    print(f" Training done | Best EMA Dice: {best_dice:.4f}")
    print(f"{'='*65}")

    run_inference(
        "best_mamba_unetpp_v4.pth",
        img_dir,
        msk_dir,
        num_samples=4,
        device=device,
    )


if __name__ == "__main__":
    train()
```

```python
"""
Mamba-UNet++ v5 — DFUC2022 / Fundus Edition
Optimized for NVIDIA A40 (High VRAM, PyTorch 2.x Compiled)
"""

import os
# Thread guards MUST come before any cv2/numpy import
os.environ["OMP_NUM_THREADS"] = "1"
os.environ["MKL_NUM_THREADS"] = "1"
os.environ["OPENBLAS_NUM_THREADS"] = "1"
os.environ["NUMEXPR_NUM_THREADS"] = "1"

import math, random, warnings
import cv2

cv2.setNumThreads(0)
cv2.ocl.setUseOpenCL(False)

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from scipy.ndimage import distance_transform_edt
from copy import deepcopy

warnings.filterwarnings("ignore")
torch.backends.cudnn.benchmark = True
torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True

try:
    from mamba_ssm import Mamba
    USE_FAST_MAMBA = True
    print("✓ Fast CUDA Mamba loaded")
except ImportError:
    USE_FAST_MAMBA = False
    print("⚠  mamba-ssm not found — using CBAM attention (fast fallback)")

try:
    import timm
    HAS_TIMM = True
    print("✓ timm loaded")
except ImportError:
    HAS_TIMM = False
    print("⚠  timm not found")

# ── DFUC2022 / Scaled Constants ──────────────────────────────────────
TRAIN_PATCH_SIZE = 512   # Smart crop size for DFUC2022
PAD_RESIZE = 640         # Pad-to-square size for Val/Inference
BATCH_SIZE = 16          # 16 is safe for 512x512 + Mamba on 48GB A40
NUM_WORKERS = 8          

def set_seed(s=42):
    random.seed(s)
    np.random.seed(s)
    torch.manual_seed(s)
    torch.cuda.manual_seed_all(s)

set_seed()

image_dir = "dataset/train/images"
mask_dir  = "dataset/train/labels"

# =====================================================================
# 1. DATASET & ALBUMENTATIONS
# =====================================================================
class MedicalDataset(Dataset):
    def __init__(self, samples, transform=None):
        self.transform = transform
        self.samples = samples  # List of (img_path, mask_path)

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, mask_path = self.samples[idx]

        im = cv2.imread(img_path)
        im = cv2.cvtColor(im, cv2.COLOR_BGR2RGB) if im is not None else np.zeros((PAD_RESIZE, PAD_RESIZE, 3), np.uint8)
        
        mk = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE) if mask_path else np.zeros(im.shape[:2], np.uint8)
        mk = (mk > 127).astype(np.float32)

        if self.transform:
            aug = self.transform(image=im, mask=mk)
            im, mk = aug["image"], aug["mask"].unsqueeze(0)
            
        return im, mk

def get_train_transforms():
    return A.Compose([
        A.PadIfNeeded(min_height=TRAIN_PATCH_SIZE, min_width=TRAIN_PATCH_SIZE, 
                      border_mode=cv2.BORDER_CONSTANT, value=0, mask_value=0),
        A.CropNonEmptyMaskIfExists(height=TRAIN_PATCH_SIZE, width=TRAIN_PATCH_SIZE, p=1.0),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.RandomRotate90(p=0.5),
        A.Affine(scale=(0.85, 1.15), translate_percent=(-0.08, 0.08), rotate=(-30, 30), p=0.5),
        A.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1, p=0.4),
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ToTensorV2(),
    ])

def get_val_transforms():
    return A.Compose([
        A.LongestMaxSize(max_size=PAD_RESIZE),
        A.PadIfNeeded(min_height=PAD_RESIZE, min_width=PAD_RESIZE, 
                      border_mode=cv2.BORDER_CONSTANT, value=0, mask_value=0),
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ToTensorV2(),
    ])

# =====================================================================
# 2. METRICS, LOSSES, & PLOTTING
# =====================================================================
def calc_metrics(pred_logits, target, thr=0.5):
    pred = (torch.sigmoid(pred_logits) > thr).float().cpu().numpy()
    tgt = target.float().cpu().numpy()
    dice_l, iou_l, prec_l, rec_l = [], [], [], []
    for p, t in zip(pred, tgt):
        p, t = p.squeeze(), t.squeeze()
        tp = float((p * t).sum())
        fp = float((p * (1 - t)).sum())
        fn = float(((1 - p) * t).sum())
        dice_l.append((2 * tp + 1e-6) / (2 * tp + fp + fn + 1e-6))
        iou_l.append((tp + 1e-6) / (tp + fp + fn + 1e-6))
        prec_l.append((tp + 1e-6) / (tp + fp + 1e-6))
        rec_l.append((tp + 1e-6) / (tp + fn + 1e-6))
    return map(np.mean, [dice_l, iou_l, prec_l, rec_l])

def hausdorff_95(pred_logits, target, thr=0.5):
    """Expensive CPU metric — only called at final inference to save time."""
    pred = (torch.sigmoid(pred_logits) > thr).float().cpu().numpy()
    tgt = target.float().cpu().numpy()
    hd_l = []
    for p, t in zip(pred, tgt):
        p, t = p.squeeze().astype(bool), t.squeeze().astype(bool)
        if not p.any() or not t.any():
            hd_l.append(float(np.sqrt(p.shape[0] ** 2 + p.shape[1] ** 2)))
            continue
        d1 = distance_transform_edt(~p)[t]
        d2 = distance_transform_edt(~t)[p]
        hd_l.append(np.percentile(np.concatenate([d1, d2]), 95))
    return np.mean(hd_l)

def plot_all(history, path="training_metrics.png"):
    eps = range(1, len(history["train_loss"]) + 1)
    fig, ax = plt.subplots(2, 3, figsize=(18, 10))
    fig.suptitle("Training Dashboard", fontsize=16, fontweight="bold")

    def _p(a, key, label, color):
        a.plot(eps, history[key], color=color, marker="o", markersize=2, label=label)
        a.legend(fontsize=8)
        a.grid(True, alpha=0.3)

    _p(ax[0, 0], "train_loss", "Train Loss", "red")
    ax[0, 0].set_title("Loss")
    _p(ax[0, 1], "val_dice", "Val Dice", "royalblue")
    ax[0, 1].set_title("Dice")
    _p(ax[0, 2], "val_iou", "Val IoU", "seagreen")
    ax[0, 2].set_title("IoU")

    ax[1, 0].plot(eps, history["val_precision"], color="purple", marker="o", markersize=2, label="Precision")
    ax[1, 0].plot(eps, history["val_recall"], color="teal", marker="s", markersize=2, label="Recall")
    ax[1, 0].legend(fontsize=8)
    ax[1, 0].grid(True, alpha=0.3)
    ax[1, 0].set_title("Precision & Recall")

    ax[1, 1].axis("off")
    ax[1, 1].text(0.5, 0.5, "HD95 computed\nonly at inference\n(removed from epoch loop)", 
                  ha="center", va="center", fontsize=12, style="italic", color="gray")

    ax[1, 2].plot(eps, history["lr"], "k", marker="o", markersize=2)
    ax[1, 2].set_title("Learning Rate")
    ax[1, 2].set_yscale("log")
    ax[1, 2].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig(path, dpi=150, bbox_inches="tight")
    plt.close()

class DiceBCELoss(nn.Module):
    def __init__(self, smooth=1.0):
        super().__init__()
        self.smooth = smooth
        self.bce = nn.BCEWithLogitsLoss()

    def forward(self, logits, targets):
        bce = self.bce(logits.float(), targets.float())
        p, t = torch.sigmoid(logits).view(-1), targets.view(-1)
        dice = 1 - (2 * (p * t).sum() + self.smooth) / (p.sum() + t.sum() + self.smooth)
        return bce, dice

class LovaszHingeLoss(nn.Module):
    def forward(self, logits, targets):
        probs, tgts = torch.sigmoid(logits.float()).view(-1), targets.float().view(-1)
        if tgts.sum() == 0 or tgts.sum() == tgts.numel(): return (logits * 0.0).mean()
        errors = (tgts - probs).abs()
        errors_sorted, perm = torch.sort(errors, descending=True)
        return torch.clamp(torch.dot(errors_sorted, self._lovasz_grad(tgts[perm])), 0.0, 2.0)

    @staticmethod
    def _lovasz_grad(gt_sorted):
        gts = gt_sorted.sum()
        intersection = gts - gt_sorted.cumsum(0)
        union = gts + (1 - gt_sorted).cumsum(0)
        jaccard = 1.0 - intersection / (union + 1e-6)
        if gt_sorted.numel() > 1: jaccard[1:] = jaccard[1:] - jaccard[:-1]
        return jaccard

class TverskyLoss(nn.Module):
    def __init__(self, alpha=0.7, beta=0.3, gamma=0.75, smooth=1e-3):
        super().__init__()
        self.alpha, self.beta, self.gamma, self.smooth = alpha, beta, gamma, smooth

    def forward(self, inputs, targets):
        p, t = torch.sigmoid(inputs.float()).view(-1), targets.float().view(-1)
        tp, fp, fn = (p * t).sum(), (p * (1 - t)).sum(), ((1 - p) * t).sum()
        return (1 - ((tp + self.smooth) / (tp + self.alpha * fp + self.beta * fn + self.smooth))) ** self.gamma

class ScheduledLoss(nn.Module):
    def __init__(self):
        super().__init__()
        self.dice_bce = DiceBCELoss()
        self.tversky = TverskyLoss(alpha=0.7, beta=0.3, gamma=0.75)
        self.lovasz = LovaszHingeLoss()
        self._epoch = 0

    def set_epoch(self, e): self._epoch = e

    def forward(self, logits, targets):
        bce, dice = self.dice_bce(logits, targets)
        tv, lov = self.tversky(logits, targets), self.lovasz(logits, targets)
        safe = lambda t: t if torch.isfinite(t) else torch.zeros(1, device=logits.device)[0]

        e = self._epoch
        w = (0.25, 0.25, 0.40, 0.10) if e < 20 else (0.20, 0.30, 0.30, 0.20) if e < 50 else (0.15, 0.30, 0.25, 0.30) if e < 90 else (0.10, 0.30, 0.20, 0.40)
        return w[0] * safe(bce) + w[1] * safe(dice) + w[2] * safe(tv) + w[3] * safe(lov)

# =====================================================================
# 3. BUILDING BLOCKS & ARCHITECTURE
# =====================================================================
class SEBlock(nn.Module):
    def __init__(self, c, r=8):
        super().__init__()
        self.se = nn.Sequential(nn.AdaptiveAvgPool2d(1), nn.Flatten(), nn.Linear(c, c // r, bias=False), nn.ReLU(inplace=True), nn.Linear(c // r, c, bias=False), nn.Sigmoid())
    def forward(self, x): return x * self.se(x).view(x.size(0), x.size(1), 1, 1)

class CBAM(nn.Module):
    def __init__(self, c, r=8):
        super().__init__()
        self.mlp = nn.Sequential(nn.Linear(c, c // r, bias=False), nn.ReLU(inplace=True), nn.Linear(c // r, c, bias=False))
        self.spatial = nn.Sequential(nn.Conv2d(2, 1, 7, padding=3, bias=False), nn.BatchNorm2d(1), nn.Sigmoid())
    def forward(self, x):
        ca = torch.sigmoid(self.mlp(x.mean(dim=(2, 3))) + self.mlp(x.amax(dim=(2, 3)))).unsqueeze(-1).unsqueeze(-1)
        x = x * ca
        return x * self.spatial(torch.cat([x.mean(dim=1, keepdim=True), x.amax(dim=1, keepdim=True)], dim=1))

class ConvBlock(nn.Module):
    def __init__(self, in_c, out_c, drop=0.1):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_c, out_c, 3, padding=1, bias=False), nn.BatchNorm2d(out_c), nn.GELU(), nn.Dropout2d(drop),
            nn.Conv2d(out_c, out_c, 3, padding=1, bias=False), nn.BatchNorm2d(out_c), nn.GELU(),
        )
        self.se = SEBlock(out_c)
        self.proj = nn.Conv2d(in_c, out_c, 1, bias=False) if in_c != out_c else nn.Identity()
    def forward(self, x): return self.se(self.conv(x)) + self.proj(x)

class ASPP(nn.Module):
    def __init__(self, in_c, out_c=256):
        super().__init__()
        rates = [1, 6, 12, 18]
        self.convs = nn.ModuleList([
            nn.Sequential(nn.Conv2d(in_c, out_c, 3 if r > 1 else 1, padding=r if r > 1 else 0, dilation=r, bias=False), nn.BatchNorm2d(out_c), nn.ReLU(inplace=True)) for r in rates
        ])
        self.pool = nn.Sequential(nn.AdaptiveAvgPool2d(1), nn.Conv2d(in_c, out_c, 1, bias=False), nn.BatchNorm2d(out_c), nn.ReLU(inplace=True))
        self.proj = nn.Sequential(nn.Conv2d(out_c * (len(rates) + 1), out_c, 1, bias=False), nn.BatchNorm2d(out_c), nn.ReLU(inplace=True), nn.Dropout2d(0.1))
    def forward(self, x):
        feats = [c(x) for c in self.convs] + [F.interpolate(self.pool(x), size=x.shape[2:], mode="bilinear", align_corners=True)]
        return self.proj(torch.cat(feats, 1))

class AttentionGate(nn.Module):
    def __init__(self, Fg, Fl, Fi):
        super().__init__()
        self.Wg = nn.Sequential(nn.Conv2d(Fg, Fi, 1, bias=False), nn.BatchNorm2d(Fi))
        self.Wx = nn.Sequential(nn.Conv2d(Fl, Fi, 1, bias=False), nn.BatchNorm2d(Fi))
        self.psi = nn.Sequential(nn.Conv2d(Fi, 1, 1, bias=False), nn.BatchNorm2d(1), nn.Sigmoid())
    def forward(self, g, x):
        g1 = self.Wg(F.interpolate(g, x.shape[2:], mode="bilinear", align_corners=True))
        return x * self.psi(F.relu(g1 + self.Wx(x), inplace=True))

class FastMambaBlock(nn.Module):
    def __init__(self, d):
        super().__init__()
        self.mamba = Mamba(d_model=d, d_state=16, d_conv=4, expand=2)
        self.norm = nn.LayerNorm(d)
    def _scan(self, x, flip=False):
        s = torch.flip(x, [1]) if flip else x
        o = self.mamba(self.norm(s))
        return torch.flip(o, [1]) if flip else o
    def forward(self, x):
        B, C, H, W = x.shape
        h = x.view(B, C, -1).transpose(1, 2)
        oh = (self._scan(h) + self._scan(h, flip=True)) / 2
        v = x.permute(0, 1, 3, 2).contiguous().view(B, C, -1).transpose(1, 2)
        ov = (self._scan(v) + self._scan(v, flip=True)) / 2
        return ((oh + ov) / 2).transpose(1, 2).view(B, C, H, W) + x

class ResidualAdapter(nn.Module):
    def __init__(self, in_c, out_c):
        super().__init__()
        self.needs_proj = in_c != out_c
        if self.needs_proj:
            self.proj, self.bn = nn.Conv2d(in_c, out_c, 1, bias=False), nn.BatchNorm2d(out_c)
            nn.init.kaiming_normal_(self.proj.weight, mode="fan_out")
    def forward(self, x): return self.bn(self.proj(x)) if self.needs_proj else x

class EfficientNetEncoder(nn.Module):
    def __init__(self, pretrained=True):
        super().__init__()
        self.stages = timm.create_model("efficientnet_b4", pretrained=pretrained, features_only=True)
        self.channels = self.stages.feature_info.channels()
    def forward(self, x): return self.stages(x)

class ScratchEncoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.channels = [32, 64, 128, 256, 512]
        self.s0 = ConvBlock(3, self.channels[0])
        self.s1 = nn.Sequential(nn.MaxPool2d(2), ConvBlock(self.channels[0], self.channels[1]))
        self.s2 = nn.Sequential(nn.MaxPool2d(2), ConvBlock(self.channels[1], self.channels[2]))
        self.s3 = nn.Sequential(nn.MaxPool2d(2), ConvBlock(self.channels[2], self.channels[3]), CBAM(self.channels[3]))
        self.s4 = nn.Sequential(nn.MaxPool2d(2), ConvBlock(self.channels[3], self.channels[4]), CBAM(self.channels[4]))
    def forward(self, x):
        f0 = self.s0(x); f1 = self.s1(f0); f2 = self.s2(f1); f3 = self.s3(f2); f4 = self.s4(f3)
        return [f0, f1, f2, f3, f4]

class MambaUNetPPv4(nn.Module):
    def __init__(self, num_classes=1, deep_supervision=True, pretrained=True):
        super().__init__()
        self.deep_supervision = deep_supervision
        ContextBlock = FastMambaBlock if USE_FAST_MAMBA else CBAM

        self.encoder = EfficientNetEncoder(pretrained=pretrained) if HAS_TIMM else ScratchEncoder()
        enc_ch, dec_ch = self.encoder.channels, [64, 128, 192, 256, 256]

        self.adapt = nn.ModuleList([ResidualAdapter(ec, dc) for ec, dc in zip(enc_ch, dec_ch)])
        self.aspp, self.ctx3, self.ctx4 = ASPP(dec_ch[4], dec_ch[4]), ContextBlock(dec_ch[3]), ContextBlock(dec_ch[4])
        self.up = nn.Upsample(scale_factor=2, mode="bilinear", align_corners=True)

        C = dec_ch
        self.ag = nn.ModuleDict({
            f"{i}{j}": AttentionGate(C[i+1], C[i], C[i] // 2) for j in range(1, 5) for i in range(5-j)
        })

        self.d01, self.d11, self.d21, self.d31 = ConvBlock(C[0]+C[1], C[0]), ConvBlock(C[1]+C[2], C[1]), ConvBlock(C[2]+C[3], C[2]), ConvBlock(C[3]+C[4], C[3])
        self.d02, self.d12, self.d22 = ConvBlock(C[0]*2+C[1], C[0]), ConvBlock(C[1]*2+C[2], C[1]), ConvBlock(C[2]*2+C[3], C[2])
        self.d03, self.d13 = ConvBlock(C[0]*3+C[1], C[0]), ConvBlock(C[1]*3+C[2], C[1])
        self.d04 = ConvBlock(C[0]*4+C[1], C[0])

        self.heads = nn.ModuleList([nn.Conv2d(C[0], num_classes, 1) for _ in range(4)]) if deep_supervision else nn.Conv2d(C[0], num_classes, 1)

    def forward(self, x):
        f = [ad(fe) for ad, fe in zip(self.adapt, self.encoder(x))]
        f[3], f[4] = self.ctx3(f[3]), self.ctx4(self.aspp(f[4]))

        x01 = self.d01(torch.cat([self.ag["01"](f[1], f[0]), self.up(f[1])], 1))
        x11 = self.d11(torch.cat([self.ag["11"](f[2], f[1]), self.up(f[2])], 1))
        x21 = self.d21(torch.cat([self.ag["21"](f[3], f[2]), self.up(f[3])], 1))
        x31 = self.d31(torch.cat([self.ag["31"](f[4], f[3]), self.up(f[4])], 1))

        x02 = self.d02(torch.cat([f[0], self.ag["02"](x11, x01), self.up(x11)], 1))
        x12 = self.d12(torch.cat([f[1], self.ag["12"](x21, x11), self.up(x21)], 1))
        x22 = self.d22(torch.cat([f[2], self.ag["22"](x31, x21), self.up(x31)], 1))

        x03 = self.d03(torch.cat([f[0], x01, self.ag["03"](x12, x02), self.up(x12)], 1))
        x13 = self.d13(torch.cat([f[1], x11, self.ag["13"](x22, x12), self.up(x22)], 1))
        x04 = self.d04(torch.cat([f[0], x01, x02, self.ag["04"](x13, x03), self.up(x13)], 1))

        H, W = x.shape[2:]
        if self.deep_supervision:
            return [F.interpolate(h(o), size=(H, W), mode="bilinear", align_corners=True) for h, o in zip(self.heads, [x01, x02, x03, x04])]
        return F.interpolate(self.heads(x04), size=(H, W), mode="bilinear", align_corners=True)

# =====================================================================
# 4. EMA & TTA
# =====================================================================
class EMA:
    def __init__(self, model, decay=0.9995):
        self.decay = decay
        self.shadow = deepcopy(model).eval()
        for p in self.shadow.parameters(): p.requires_grad_(False)
    @torch.no_grad()
    def update(self, model):
        for s, m in zip(self.shadow.parameters(), model.parameters()): s.copy_(self.decay * s + (1 - self.decay) * m.data)
        for s, m in zip(self.shadow.buffers(), model.buffers()): s.copy_(m)
    def __call__(self, *args, **kwargs): return self.shadow(*args, **kwargs)

@torch.no_grad()
def tta_predict(model, imgs):
    logit_sum = torch.zeros_like(imgs[:, :1, :, :])
    for fh in [False, True]:
        for fv in [False, True]:
            x = imgs.clone()
            if fh: x = torch.flip(x, [3])
            if fv: x = torch.flip(x, [2])
            with torch.amp.autocast("cuda"):
                out = model(x)
                logits = out[-1] if isinstance(out, list) else out
            if fv: logits = torch.flip(logits, [2])
            if fh: logits = torch.flip(logits, [3])
            logit_sum += logits.float()
    return logit_sum / 4.0

# =====================================================================
# 5. FAST INFERENCE (Val Subset Only)
# =====================================================================
def run_inference(model_path, val_samples, num_samples=5, device=None):
    if device is None: device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    
    model = MambaUNetPPv4(deep_supervision=False, pretrained=False).to(device)
    model.load_state_dict(torch.load(model_path, map_location=device), strict=False)
    model.eval()

    # ONLY pick from the validation subset provided by the epoch loop
    picks = random.sample(val_samples, min(num_samples, len(val_samples)))
    val_transform = get_val_transforms()

    fig, axes = plt.subplots(num_samples, 4, figsize=(17, 4 * num_samples))
    if num_samples == 1: axes = axes[np.newaxis]
    for c, t in enumerate(["Input", "Ground Truth", "Prediction", "Overlay"]):
        axes[0, c].set_title(t, fontsize=11, fontweight="bold")

    all_hd95 = []

    for row, (ip, mp) in enumerate(picks):
        name = os.path.basename(ip)
        orig = cv2.cvtColor(cv2.imread(ip), cv2.COLOR_BGR2RGB)
        gt_raw = cv2.imread(mp, cv2.IMREAD_GRAYSCALE) if mp and os.path.exists(mp) else np.zeros(orig.shape[:2], np.uint8)
        gt = (gt_raw > 127).astype(np.float32)

        aug = val_transform(image=orig, mask=gt)
        img_t = aug["image"].unsqueeze(0).to(device)
        gt_np = aug["mask"].numpy()

        with torch.amp.autocast("cuda"):
            logit_map = tta_predict(model, img_t).float().cpu().numpy()[0, 0]
            
        pred = (1 / (1 + np.exp(-logit_map)) > 0.5).astype(np.float32)

        gtt, pt = torch.from_numpy(gt_np).unsqueeze(0).unsqueeze(0), torch.from_numpy(pred).unsqueeze(0).unsqueeze(0)
        dice = float((2 * (pt * gtt).sum() + 1e-6) / (pt.sum() + gtt.sum() + 1e-6))
        
        # Calculate HD95 here (keeps it out of the main loop)
        hd = hausdorff_95(torch.from_numpy(logit_map).unsqueeze(0).unsqueeze(0), gtt)
        all_hd95.append(hd)

        img_disp = aug["image"].permute(1, 2, 0).numpy()
        img_disp = ((img_disp * np.array([0.229, 0.224, 0.225])) + np.array([0.485, 0.456, 0.406])) * 255.0

        ov = img_disp.copy().astype(np.float32)
        tp, fp, fn = (pred > 0.5) & (gt_np > 0.5), (pred > 0.5) & (gt_np <= 0.5), (pred <= 0.5) & (gt_np > 0.5)
        ov[tp] = ov[tp] * 0.45 + np.array([0, 220, 0]) * 0.55
        ov[fp] = ov[fp] * 0.45 + np.array([220, 0, 0]) * 0.55
        ov[fn] = ov[fn] * 0.45 + np.array([0, 0, 220]) * 0.55

        axes[row, 0].imshow(np.clip(img_disp, 0, 255).astype(np.uint8))
        axes[row, 0].set_ylabel(name[:18], fontsize=7)
        axes[row, 1].imshow(gt_np, cmap="gray")
        axes[row, 2].imshow(pred, cmap="gray")
        axes[row, 2].set_xlabel(f"Dice={dice:.4f}  HD95={hd:.1f}px", fontsize=8)
        axes[row, 3].imshow(np.clip(ov, 0, 255).astype(np.uint8))
        for ax in axes[row]: ax.axis("off")

    plt.tight_layout()
    plt.savefig("inference_results_v5.png", dpi=150, bbox_inches="tight")
    plt.close()
    print(f"\n✓ Inference saved on {num_samples} validation images. Mean HD95: {np.mean(all_hd95):.1f} px")

# =====================================================================
# 6. FAST TRAINING LOOP
# =====================================================================
def train():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"\nDevice: {device} | GPU: {torch.cuda.get_device_name(0)}")

    base = "dataset/train"
    img_dir, msk_dir = os.path.join(base, "images"), os.path.join(base, "labels")
    if not os.path.exists(img_dir): img_dir = msk_dir = base

    all_names = sorted([f for f in os.listdir(img_dir) if f.lower().endswith((".png", ".jpg", ".jpeg"))])
    
    samples = []
    for n in all_names:
        ip = os.path.join(img_dir, n)
        stem = n.rsplit(".", 1)[0]
        mp_png = os.path.join(msk_dir, stem + ".png")
        mp_jpg = os.path.join(msk_dir, n)
        mp = mp_png if os.path.exists(mp_png) else (mp_jpg if os.path.exists(mp_jpg) else None)
        if mp: samples.append((ip, mp))

    random.shuffle(samples)
    split = int(0.8 * len(samples))
    tr_samples, vl_samples = samples[:split], samples[split:]

    tr_ds = MedicalDataset(tr_samples, transform=get_train_transforms())
    vl_ds = MedicalDataset(vl_samples, transform=get_val_transforms())

    tr_ld = DataLoader(tr_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True, drop_last=True)
    vl_ld = DataLoader(vl_ds, batch_size=BATCH_SIZE//2, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

    model = MambaUNetPPv4(deep_supervision=True, pretrained=HAS_TIMM).to(device)
    
    if hasattr(torch, "compile"):
        print("⚡ Compiling model with torch.compile() for A40 speed boost...")
        model = torch.compile(model, mode="reduce-overhead")

    ema = EMA(model, decay=0.9995)
    crit = ScheduledLoss()

    enc_params = list(model.encoder.parameters() if not hasattr(model, "_orig_mod") else model._orig_mod.encoder.parameters())
    enc_param_ids = {id(p) for p in enc_params}
    dec_params = [p for p in model.parameters() if id(p) not in enc_param_ids]

    optimizer = torch.optim.AdamW([
        {"params": enc_params, "lr": 5e-5, "name": "encoder"},
        {"params": dec_params, "lr": 3e-4, "name": "decoder"},
    ], weight_decay=1e-4, betas=(0.9, 0.999))

    epochs = 150
    sched = torch.optim.lr_scheduler.OneCycleLR(
        optimizer, max_lr=[5e-4, 3e-3], epochs=epochs, steps_per_epoch=len(tr_ld),
        pct_start=0.05, anneal_strategy="cos", div_factor=10, final_div_factor=100,
    )

    scaler = torch.amp.GradScaler("cuda")
    best_dice = 0.0
    patience, PATIENCE = 0, 30
    
    history = {"train_loss": [], "val_dice": [], "val_iou": [], "val_precision": [], "val_recall": [], "lr": []}

    for epoch in range(epochs):
        model.train()
        crit.set_epoch(epoch)
        tloss = 0.0
        
        pbar = tqdm(tr_ld, desc=f"E{epoch+1:03d}/{epochs} [Train]", ncols=100)
        for imgs, masks in pbar:
            imgs, masks = imgs.to(device, non_blocking=True), masks.to(device, non_blocking=True)

            with torch.amp.autocast("cuda"):
                seg_outs = model(imgs)
                if isinstance(seg_outs, tuple): seg_outs = seg_outs[0]
                if not isinstance(seg_outs, list): seg_outs = [seg_outs]

                ds_w = [0.05, 0.15, 0.30, 0.50][-len(seg_outs):]
                loss = sum(w * crit(o, masks) for w, o in zip(ds_w, seg_outs))

            optimizer.zero_grad(set_to_none=True)
            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            scaler.step(optimizer)
            scaler.update()
            
            ema.update(model)
            sched.step()

            tloss += loss.item()
            pbar.set_postfix({"loss": f"{loss.item():.4f}"})

        # ── Validate ───────────────────────────────────────────
        ema.shadow.eval()
        d_l, i_l, p_l, r_l = [], [], [], []
        use_tta = (epoch + 1) >= 80 and (epoch + 1) % 5 == 0

        with torch.no_grad():
            for imgs, masks in tqdm(vl_ld, desc=f"E{epoch+1:03d}/{epochs} [Val]  ", leave=False, ncols=100):
                imgs, masks = imgs.to(device, non_blocking=True), masks.to(device, non_blocking=True)
                
                if use_tta:
                    final_out = tta_predict(ema.shadow, imgs)
                else:
                    with torch.amp.autocast("cuda"):
                        out = ema.shadow(imgs)
                        final_out = out[-1] if isinstance(out, list) else out

                d, i, p, r = calc_metrics(final_out, masks)
                d_l.append(d); i_l.append(i); p_l.append(p); r_l.append(r)

        epoch_loss = tloss/len(tr_ld)
        vd, vi, vp, vr = np.mean(d_l), np.mean(i_l), np.mean(p_l), np.mean(r_l)
        cur_lr = optimizer.param_groups[0]["lr"]

        history["train_loss"].append(epoch_loss)
        history["val_dice"].append(vd)
        history["val_iou"].append(vi)
        history["val_precision"].append(vp)
        history["val_recall"].append(vr)
        history["lr"].append(cur_lr)

        tta_tag = " [TTA]" if use_tta else ""
        print(f"  Loss:{epoch_loss:.4f} | Dice:{vd:.4f} | IoU:{vi:.4f} | Prec:{vp:.4f} | Rec:{vr:.4f}{tta_tag}")

        if vd > best_dice:
            best_dice = vd
            torch.save(ema.shadow.state_dict(), "best_model_v5.pth")
            patience = 0
        else:
            patience += 1
            if patience >= PATIENCE:
                print(f"Early stop at epoch {epoch+1}.")
                break

        if (epoch + 1) % 5 == 0:
            plot_all(history)

    print(f"\nTraining done | Best EMA Dice: {best_dice:.4f}")
    
    # Run inference ONLY on validation samples
    run_inference("best_model_v5.pth", vl_samples, num_samples=5, device=device)

if __name__ == "__main__":
    train()
```

```python
"""
Mamba-UNet++ v5 — Extreme Performance Edition
Optimized for NVIDIA A40 (High VRAM, Max Throughput)
Target: ~30s per epoch at Batch Size 16
"""

import os
# SPEEDUP 1: Thread guards strictly enforced to prevent Dataloader multi-processing locks
os.environ["OMP_NUM_THREADS"] = "1"
os.environ["MKL_NUM_THREADS"] = "1"
os.environ["OPENBLAS_NUM_THREADS"] = "1"
os.environ["NUMEXPR_NUM_THREADS"] = "1"

import math, random, warnings
import cv2

cv2.setNumThreads(0)
cv2.ocl.setUseOpenCL(False)

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
import albumentations as A
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from scipy.ndimage import distance_transform_edt
from copy import deepcopy

warnings.filterwarnings("ignore")
# SPEEDUP 2: TF32 and CuDNN Benchmarking enable aggressive kernel selection on A40
torch.backends.cudnn.benchmark = True
torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True

try:
    from mamba_ssm import Mamba
    USE_FAST_MAMBA = True
    print("✓ Fast CUDA Mamba loaded")
except ImportError:
    USE_FAST_MAMBA = False
    print("⚠  mamba-ssm not found — using CBAM attention (fast fallback)")

try:
    import timm
    HAS_TIMM = True
    print("✓ timm loaded")
except ImportError:
    HAS_TIMM = False
    print("⚠  timm not found")

# ── Dataset paths & Scaled Constants ─────────────────────────────────
image_dir = "dataset/train/images"
mask_dir  = "dataset/train/labels"

TRAIN_PATCH_SIZE = 512   
PAD_RESIZE = 640         
BATCH_SIZE = 16          
# SPEEDUP 3: Max out A40 prefetching. 8 workers + prefetch factor keeps GPU fed.
NUM_WORKERS = 8          

def set_seed(s=42):
    random.seed(s)
    np.random.seed(s)
    torch.manual_seed(s)
    torch.cuda.manual_seed_all(s)

set_seed()

# =====================================================================
# 1. DATASET & CPU-TO-GPU OPTIMIZATIONS
# =====================================================================
class MedicalDataset(Dataset):
    def __init__(self, samples, transform=None):
        self.transform = transform
        self.samples = samples  

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, mask_path = self.samples[idx]

        im = cv2.imread(img_path)
        im = cv2.cvtColor(im, cv2.COLOR_BGR2RGB) if im is not None else np.zeros((PAD_RESIZE, PAD_RESIZE, 3), np.uint8)
        
        mk = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE) if mask_path else np.zeros(im.shape[:2], np.uint8)
        mk = (mk > 127).astype(np.float32)

        if self.transform:
            aug = self.transform(image=im, mask=mk)
            im, mk = aug["image"], aug["mask"]
            
        # SPEEDUP 4: Return UINT8 arrays from Dataloader. Do NOT normalize on CPU.
        # This reduces PCI-e bus bandwidth by 75% compared to Float32.
        im_t = torch.from_numpy(im.transpose(2, 0, 1)).contiguous() # (3, H, W)
        mk_t = torch.from_numpy(mk).unsqueeze(0).contiguous()       # (1, H, W)
            
        return im_t, mk_t

def get_train_transforms():
    # SPEEDUP 5: Replaced slow A.Affine with optimized A.ShiftScaleRotate. 
    # Removed A.Normalize and ToTensorV2 (Moved to GPU loop).
    return A.Compose([
        A.PadIfNeeded(min_height=TRAIN_PATCH_SIZE, min_width=TRAIN_PATCH_SIZE, border_mode=cv2.BORDER_CONSTANT, value=0),
        A.CropNonEmptyMaskIfExists(height=TRAIN_PATCH_SIZE, width=TRAIN_PATCH_SIZE, p=1.0),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.RandomRotate90(p=0.5),
        A.ShiftScaleRotate(scale_limit=0.15, rotate_limit=30, shift_limit=0.08, p=0.5, border_mode=cv2.BORDER_CONSTANT),
        A.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1, p=0.4),
    ])

def get_val_transforms():
    return A.Compose([
        A.LongestMaxSize(max_size=PAD_RESIZE),
        A.PadIfNeeded(min_height=PAD_RESIZE, min_width=PAD_RESIZE, border_mode=cv2.BORDER_CONSTANT, value=0),
    ])

# =====================================================================
# 2. METRICS ON GPU (Zero CPU/NumPy Conversions)
# =====================================================================
@torch.no_grad()
def calc_metrics_gpu(pred_logits, target):
    # SPEEDUP 6: 100% GPU-based metrics. Eliminates slow .cpu().numpy() loops.
    # Sigmoid > 0.5 is mathematically identical to logits > 0.0
    pred = (pred_logits > 0.0).float() 
    
    tp = (pred * target).sum(dim=(1, 2, 3))
    fp = (pred * (1 - target)).sum(dim=(1, 2, 3))
    fn = ((1 - pred) * target).sum(dim=(1, 2, 3))
    
    dice = (2 * tp + 1e-6) / (2 * tp + fp + fn + 1e-6)
    iou = (tp + 1e-6) / (tp + fp + fn + 1e-6)
    prec = (tp + 1e-6) / (tp + fp + 1e-6)
    rec = (tp + 1e-6) / (tp + fn + 1e-6)
    
    return dice.mean().item(), iou.mean().item(), prec.mean().item(), rec.mean().item()

class DiceBCELoss(nn.Module):
    def __init__(self, smooth=1.0):
        super().__init__()
        self.smooth = smooth
        self.bce = nn.BCEWithLogitsLoss()

    def forward(self, logits, targets):
        bce = self.bce(logits, targets)
        p, t = torch.sigmoid(logits).view(-1), targets.view(-1)
        dice = 1 - (2 * (p * t).sum() + self.smooth) / (p.sum() + t.sum() + self.smooth)
        return bce, dice

class LovaszHingeLoss(nn.Module):
    def forward(self, logits, targets):
        probs, tgts = torch.sigmoid(logits).view(-1), targets.view(-1)
        if tgts.sum() == 0 or tgts.sum() == tgts.numel(): return (logits * 0.0).mean()
        errors = (tgts - probs).abs()
        errors_sorted, perm = torch.sort(errors, descending=True)
        return torch.clamp(torch.dot(errors_sorted, self._lovasz_grad(tgts[perm])), 0.0, 2.0)

    @staticmethod
    def _lovasz_grad(gt_sorted):
        gts = gt_sorted.sum()
        intersection = gts - gt_sorted.cumsum(0)
        union = gts + (1 - gt_sorted).cumsum(0)
        jaccard = 1.0 - intersection / (union + 1e-6)
        if gt_sorted.numel() > 1: jaccard[1:] = jaccard[1:] - jaccard[:-1]
        return jaccard

class TverskyLoss(nn.Module):
    def __init__(self, alpha=0.7, beta=0.3, gamma=0.75, smooth=1e-3):
        super().__init__()
        self.alpha, self.beta, self.gamma, self.smooth = alpha, beta, gamma, smooth

    def forward(self, inputs, targets):
        p, t = torch.sigmoid(inputs).view(-1), targets.view(-1)
        tp, fp, fn = (p * t).sum(), (p * (1 - t)).sum(), ((1 - p) * t).sum()
        return (1 - ((tp + self.smooth) / (tp + self.alpha * fp + self.beta * fn + self.smooth))) ** self.gamma

class ScheduledLoss(nn.Module):
    def __init__(self):
        super().__init__()
        self.dice_bce = DiceBCELoss()
        self.tversky = TverskyLoss()
        self.lovasz = LovaszHingeLoss()
        self._epoch = 0

    def set_epoch(self, e): self._epoch = e

    def forward(self, logits, targets):
        logits, targets = logits.float(), targets.float() # Ensure float32 for loss
        bce, dice = self.dice_bce(logits, targets)
        tv, lov = self.tversky(logits, targets), self.lovasz(logits, targets)
        safe = lambda t: t if torch.isfinite(t) else torch.zeros_like(t)

        e = self._epoch
        w = (0.25, 0.25, 0.40, 0.10) if e < 20 else (0.20, 0.30, 0.30, 0.20) if e < 50 else (0.15, 0.30, 0.25, 0.30) if e < 90 else (0.10, 0.30, 0.20, 0.40)
        return w[0] * safe(bce) + w[1] * safe(dice) + w[2] * safe(tv) + w[3] * safe(lov)

# =====================================================================
# 3. BUILDING BLOCKS & BATCHED MAMBA
# =====================================================================
class FastMambaBlock(nn.Module):
    def __init__(self, d):
        super().__init__()
        self.mamba = Mamba(d_model=d, d_state=16, d_conv=4, expand=2)
        self.norm = nn.LayerNorm(d)
        
    def forward(self, x):
        # SPEEDUP 7: Batched Mamba Scans. 
        # Instead of launching 4 separate Mamba kernels per block (forward/back x horiz/vert),
        # we concatenate the flipped tensors to process 2x batch size simultaneously.
        B, C, H, W = x.shape
        L = H * W
        
        # --- Horizontal Scan (Forward + Backward batched) ---
        h = x.view(B, C, L).transpose(1, 2)
        h_flip = torch.flip(h, [1])
        h_batched = torch.cat([h, h_flip], dim=0) # Shape: (2B, L, C)
        
        h_out_batched = self.mamba(self.norm(h_batched))
        h_out, h_out_flip = h_out_batched.chunk(2, dim=0)
        oh = (h_out + torch.flip(h_out_flip, [1])) / 2
        
        # --- Vertical Scan (Forward + Backward batched) ---
        v = x.permute(0, 1, 3, 2).reshape(B, C, L).transpose(1, 2)
        v_flip = torch.flip(v, [1])
        v_batched = torch.cat([v, v_flip], dim=0) # Shape: (2B, L, C)
        
        v_out_batched = self.mamba(self.norm(v_batched))
        v_out, v_out_flip = v_out_batched.chunk(2, dim=0)
        ov = (v_out + torch.flip(v_out_flip, [1])) / 2
        
        out = (oh + ov) / 2
        return out.transpose(1, 2).view(B, C, H, W) + x

class SEBlock(nn.Module):
    def __init__(self, c, r=8):
        super().__init__()
        self.se = nn.Sequential(nn.AdaptiveAvgPool2d(1), nn.Flatten(), nn.Linear(c, c // r, bias=False), nn.ReLU(inplace=True), nn.Linear(c // r, c, bias=False), nn.Sigmoid())
    def forward(self, x): return x * self.se(x).view(x.size(0), x.size(1), 1, 1)

class ConvBlock(nn.Module):
    def __init__(self, in_c, out_c, drop=0.1):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_c, out_c, 3, padding=1, bias=False), nn.BatchNorm2d(out_c), nn.GELU(), nn.Dropout2d(drop),
            nn.Conv2d(out_c, out_c, 3, padding=1, bias=False), nn.BatchNorm2d(out_c), nn.GELU(),
        )
        self.se = SEBlock(out_c)
        self.proj = nn.Conv2d(in_c, out_c, 1, bias=False) if in_c != out_c else nn.Identity()
    def forward(self, x): return self.se(self.conv(x)) + self.proj(x)

class ASPP(nn.Module):
    def __init__(self, in_c, out_c=256):
        super().__init__()
        rates = [1, 6, 12, 18]
        self.convs = nn.ModuleList([
            nn.Sequential(nn.Conv2d(in_c, out_c, 3 if r > 1 else 1, padding=r if r > 1 else 0, dilation=r, bias=False), nn.BatchNorm2d(out_c), nn.ReLU(inplace=True)) for r in rates
        ])
        self.pool = nn.Sequential(nn.AdaptiveAvgPool2d(1), nn.Conv2d(in_c, out_c, 1, bias=False), nn.BatchNorm2d(out_c), nn.ReLU(inplace=True))
        self.proj = nn.Sequential(nn.Conv2d(out_c * (len(rates) + 1), out_c, 1, bias=False), nn.BatchNorm2d(out_c), nn.ReLU(inplace=True), nn.Dropout2d(0.1))
    def forward(self, x):
        feats = [c(x) for c in self.convs] + [F.interpolate(self.pool(x), size=x.shape[2:], mode="bilinear", align_corners=True)]
        return self.proj(torch.cat(feats, 1))

class AttentionGate(nn.Module):
    def __init__(self, Fg, Fl, Fi):
        super().__init__()
        self.Wg = nn.Sequential(nn.Conv2d(Fg, Fi, 1, bias=False), nn.BatchNorm2d(Fi))
        self.Wx = nn.Sequential(nn.Conv2d(Fl, Fi, 1, bias=False), nn.BatchNorm2d(Fi))
        self.psi = nn.Sequential(nn.Conv2d(Fi, 1, 1, bias=False), nn.BatchNorm2d(1), nn.Sigmoid())
    def forward(self, g, x):
        g1 = self.Wg(F.interpolate(g, x.shape[2:], mode="bilinear", align_corners=True))
        return x * self.psi(F.relu(g1 + self.Wx(x), inplace=True))

class ResidualAdapter(nn.Module):
    def __init__(self, in_c, out_c):
        super().__init__()
        self.needs_proj = in_c != out_c
        if self.needs_proj:
            self.proj, self.bn = nn.Conv2d(in_c, out_c, 1, bias=False), nn.BatchNorm2d(out_c)
            nn.init.kaiming_normal_(self.proj.weight, mode="fan_out")
    def forward(self, x): return self.bn(self.proj(x)) if self.needs_proj else x

class EfficientNetEncoder(nn.Module):
    def __init__(self, pretrained=True):
        super().__init__()
        self.stages = timm.create_model("efficientnet_b4", pretrained=pretrained, features_only=True)
        self.channels = self.stages.feature_info.channels()
    def forward(self, x): return self.stages(x)

class MambaUNetPPv4(nn.Module):
    def __init__(self, num_classes=1, deep_supervision=True, pretrained=True):
        super().__init__()
        self.deep_supervision = deep_supervision
        ContextBlock = FastMambaBlock if USE_FAST_MAMBA else None 

        self.encoder = EfficientNetEncoder(pretrained=pretrained) 
        enc_ch, dec_ch = self.encoder.channels, [64, 128, 192, 256, 256]

        self.adapt = nn.ModuleList([ResidualAdapter(ec, dc) for ec, dc in zip(enc_ch, dec_ch)])
        self.aspp, self.ctx3, self.ctx4 = ASPP(dec_ch[4], dec_ch[4]), ContextBlock(dec_ch[3]), ContextBlock(dec_ch[4])
        self.up = nn.Upsample(scale_factor=2, mode="bilinear", align_corners=True)

        C = dec_ch
        self.ag = nn.ModuleDict({
            f"{i}{j}": AttentionGate(C[i+1], C[i], C[i] // 2) for j in range(1, 5) for i in range(5-j)
        })

        self.d01, self.d11, self.d21, self.d31 = ConvBlock(C[0]+C[1], C[0]), ConvBlock(C[1]+C[2], C[1]), ConvBlock(C[2]+C[3], C[2]), ConvBlock(C[3]+C[4], C[3])
        self.d02, self.d12, self.d22 = ConvBlock(C[0]*2+C[1], C[0]), ConvBlock(C[1]*2+C[2], C[1]), ConvBlock(C[2]*2+C[3], C[2])
        self.d03, self.d13 = ConvBlock(C[0]*3+C[1], C[0]), ConvBlock(C[1]*3+C[2], C[1])
        self.d04 = ConvBlock(C[0]*4+C[1], C[0])

        self.heads = nn.ModuleList([nn.Conv2d(C[0], num_classes, 1) for _ in range(4)])

    def forward(self, x):
        f = [ad(fe) for ad, fe in zip(self.adapt, self.encoder(x))]
        f[3], f[4] = self.ctx3(f[3]), self.ctx4(self.aspp(f[4]))

        x01 = self.d01(torch.cat([self.ag["01"](f[1], f[0]), self.up(f[1])], 1))
        x11 = self.d11(torch.cat([self.ag["11"](f[2], f[1]), self.up(f[2])], 1))
        x21 = self.d21(torch.cat([self.ag["21"](f[3], f[2]), self.up(f[3])], 1))
        x31 = self.d31(torch.cat([self.ag["31"](f[4], f[3]), self.up(f[4])], 1))

        x02 = self.d02(torch.cat([f[0], self.ag["02"](x11, x01), self.up(x11)], 1))
        x12 = self.d12(torch.cat([f[1], self.ag["12"](x21, x11), self.up(x21)], 1))
        x22 = self.d22(torch.cat([f[2], self.ag["22"](x31, x21), self.up(x31)], 1))

        x03 = self.d03(torch.cat([f[0], x01, self.ag["03"](x12, x02), self.up(x12)], 1))
        x13 = self.d13(torch.cat([f[1], x11, self.ag["13"](x22, x12), self.up(x22)], 1))
        x04 = self.d04(torch.cat([f[0], x01, x02, self.ag["04"](x13, x03), self.up(x13)], 1))

        H, W = x.shape[2:]
        return [F.interpolate(h(o), size=(H, W), mode="bilinear", align_corners=True) for h, o in zip(self.heads, [x01, x02, x03, x04])]


class EMA:
    def __init__(self, model, decay=0.9995):
        self.decay = decay
        self.shadow = deepcopy(model).eval()
        for p in self.shadow.parameters(): p.requires_grad_(False)
    @torch.no_grad()
    def update(self, model):
        for s, m in zip(self.shadow.parameters(), model.parameters()): s.copy_(self.decay * s + (1 - self.decay) * m.data)
        for s, m in zip(self.shadow.buffers(), model.buffers()): s.copy_(m)
    def __call__(self, *args, **kwargs): return self.shadow(*args, **kwargs)

# =====================================================================
# 4. EXTREME SPEED TRAINING LOOP
# =====================================================================
def train():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"\nDevice: {device} | GPU: {torch.cuda.get_device_name(0)}")

    if not os.path.exists(image_dir) or not os.path.exists(mask_dir):
        raise ValueError(f"Paths not found: Ensure {image_dir} and {mask_dir} exist.")

    all_names = sorted([f for f in os.listdir(image_dir) if f.lower().endswith((".png", ".jpg", ".jpeg"))])
    samples = []
    for n in all_names:
        ip = os.path.join(image_dir, n)
        stem = n.rsplit(".", 1)[0]
        mp = os.path.join(mask_dir, stem + ".png") if os.path.exists(os.path.join(mask_dir, stem + ".png")) else None
        if mp: samples.append((ip, mp))

    random.shuffle(samples)
    split = int(0.8 * len(samples))
    tr_samples, vl_samples = samples[:split], samples[split:]

    tr_ds = MedicalDataset(tr_samples, transform=get_train_transforms())
    vl_ds = MedicalDataset(vl_samples, transform=get_val_transforms())

    # SPEEDUP 8: Max workers + persistent_workers + prefetch_factor keeps the A40 saturated
    tr_ld = DataLoader(tr_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, 
                       pin_memory=True, drop_last=True, persistent_workers=True, prefetch_factor=4)
    vl_ld = DataLoader(vl_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, 
                       pin_memory=True, persistent_workers=True)

    model = MambaUNetPPv4(deep_supervision=True, pretrained=HAS_TIMM).to(device)

    # SPEEDUP 9: SAFE PyTorch 2.x Compile. 
    # Mamba kernels deadlock `torch.compile`. We manually compile ONLY the CNN blocks.
    if hasattr(torch, "compile"):
        print("⚡ Safely compiling CNN Encoders/Decoders for massive speed boost...")
        model.encoder = torch.compile(model.encoder, mode="max-autotune")
        model.aspp = torch.compile(model.aspp, mode="max-autotune")
        model.d01 = torch.compile(model.d01); model.d11 = torch.compile(model.d11); model.d21 = torch.compile(model.d21)
        model.d02 = torch.compile(model.d02); model.d12 = torch.compile(model.d12); model.d22 = torch.compile(model.d22)
        model.d03 = torch.compile(model.d03); model.d13 = torch.compile(model.d13)
        model.d04 = torch.compile(model.d04)

    ema = EMA(model, decay=0.9995)
    crit = ScheduledLoss()

    enc_params = list(model.encoder.parameters())
    dec_params = [p for p in model.parameters() if id(p) not in {id(ep) for ep in enc_params}]

    optimizer = torch.optim.AdamW([
        {"params": enc_params, "lr": 5e-5},
        {"params": dec_params, "lr": 3e-4},
    ], weight_decay=1e-4)

    epochs = 150
    sched = torch.optim.lr_scheduler.OneCycleLR(optimizer, max_lr=[5e-4, 3e-3], epochs=epochs, 
                                                steps_per_epoch=len(tr_ld), pct_start=0.05)

    scaler = torch.amp.GradScaler("cuda")
    best_dice = 0.0
    
    # SPEEDUP 10: Tensors for GPU Normalization created ONCE
    mean_t = torch.tensor([0.485, 0.456, 0.406], device=device).view(1, 3, 1, 1)
    std_t = torch.tensor([0.229, 0.224, 0.225], device=device).view(1, 3, 1, 1)

    history = {"train_loss": [], "val_dice": [], "val_iou": [], "val_precision": [], "val_recall": [], "lr": []}
    
    # Track previous val metrics to pad history array on skipped epochs
    prev_vd, prev_vi, prev_vp, prev_vr = 0, 0, 0, 0

    for epoch in range(epochs):
        model.train()
        crit.set_epoch(epoch)
        tloss = 0.0
        
        pbar = tqdm(tr_ld, desc=f"E{epoch+1:03d}/{epochs} [Train]", ncols=100)
        for imgs, masks in pbar:
            # GPU Normalization is ~10x faster than CPU Albumentations
            imgs = imgs.to(device, non_blocking=True).float() / 255.0
            imgs = (imgs - mean_t) / std_t
            masks = masks.to(device, non_blocking=True)

            with torch.amp.autocast("cuda"):
                seg_outs = model(imgs)
                ds_w = [0.05, 0.15, 0.30, 0.50]
                loss = sum(w * crit(o, masks) for w, o in zip(ds_w, seg_outs))

            optimizer.zero_grad(set_to_none=True)
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
            
            ema.update(model)
            sched.step()

            tloss += loss.item()
            pbar.set_postfix({"loss": f"{loss.item():.4f}"})

        epoch_loss = tloss / len(tr_ld)
        cur_lr = optimizer.param_groups[0]["lr"]

        # SPEEDUP 11: Only run Validation every 5 epochs to prevent overhead
        run_val = (epoch + 1) % 5 == 0 or epoch == 0
        
        if run_val:
            ema.shadow.eval()
            d_l, i_l, p_l, r_l = [], [], [], []

            with torch.no_grad():
                for imgs, masks in tqdm(vl_ld, desc="[Val]", leave=False, ncols=100):
                    imgs = imgs.to(device, non_blocking=True).float() / 255.0
                    imgs = (imgs - mean_t) / std_t
                    masks = masks.to(device, non_blocking=True)
                    
                    with torch.amp.autocast("cuda"):
                        out = ema.shadow(imgs)[-1] # Take deep supervision top level

                    d, i, p, r = calc_metrics_gpu(out, masks)
                    d_l.append(d); i_l.append(i); p_l.append(p); r_l.append(r)

            prev_vd, prev_vi, prev_vp, prev_vr = np.mean(d_l), np.mean(i_l), np.mean(p_l), np.mean(r_l)

        history["train_loss"].append(epoch_loss)
        history["val_dice"].append(prev_vd)
        history["val_iou"].append(prev_vi)
        history["val_precision"].append(prev_vp)
        history["val_recall"].append(prev_vr)
        history["lr"].append(cur_lr)

        val_str = f"| Dice:{prev_vd:.4f} | IoU:{prev_vi:.4f} | Prec:{prev_vp:.4f} | Rec:{prev_vr:.4f}" if run_val else ""
        print(f"  Loss:{epoch_loss:.4f} {val_str}")

        if run_val and prev_vd > best_dice:
            best_dice = prev_vd
            torch.save(ema.shadow.state_dict(), "best_model_v5.pth")

if __name__ == "__main__":
    train()
```

```python
"""
Mamba-UNet++ v4 — FUSC2021 Edition (FULLY FIXED)
=================================================
All fixes integrated:
  1. DiceBCELoss pos_weight=15.0 (register_buffer, auto GPU move)
  2. TverskyLoss alpha=0.3/beta=0.7 (correct FN-penalising direction)
  3. 3-stage ScheduledLoss curriculum (BCE+Dice → Dice+Tversky → +Lovász)
  4. CropNonEmptyMaskIfExists with ignore_values=[0] (no blank crops)
  5. OneCycleLR max_lr=[5e-5,4e-4], pct_start=0.15, div_factor=50
  6. Centered pad-to-square (foot dead-centre, no top-left anchoring)
  7. TTA averages raw logits only (no inverse-sigmoid)
  8. HD95 removed from epoch loop (inference only)
  9. Thread guards + num_workers=4 + persistent_workers
  10. LovaszHingeLoss
"""

# ── Thread guards — MUST precede cv2/numpy ───────────────────────────
import os
os.environ["OMP_NUM_THREADS"]      = "1"
os.environ["MKL_NUM_THREADS"]      = "1"
os.environ["OPENBLAS_NUM_THREADS"] = "1"
os.environ["NUMEXPR_NUM_THREADS"]  = "1"

import random, warnings
import cv2
cv2.setNumThreads(0)
cv2.ocl.setUseOpenCL(False)

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from scipy.ndimage import distance_transform_edt
from copy import deepcopy

warnings.filterwarnings("ignore")
torch.backends.cudnn.benchmark = True

try:
    from mamba_ssm import Mamba
    USE_FAST_MAMBA = True
    print("✓ Fast CUDA Mamba loaded")
except ImportError:
    USE_FAST_MAMBA = False
    print("⚠  mamba-ssm not found — using CBAM fallback")

try:
    import timm
    HAS_TIMM = True
    print("✓ timm loaded — EfficientNet-B2 encoder")
except ImportError:
    HAS_TIMM = False
    print("⚠  timm not found — scratch encoder")

# ── Constants ────────────────────────────────────────────────────────
IMG_SIZE     = 320
CROP_SIZE    = 256
INFER_SIZE   = 320
INFER_PATCH  = 320
INFER_STRIDE = 240

image_dir = "dataset/train/images"
mask_dir  = "dataset/train/labels"


def set_seed(s=42):
    random.seed(s); np.random.seed(s)
    torch.manual_seed(s); torch.cuda.manual_seed_all(s)

set_seed()


# =====================================================================
# 1. CENTRED PAD-TO-SQUARE
# =====================================================================
def pad_to_square_center(image, mask=None, pad_value=0):
    h, w = image.shape[:2]
    if h == w:
        return (image, mask, (0,0,0,0)) if mask is not None else (image, (0,0,0,0))
    if h < w:
        total = w - h; top, bottom, left, right = total//2, total - total//2, 0, 0
    else:
        total = h - w; top, bottom, left, right = 0, 0, total//2, total - total//2
    kw = dict(borderType=cv2.BORDER_CONSTANT)
    image = cv2.copyMakeBorder(image, top, bottom, left, right,
                                value=(pad_value, pad_value, pad_value), **kw)
    if mask is not None:
        mask = cv2.copyMakeBorder(mask, top, bottom, left, right, value=0, **kw)
        return image, mask, (top, bottom, left, right)
    return image, (top, bottom, left, right)


def unpad(arr, pad_info, orig_h, orig_w):
    top, bot, left, right = pad_info
    h, w = arr.shape[:2]
    arr = arr[top: h - bot if bot else h,
              left: w - right if right else w]
    if arr.shape[:2] != (orig_h, orig_w):
        arr = cv2.resize(arr.astype(np.float32), (orig_w, orig_h),
                         interpolation=cv2.INTER_LINEAR)
    return arr


# =====================================================================
# 2. DATASET
# =====================================================================
class FUSC2021Dataset(Dataset):
    def __init__(self, image_dir, mask_dir, names, transform=None, training=True):
        self.transform = transform
        self.training  = training
        self.samples   = []
        for n in names:
            ip = os.path.join(image_dir, n)
            if not os.path.exists(ip) or os.path.getsize(ip) == 0:
                continue
            stem = n.rsplit(".", 1)[0]
            mp = os.path.join(mask_dir, stem + ".png")
            if not os.path.exists(mp):
                mp = os.path.join(mask_dir, n)
            self.samples.append({"img": ip, "mask": mp})

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        s  = self.samples[idx]
        im = cv2.imread(s["img"])
        im = cv2.cvtColor(im, cv2.COLOR_BGR2RGB) if im is not None \
             else np.zeros((IMG_SIZE, IMG_SIZE, 3), np.uint8)
        mk = cv2.imread(s["mask"], cv2.IMREAD_GRAYSCALE) \
             if os.path.exists(s["mask"]) \
             else np.zeros(im.shape[:2], np.uint8)
        mk = (mk > 127).astype(np.float32)

        # centred pad → resize
        im, mk, _ = pad_to_square_center(im, mk, pad_value=0)
        im = cv2.resize(im, (IMG_SIZE, IMG_SIZE))
        mk = cv2.resize(mk, (IMG_SIZE, IMG_SIZE), interpolation=cv2.INTER_NEAREST)

        if self.transform:
            aug = self.transform(image=im, mask=mk)
            im, mk = aug["image"], aug["mask"].unsqueeze(0)
        return im, mk


# =====================================================================
# 3. TRANSFORMS
# =====================================================================
def get_train_transforms():
    return A.Compose([
        # Guarantees every crop overlaps foreground — no blank-space training
        A.CropNonEmptyMaskIfExists(
            CROP_SIZE, CROP_SIZE,
            ignore_values=[0],
            p=1.0,
        ),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.3),
        A.RandomRotate90(p=0.4),
        A.Affine(scale=(0.9, 1.1), translate_percent=(-0.05, 0.05),
                 rotate=(-20, 20), p=0.4),
        A.CLAHE(clip_limit=3.0, tile_grid_size=(8, 8), p=0.3),
        A.RandomBrightnessContrast(0.15, 0.15, p=0.4),
        A.HueSaturationValue(6, 15, 8, p=0.3),
        A.Normalize(mean=(0.485, 0.456, 0.406),
                    std =(0.229, 0.224, 0.225)),
        ToTensorV2(),
    ])


def get_val_transforms():
    return A.Compose([
        A.Normalize(mean=(0.485, 0.456, 0.406),
                    std =(0.229, 0.224, 0.225)),
        ToTensorV2(),
    ])


# =====================================================================
# 4. METRICS  (no HD95 in epoch loop)
# =====================================================================
def calc_metrics(pred_logits, target, thr=0.5):
    pred = (torch.sigmoid(pred_logits) > thr).float().cpu().numpy()
    tgt  = target.float().cpu().numpy()
    dice_l, iou_l, prec_l, rec_l = [], [], [], []
    for p, t in zip(pred, tgt):
        p, t = p.squeeze(), t.squeeze()
        tp = float((p * t).sum())
        fp = float((p * (1 - t)).sum())
        fn = float(((1 - p) * t).sum())
        dice_l.append((2*tp + 1e-6) / (2*tp + fp + fn + 1e-6))
        iou_l .append((tp   + 1e-6) / (tp   + fp + fn + 1e-6))
        prec_l.append((tp   + 1e-6) / (tp   + fp      + 1e-6))
        rec_l .append((tp   + 1e-6) / (tp   + fn      + 1e-6))
    return map(np.mean, [dice_l, iou_l, prec_l, rec_l])


def hausdorff_95(pred_logits, target, thr=0.5):
    """Called ONLY at final inference — never during training."""
    pred = (torch.sigmoid(pred_logits) > thr).float().cpu().numpy()
    tgt  = target.float().cpu().numpy()
    hd_l = []
    for p, t in zip(pred, tgt):
        p, t = p.squeeze().astype(bool), t.squeeze().astype(bool)
        if not p.any() or not t.any():
            hd_l.append(float(np.sqrt(p.shape[0]**2 + p.shape[1]**2)))
            continue
        hd_l.append(np.percentile(np.concatenate([
            distance_transform_edt(~p)[t],
            distance_transform_edt(~t)[p]]), 95))
    return np.mean(hd_l)


# =====================================================================
# 5. LOSSES
# =====================================================================
class DiceBCELoss(nn.Module):
    """
    pos_weight=15.0 — FUSC foreground is <3% of pixels.
    register_buffer ensures the weight tensor moves to GPU automatically.
    """
    def __init__(self, smooth=1.0, pos_weight=15.0):
        super().__init__()
        self.smooth = smooth
        self.register_buffer("pw", torch.tensor([pos_weight]))

    def forward(self, logits, targets):
        logits, targets = logits.float(), targets.float()
        bce   = F.binary_cross_entropy_with_logits(
                    logits, targets, pos_weight=self.pw.to(logits.device))
        p, t  = torch.sigmoid(logits).view(-1), targets.view(-1)
        inter = (p * t).sum()
        dice  = 1 - (2 * inter + self.smooth) / (p.sum() + t.sum() + self.smooth)
        return bce, dice


class LovaszHingeLoss(nn.Module):
    def forward(self, logits, targets):
        logits, targets = logits.float(), targets.float()
        probs = torch.sigmoid(logits).view(-1)
        tgts  = targets.view(-1)
        if tgts.sum() == 0 or tgts.sum() == tgts.numel():
            return (logits * 0.0).mean()
        errors_sorted, perm = torch.sort((tgts - probs).abs(), descending=True)
        grad = self._lovasz_grad(tgts[perm])
        return torch.clamp(torch.dot(errors_sorted, grad), 0.0, 2.0)

    @staticmethod
    def _lovasz_grad(gt_sorted):
        gts   = gt_sorted.sum()
        inter = gts - gt_sorted.cumsum(0)
        union = gts + (1 - gt_sorted).cumsum(0)
        jac   = 1.0 - inter / (union + 1e-6)
        if gt_sorted.numel() > 1:
            jac[1:] = jac[1:] - jac[:-1]
        return jac


class TverskyLoss(nn.Module):
    """
    alpha=0.3 penalises FP (tolerate them).
    beta=0.7  penalises FN (punish missed ulcers hard).
    DO NOT swap these — previous alpha=0.7/beta=0.3 caused Prec=0.99, Rec=0.10.
    """
    def __init__(self, alpha=0.3, beta=0.7, gamma=0.75, smooth=1e-3):
        super().__init__()
        self.alpha, self.beta, self.gamma, self.smooth = alpha, beta, gamma, smooth

    def forward(self, inputs, targets):
        p  = torch.sigmoid(inputs.float()).view(-1)
        t  = targets.float().view(-1)
        tp = (p * t).sum()
        fp = (p * (1 - t)).sum()
        fn = ((1 - p) * t).sum()
        tv = (tp + self.smooth) / (
             tp + self.alpha * fp + self.beta * fn + self.smooth)
        return (1 - tv) ** self.gamma


class ScheduledLoss(nn.Module):
    """
    3-stage curriculum designed for extreme foreground sparsity:

    Stage 1  ep  0-19 : BCE(pw=15) + Dice  @ 0.60 / 0.40
             → model learns WHERE ulcers are before caring about shape

    Stage 2  ep 20-59 : Dice + Tversky     @ 0.40 / 0.60
             → BCE removed; Tversky(β=0.7) drives recall up hard

    Stage 3  ep 60+   : Dice + Tversky + Lovász  @ 0.30 / 0.40 / 0.30
             → boundary refinement once recall is healthy
    """
    def __init__(self):
        super().__init__()
        self.dice_bce = DiceBCELoss(pos_weight=15.0)
        self.tversky  = TverskyLoss(alpha=0.3, beta=0.7, gamma=0.75)
        self.lovasz   = LovaszHingeLoss()
        self._epoch   = 0

    def set_epoch(self, e):
        self._epoch = e

    def forward(self, logits, targets):
        safe = lambda t: t if torch.isfinite(t) \
                         else torch.zeros(1, device=logits.device)[0]
        e = self._epoch

        if e < 20:
            bce, dice = self.dice_bce(logits, targets)
            return 0.60 * safe(bce) + 0.40 * safe(dice)

        elif e < 60:
            _, dice = self.dice_bce(logits, targets)
            tv      = self.tversky(logits, targets)
            return 0.40 * safe(dice) + 0.60 * safe(tv)

        else:
            _, dice = self.dice_bce(logits, targets)
            tv      = self.tversky(logits, targets)
            lov     = self.lovasz(logits, targets)
            return 0.30 * safe(dice) + 0.40 * safe(tv) + 0.30 * safe(lov)


# =====================================================================
# 6. BUILDING BLOCKS
# =====================================================================
class SEBlock(nn.Module):
    def __init__(self, c, r=8):
        super().__init__()
        self.se = nn.Sequential(
            nn.AdaptiveAvgPool2d(1), nn.Flatten(),
            nn.Linear(c, c // r, bias=False), nn.ReLU(inplace=True),
            nn.Linear(c // r, c, bias=False), nn.Sigmoid())

    def forward(self, x):
        return x * self.se(x).view(x.size(0), x.size(1), 1, 1)


class CBAM(nn.Module):
    def __init__(self, c, r=8):
        super().__init__()
        self.mlp = nn.Sequential(
            nn.Linear(c, c // r, bias=False), nn.ReLU(inplace=True),
            nn.Linear(c // r, c, bias=False))
        self.spatial = nn.Sequential(
            nn.Conv2d(2, 1, 7, padding=3, bias=False),
            nn.BatchNorm2d(1), nn.Sigmoid())

    def forward(self, x):
        avg, mx = x.mean(dim=(2, 3)), x.amax(dim=(2, 3))
        ca = torch.sigmoid(self.mlp(avg) + self.mlp(mx)).unsqueeze(-1).unsqueeze(-1)
        x  = x * ca
        sa = torch.cat([x.mean(1, keepdim=True), x.amax(1, keepdim=True)], 1)
        return x * self.spatial(sa)


class ConvBlock(nn.Module):
    def __init__(self, in_c, out_c, drop=0.1):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_c, out_c, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_c), nn.GELU(), nn.Dropout2d(drop),
            nn.Conv2d(out_c, out_c, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_c), nn.GELU())
        self.se   = SEBlock(out_c)
        self.proj = nn.Conv2d(in_c, out_c, 1, bias=False) \
                    if in_c != out_c else nn.Identity()

    def forward(self, x):
        return self.se(self.conv(x)) + self.proj(x)


class ASPP(nn.Module):
    def __init__(self, in_c, out_c=256):
        super().__init__()
        rates = [1, 6, 12, 18]
        self.convs = nn.ModuleList([
            nn.Sequential(
                nn.Conv2d(in_c, out_c,
                          3 if r > 1 else 1,
                          padding=r if r > 1 else 0,
                          dilation=r, bias=False),
                nn.BatchNorm2d(out_c), nn.ReLU(inplace=True))
            for r in rates])
        self.pool = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Conv2d(in_c, out_c, 1, bias=False),
            nn.BatchNorm2d(out_c), nn.ReLU(inplace=True))
        self.proj = nn.Sequential(
            nn.Conv2d(out_c * (len(rates) + 1), out_c, 1, bias=False),
            nn.BatchNorm2d(out_c), nn.ReLU(inplace=True), nn.Dropout2d(0.1))

    def forward(self, x):
        h, w  = x.shape[2:]
        feats = [c(x) for c in self.convs]
        feats.append(F.interpolate(self.pool(x), (h, w),
                                   mode="bilinear", align_corners=True))
        return self.proj(torch.cat(feats, 1))


class AttentionGate(nn.Module):
    def __init__(self, Fg, Fl, Fi):
        super().__init__()
        self.Wg  = nn.Sequential(nn.Conv2d(Fg, Fi, 1, bias=False), nn.BatchNorm2d(Fi))
        self.Wx  = nn.Sequential(nn.Conv2d(Fl, Fi, 1, bias=False), nn.BatchNorm2d(Fi))
        self.psi = nn.Sequential(nn.Conv2d(Fi, 1,  1, bias=False), nn.BatchNorm2d(1),
                                  nn.Sigmoid())

    def forward(self, g, x):
        g1 = self.Wg(F.interpolate(g, x.shape[2:], mode="bilinear", align_corners=True))
        return x * self.psi(F.relu(g1 + self.Wx(x), inplace=True))


class FastMambaBlock(nn.Module):
    def __init__(self, d):
        super().__init__()
        self.mamba = Mamba(d_model=d, d_state=16, d_conv=4, expand=2)
        self.norm  = nn.LayerNorm(d)

    def _scan(self, x, flip=False):
        s = torch.flip(x, [1]) if flip else x
        o = self.mamba(self.norm(s))
        return torch.flip(o, [1]) if flip else o

    def forward(self, x):
        B, C, H, W = x.shape
        h  = x.view(B, C, -1).transpose(1, 2)
        oh = (self._scan(h) + self._scan(h, flip=True)) / 2
        v  = x.permute(0, 1, 3, 2).contiguous().view(B, C, -1).transpose(1, 2)
        ov = (self._scan(v) + self._scan(v, flip=True)) / 2
        return ((oh + ov) / 2).transpose(1, 2).view(B, C, H, W) + x


# =====================================================================
# 7. ENCODER
# =====================================================================
class EfficientNetEncoder(nn.Module):
    def __init__(self, pretrained=True):
        super().__init__()
        self.stages   = timm.create_model("efficientnet_b2", pretrained=pretrained,
                                           features_only=True)
        self.channels = self.stages.feature_info.channels()
        print(f"  EfficientNet-B2 channels: {self.channels}")

    def forward(self, x):
        return self.stages(x)


class ScratchEncoder(nn.Module):
    def __init__(self):
        super().__init__()
        F_ = [32, 64, 128, 256, 512]
        self.channels = F_
        self.s0 = ConvBlock(3, F_[0])
        self.s1 = nn.Sequential(nn.MaxPool2d(2), ConvBlock(F_[0], F_[1]))
        self.s2 = nn.Sequential(nn.MaxPool2d(2), ConvBlock(F_[1], F_[2]))
        self.s3 = nn.Sequential(nn.MaxPool2d(2), ConvBlock(F_[2], F_[3]), CBAM(F_[3]))
        self.s4 = nn.Sequential(nn.MaxPool2d(2), ConvBlock(F_[3], F_[4]), CBAM(F_[4]))

    def forward(self, x):
        f0 = self.s0(x);  f1 = self.s1(f0); f2 = self.s2(f1)
        f3 = self.s3(f2); f4 = self.s4(f3)
        return [f0, f1, f2, f3, f4]


class ResidualAdapter(nn.Module):
    def __init__(self, in_c, out_c):
        super().__init__()
        self.needs_proj = in_c != out_c
        if self.needs_proj:
            self.proj = nn.Conv2d(in_c, out_c, 1, bias=False)
            self.bn   = nn.BatchNorm2d(out_c)

    def forward(self, x):
        return self.bn(self.proj(x)) if self.needs_proj else x


# =====================================================================
# 8. MODEL
# =====================================================================
class MambaUNetPPv4(nn.Module):
    def __init__(self, num_classes=1, deep_supervision=True, pretrained=True):
        super().__init__()
        self.deep_supervision = deep_supervision
        ContextBlock = FastMambaBlock if USE_FAST_MAMBA else CBAM

        self.encoder = EfficientNetEncoder(pretrained) if HAS_TIMM else ScratchEncoder()
        enc_ch = self.encoder.channels
        dec_ch = [48, 96, 160, 224, 224]

        self.adapt = nn.ModuleList([ResidualAdapter(ec, dc)
                                    for ec, dc in zip(enc_ch, dec_ch)])
        self.aspp  = ASPP(dec_ch[4], dec_ch[4])
        self.ctx3  = ContextBlock(dec_ch[3])
        self.ctx4  = ContextBlock(dec_ch[4])

        C        = dec_ch
        self.up  = nn.Upsample(scale_factor=2, mode="bilinear", align_corners=True)

        self.ag = nn.ModuleDict({
            "01": AttentionGate(C[1], C[0], C[0]//2),
            "11": AttentionGate(C[2], C[1], C[1]//2),
            "21": AttentionGate(C[3], C[2], C[2]//2),
            "31": AttentionGate(C[4], C[3], C[3]//2),
            "02": AttentionGate(C[1], C[0], C[0]//2),
            "12": AttentionGate(C[2], C[1], C[1]//2),
            "22": AttentionGate(C[3], C[2], C[2]//2),
            "03": AttentionGate(C[1], C[0], C[0]//2),
            "13": AttentionGate(C[2], C[1], C[1]//2),
            "04": AttentionGate(C[1], C[0], C[0]//2),
        })

        self.d01 = ConvBlock(C[0] + C[1],     C[0])
        self.d11 = ConvBlock(C[1] + C[2],     C[1])
        self.d21 = ConvBlock(C[2] + C[3],     C[2])
        self.d31 = ConvBlock(C[3] + C[4],     C[3])
        self.d02 = ConvBlock(C[0]*2 + C[1],   C[0])
        self.d12 = ConvBlock(C[1]*2 + C[2],   C[1])
        self.d22 = ConvBlock(C[2]*2 + C[3],   C[2])
        self.d03 = ConvBlock(C[0]*3 + C[1],   C[0])
        self.d13 = ConvBlock(C[1]*3 + C[2],   C[1])
        self.d04 = ConvBlock(C[0]*4 + C[1],   C[0])

        if deep_supervision:
            self.heads = nn.ModuleList([nn.Conv2d(C[0], num_classes, 1)
                                        for _ in range(4)])
        else:
            self.head = nn.Conv2d(C[0], num_classes, 1)

        self._init_weights()

    def _init_weights(self):
        enc_ids = {id(p) for p in self.encoder.parameters()}
        for m in self.modules():
            if id(m) in enc_ids:
                continue
            if isinstance(m, nn.Conv2d):
                nn.init.kaiming_normal_(m.weight, mode="fan_out", nonlinearity="relu")
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.ones_(m.weight); nn.init.zeros_(m.bias)

    def forward(self, x):
        feats = self.encoder(x)
        f     = [ad(fe) for ad, fe in zip(self.adapt, feats)]
        f[3]  = self.ctx3(f[3])
        f[4]  = self.ctx4(self.aspp(f[4]))

        x01 = self.d01(torch.cat([self.ag["01"](f[1], f[0]),  self.up(f[1])], 1))
        x11 = self.d11(torch.cat([self.ag["11"](f[2], f[1]),  self.up(f[2])], 1))
        x21 = self.d21(torch.cat([self.ag["21"](f[3], f[2]),  self.up(f[3])], 1))
        x31 = self.d31(torch.cat([self.ag["31"](f[4], f[3]),  self.up(f[4])], 1))

        x02 = self.d02(torch.cat([f[0], self.ag["02"](x11, x01), self.up(x11)], 1))
        x12 = self.d12(torch.cat([f[1], self.ag["12"](x21, x11), self.up(x21)], 1))
        x22 = self.d22(torch.cat([f[2], self.ag["22"](x31, x21), self.up(x31)], 1))

        x03 = self.d03(torch.cat([f[0], x01, self.ag["03"](x12, x02), self.up(x12)], 1))
        x13 = self.d13(torch.cat([f[1], x11, self.ag["13"](x22, x12), self.up(x22)], 1))
        x04 = self.d04(torch.cat([f[0], x01, x02,
                                   self.ag["04"](x13, x03), self.up(x13)], 1))

        H, W = x.shape[2:]
        if self.deep_supervision:
            return [F.interpolate(h(o), (H, W), mode="bilinear", align_corners=True)
                    for h, o in zip(self.heads, [x01, x02, x03, x04])]
        return F.interpolate(self.head(x04), (H, W),
                             mode="bilinear", align_corners=True)


# =====================================================================
# 9. EMA
# =====================================================================
class EMA:
    def __init__(self, model, decay=0.9995):
        self.decay  = decay
        self.shadow = deepcopy(model).eval()
        for p in self.shadow.parameters():
            p.requires_grad_(False)

    @torch.no_grad()
    def update(self, model):
        for s, m in zip(self.shadow.parameters(), model.parameters()):
            s.copy_(self.decay * s + (1 - self.decay) * m.data)
        for s, m in zip(self.shadow.buffers(), model.buffers()):
            s.copy_(m)

    def __call__(self, *a, **kw):
        return self.shadow(*a, **kw)


# =====================================================================
# 10. TTA — averages RAW LOGITS, sigmoid applied once
# =====================================================================
@torch.no_grad()
def tta_predict(model, imgs, device):
    logit_sum = torch.zeros_like(imgs[:, :1])
    count = 0
    for fh in [False, True]:
        for fv in [False, True]:
            x = torch.flip(imgs, [3]) if fh else imgs.clone()
            if fv: x = torch.flip(x, [2])
            with torch.amp.autocast("cuda"):
                out = model(x)
                lg  = out[-1] if isinstance(out, list) else out
            if fv: lg = torch.flip(lg, [2])
            if fh: lg = torch.flip(lg, [3])
            logit_sum += lg.float()
            count += 1
    return logit_sum / count


# =====================================================================
# 11. SLIDING-WINDOW INFERENCE
# =====================================================================
@torch.no_grad()
def sliding_window_inference(model, image_np, device,
                              patch_size=INFER_PATCH, stride=INFER_STRIDE,
                              use_tta=False):
    mean = np.array([0.485, 0.456, 0.406], dtype=np.float32)
    std  = np.array([0.229, 0.224, 0.225], dtype=np.float32)
    H, W = image_np.shape[:2]
    logit_acc = np.zeros((1, 1, H, W), dtype=np.float32)
    count_acc = np.zeros((1, 1, H, W), dtype=np.float32)

    pad_h = (stride - (H - patch_size) % stride) % stride
    pad_w = (stride - (W - patch_size) % stride) % stride
    img_p = cv2.copyMakeBorder(image_np, 0, pad_h, 0, pad_w,
                                cv2.BORDER_CONSTANT, value=(0, 0, 0))
    Hp, Wp = img_p.shape[:2]

    for y in range(0, Hp - patch_size + 1, stride):
        for x in range(0, Wp - patch_size + 1, stride):
            patch = img_p[y:y + patch_size, x:x + patch_size]
            pt = (torch.from_numpy(
                    ((patch.astype(np.float32) / 255.0 - mean) / std)
                    .transpose(2, 0, 1))
                  .unsqueeze(0).to(device))
            with torch.amp.autocast("cuda"):
                if use_tta:
                    lg = tta_predict(model, pt, device)
                else:
                    out = model(pt)
                    lg  = out[-1] if isinstance(out, list) else out
            lg_np      = lg.float().cpu().numpy()
            ye, xe     = min(y + patch_size, H), min(x + patch_size, W)
            ph, pw     = ye - y, xe - x
            logit_acc[:, :, y:ye, x:xe] += lg_np[:, :, :ph, :pw]
            count_acc[:, :, y:ye, x:xe] += 1.0

    return (logit_acc / np.maximum(count_acc, 1.0))[0, 0]


# =====================================================================
# 12. PLOTTING
# =====================================================================
def plot_all(history, path="training_metrics_fusc2021.png"):
    eps = range(1, len(history["train_loss"]) + 1)
    fig, ax = plt.subplots(2, 3, figsize=(18, 10))
    fig.suptitle("Mamba-UNet++ v4 FUSC2021", fontsize=16, fontweight="bold")

    def _p(a, key, lbl, col, tgt=None):
        a.plot(eps, history[key], color=col, marker="o", markersize=2, label=lbl)
        if tgt:
            a.axhline(tgt, color="limegreen", ls="--", lw=1, alpha=0.7)
        a.legend(fontsize=8); a.grid(True, alpha=0.3)

    _p(ax[0,0], "train_loss", "Train Loss", "red")
    ax[0,0].set_title("Loss")
    _p(ax[0,1], "val_dice",   "Val Dice",   "royalblue", 0.75)
    ax[0,1].set_title("Dice"); ax[0,1].set_ylim(0, 1)
    _p(ax[0,2], "val_iou",    "Val IoU",    "seagreen")
    ax[0,2].set_title("IoU");  ax[0,2].set_ylim(0, 1)

    ax[1,0].plot(eps, history["val_precision"], color="purple",
                 marker="o", markersize=2, label="Prec")
    ax[1,0].plot(eps, history["val_recall"],    color="teal",
                 marker="s", markersize=2, label="Rec")
    ax[1,0].legend(fontsize=8); ax[1,0].grid(True, alpha=0.3)
    ax[1,0].set_ylim(0, 1); ax[1,0].set_title("Precision & Recall")

    ax[1,1].axis("off")
    ax[1,1].text(0.5, 0.5, "HD95 only\nat inference",
                 ha="center", va="center", fontsize=12,
                 style="italic", color="gray")

    ax[1,2].plot(eps, history["lr"], "k", marker="o", markersize=2)
    ax[1,2].set_title("LR"); ax[1,2].set_yscale("log")
    ax[1,2].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig(path, dpi=150, bbox_inches="tight")
    plt.close()


# =====================================================================
# 13. INFERENCE VISUALISATION
# =====================================================================
def run_inference(model_path, image_dir, mask_dir, num_samples=4, device=None):
    if device is None:
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = MambaUNetPPv4(deep_supervision=False, pretrained=False).to(device)
    model.load_state_dict(torch.load(model_path, map_location=device), strict=False)
    model.eval()
    print(f"\n✓ Loaded '{model_path}'")

    ext   = (".png", ".jpg", ".jpeg")
    imgs  = sorted([f for f in os.listdir(image_dir) if f.lower().endswith(ext)])
    picks = random.sample(imgs, min(num_samples, len(imgs)))

    fig, axes = plt.subplots(num_samples, 4, figsize=(17, 4 * num_samples))
    if num_samples == 1:
        axes = axes[np.newaxis]
    fig.suptitle("Inference — Mamba-UNet++ v4 FUSC2021",
                 fontsize=14, fontweight="bold")
    for c, t in enumerate(["Input", "Ground Truth", "Prediction", "Overlay (TP/FP/FN)"]):
        axes[0, c].set_title(t, fontsize=11, fontweight="bold")

    all_hd95 = []
    for row, name in enumerate(picks):
        ip  = os.path.join(image_dir, name)
        mp  = os.path.join(mask_dir, name.rsplit(".", 1)[0] + ".png")
        if not os.path.exists(mp):
            mp = os.path.join(mask_dir, name)

        orig   = cv2.cvtColor(cv2.imread(ip), cv2.COLOR_BGR2RGB)
        oh, ow = orig.shape[:2]
        gt_raw = cv2.imread(mp, cv2.IMREAD_GRAYSCALE) if os.path.exists(mp) \
                 else np.zeros((oh, ow), np.uint8)
        gt = (gt_raw > 127).astype(np.float32)

        orig_sq, gt_sq, pad_info = pad_to_square_center(orig, gt, pad_value=0)
        sq_h, sq_w = orig_sq.shape[:2]
        logit_map  = sliding_window_inference(model, orig_sq, device,
                                              patch_size=INFER_PATCH,
                                              stride=INFER_STRIDE, use_tta=True)
        pred_sq = (1 / (1 + np.exp(-logit_map)) > 0.5).astype(np.float32)
        pred    = (unpad(pred_sq, pad_info, oh, ow) > 0.5).astype(np.float32)

        gtt  = torch.from_numpy(gt).unsqueeze(0).unsqueeze(0)
        pt   = torch.from_numpy(pred).unsqueeze(0).unsqueeze(0)
        dice = float((2*(pt*gtt).sum()+1e-6) / (pt.sum()+gtt.sum()+1e-6))
        iou  = float(((pt*gtt).sum()+1e-6)  / ((pt+gtt-(pt*gtt)).sum()+1e-6))

        logit_t = torch.from_numpy(logit_map).unsqueeze(0).unsqueeze(0)
        gt_rsz  = torch.from_numpy(
                    cv2.resize(gt, (sq_w, sq_h),
                               interpolation=cv2.INTER_NEAREST)
                  ).unsqueeze(0).unsqueeze(0)
        hd = hausdorff_95(logit_t, gt_rsz)
        all_hd95.append(hd)

        ds     = (320, 320)
        orig_d = cv2.resize(orig, ds)
        gt_d   = cv2.resize(gt,   ds, interpolation=cv2.INTER_NEAREST)
        pred_d = cv2.resize(pred, ds, interpolation=cv2.INTER_NEAREST)

        ov = orig_d.copy().astype(np.float32)
        tp = (pred_d > 0.5) & (gt_d > 0.5)
        fp = (pred_d > 0.5) & (gt_d <= 0.5)
        fn = (pred_d <= 0.5) & (gt_d > 0.5)
        ov[tp] = ov[tp]*0.45 + np.array([0,   220, 0  ])*0.55
        ov[fp] = ov[fp]*0.45 + np.array([220, 0,   0  ])*0.55
        ov[fn] = ov[fn]*0.45 + np.array([0,   0,   220])*0.55

        axes[row, 0].imshow(orig_d)
        axes[row, 0].set_ylabel(name[:18], fontsize=7)
        axes[row, 1].imshow(gt_d,   cmap="gray")
        axes[row, 2].imshow(pred_d, cmap="gray")
        axes[row, 2].set_xlabel(
            f"Dice={dice:.4f}  IoU={iou:.4f}  HD95={hd:.1f}px", fontsize=8)
        axes[row, 3].imshow(np.clip(ov, 0, 255).astype(np.uint8))
        for a in axes[row]:
            a.axis("off")

    patches = [mpatches.Patch(color=c, label=l)
               for c, l in [("lime","TP"), ("red","FP"), ("blue","FN")]]
    fig.legend(handles=patches, loc="lower center", ncol=3,
               fontsize=10, bbox_to_anchor=(0.5, -0.005))
    plt.tight_layout()
    plt.savefig("inference_results_fusc2021.png", dpi=150, bbox_inches="tight")
    plt.close()
    print("✓ Inference grid → 'inference_results_fusc2021.png'")
    print(f"  Mean HD95: {np.mean(all_hd95):.1f} px")


# =====================================================================
# 14. TRAINING
# =====================================================================
def train():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"\nDevice: {device}")
    if torch.cuda.is_available():
        print(f"GPU: {torch.cuda.get_device_name(0)}")

    img_dir = image_dir
    msk_dir = mask_dir
    if not os.path.exists(img_dir):
        raise ValueError(f"Not found: {img_dir}")

    ext  = (".png", ".jpg", ".jpeg")
    all_ = sorted([f for f in os.listdir(img_dir) if f.lower().endswith(ext)])
    if not all_:
        raise ValueError(f"No images in {img_dir}")

    random.shuffle(all_)
    split    = int(0.8 * len(all_))
    tr_names = all_[:split]
    vl_names = all_[split:]
    print(f"Train: {len(tr_names)} | Val: {len(vl_names)}")

    tr_ds = FUSC2021Dataset(img_dir, msk_dir, tr_names,
                             transform=get_train_transforms(), training=True)
    vl_ds = FUSC2021Dataset(img_dir, msk_dir, vl_names,
                             transform=get_val_transforms(), training=False)

    BATCH = 12
    tr_ld = DataLoader(tr_ds, batch_size=BATCH, shuffle=True,
                       num_workers=4, pin_memory=True, persistent_workers=True)
    vl_ld = DataLoader(vl_ds, batch_size=8,     shuffle=False,
                       num_workers=4, pin_memory=True, persistent_workers=True)

    model = MambaUNetPPv4(deep_supervision=True, pretrained=HAS_TIMM).to(device)
    ema   = EMA(model, decay=0.9995)
    crit  = ScheduledLoss()

    enc_params    = list(model.encoder.parameters())
    enc_param_ids = {id(p) for p in enc_params}
    dec_params    = [p for p in model.parameters()
                     if id(p) not in enc_param_ids]

    # base lr = max_lr / div_factor = 5e-5/50 = 1e-6 (encoder nearly frozen at start)
    optimizer = torch.optim.AdamW(
        [{"params": enc_params, "lr": 5e-5/50, "name": "encoder"},
         {"params": dec_params, "lr": 4e-4/50, "name": "decoder"}],
        weight_decay=1e-4, betas=(0.9, 0.999))

    epochs = 150
    sched  = torch.optim.lr_scheduler.OneCycleLR(
        optimizer,
        max_lr          = [5e-5, 4e-4],    # halved vs previous; prevents ep-5 collapse
        epochs          = epochs,
        steps_per_epoch = len(tr_ld),
        pct_start       = 0.15,             # warmup over ~22 epochs
        anneal_strategy = "cos",
        div_factor      = 50,               # start = max_lr / 50
        final_div_factor= 1000,
    )

    scaler = torch.amp.GradScaler("cuda")
    ds_w   = [0.05, 0.15, 0.30, 0.50]
    ACCUM  = 2

    best_dice, patience, PATIENCE = 0.0, 0, 35
    history = {k: [] for k in ["train_loss", "val_dice", "val_iou",
                                "val_precision", "val_recall", "lr"]}

    print(f"\nTraining: {epochs} epochs")
    print(f"  OneCycleLR  max_lr=[5e-5, 4e-4]  pct_start=0.15  div_factor=50")
    print(f"  Loss curriculum: BCE+Dice(0-19) → Dice+Tversky(20-59) → +Lovász(60+)")
    print(f"  Tversky α=0.3 β=0.7  |  BCE pos_weight=15.0\n")

    for epoch in range(epochs):
        model.train()
        crit.set_epoch(epoch)
        tloss, n_valid = 0.0, 0
        optimizer.zero_grad(set_to_none=True)

        pbar = tqdm(enumerate(tr_ld), total=len(tr_ld),
                    desc=f"E{epoch+1:03d}/{epochs} [Train]", ncols=110)

        for step, (imgs, masks) in pbar:
            imgs  = imgs.to(device,  non_blocking=True)
            masks = masks.to(device, non_blocking=True)

            with torch.amp.autocast("cuda"):
                seg_outs = model(imgs)
                if not isinstance(seg_outs, list):
                    seg_outs = [seg_outs]
                loss = sum(w * crit(o, masks)
                           for w, o in zip(ds_w[-len(seg_outs):], seg_outs)) / ACCUM

            if not torch.isfinite(loss):
                optimizer.zero_grad(set_to_none=True)
                continue

            scaler.scale(loss).backward()
            n_valid += 1

            if (step + 1) % ACCUM == 0 or (step + 1) == len(tr_ld):
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad(set_to_none=True)
                ema.update(model)

            sched.step()
            tloss += loss.item() * ACCUM
            pbar.set_postfix({"loss": f"{loss.item()*ACCUM:.4f}"})

        epoch_loss = tloss / max(n_valid, 1)

        # ── Validate with EMA — HD95 NOT computed here ───────────────
        ema.shadow.eval()
        d_l, i_l, p_l, r_l = [], [], [], []
        use_tta = (epoch + 1) >= 80 and (epoch + 1) % 5 == 0

        with torch.no_grad():
            for imgs, masks in tqdm(vl_ld,
                                    desc=f"E{epoch+1:03d}/{epochs} [Val]  ",
                                    leave=False, ncols=110):
                imgs  = imgs.to(device,  non_blocking=True)
                masks = masks.to(device, non_blocking=True)
                if use_tta:
                    final_out = tta_predict(ema.shadow, imgs, device)
                else:
                    with torch.amp.autocast("cuda"):
                        out = ema.shadow(imgs)
                        final_out = out[-1] if isinstance(out, list) else out
                d, i, p, r = calc_metrics(final_out, masks)
                d_l.append(d); i_l.append(i); p_l.append(p); r_l.append(r)

        vd, vi, vp, vr = map(np.mean, [d_l, i_l, p_l, r_l])
        cur_lr = optimizer.param_groups[0]["lr"]

        for k, v in zip(["train_loss","val_dice","val_iou",
                          "val_precision","val_recall","lr"],
                         [epoch_loss, vd, vi, vp, vr, cur_lr]):
            history[k].append(v)

        stage = "BCE+Dice" if epoch < 20 else \
                "Dice+Tversky" if epoch < 60 else "Dice+Tversky+Lovász"
        print(f"  [{stage}] Loss:{epoch_loss:.4f} | Dice:{vd:.4f} | "
              f"IoU:{vi:.4f} | Prec:{vp:.4f} | Rec:{vr:.4f} | LR:{cur_lr:.1e}"
              + (" [TTA]" if use_tta else ""))

        if vd > best_dice:
            best_dice = vd
            torch.save(ema.shadow.state_dict(),
                       "best_mamba_unetpp_v4_fusc2021.pth")
            print(f"  ✓ Best saved! (Dice={best_dice:.4f})")
            patience = 0
        else:
            patience += 1
            if patience >= PATIENCE:
                print(f"\nEarly stop epoch {epoch+1}. Best Dice={best_dice:.4f}")
                break

        if (epoch + 1) % 5 == 0:
            plot_all(history)

    print(f"\n{'='*60}\n Best EMA Dice: {best_dice:.4f}\n{'='*60}")
    run_inference("best_mamba_unetpp_v4_fusc2021.pth",
                  img_dir, msk_dir, num_samples=4, device=device)


if __name__ == "__main__":
    train()
```

```python
import os
import cv2
import numpy as np
import random
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
import matplotlib.subplots as subplots
import matplotlib.pyplot as plt
from scipy.spatial.distance import directed_hausdorff

# ==========================================
# 0. SPEED OPTIMIZATIONS
# ==========================================
torch.backends.cudnn.benchmark = True

try:
    from mamba_ssm import Mamba
    USE_FAST_MAMBA = True
except ImportError:
    USE_FAST_MAMBA = False
    print("WARNING: mamba-ssm not found! Falling back to SLOW pure PyTorch Mamba.")

# ==========================================
# 1. DATASET & AUGMENTATION
# ==========================================
class DFUDataset(Dataset):
    def __init__(self, image_dir, mask_dir, image_names, transform=None):
        self.transform = transform
        self.samples = []
        
        for name in image_names:
            img_path = os.path.join(image_dir, name)
            if os.path.getsize(img_path) > 0:
                mask_path_jpg = os.path.join(mask_dir, name)
                mask_path_png = os.path.join(mask_dir, name.rsplit('.', 1)[0] + '.png')
                valid_mask_path = mask_path_png if os.path.exists(mask_path_png) else (mask_path_jpg if os.path.exists(mask_path_jpg) else None)
                self.samples.append({'img': img_path, 'mask': valid_mask_path})

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        sample = self.samples[idx]
        image = cv2.imread(sample['img'])
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) if image is not None else np.zeros((256, 256, 3), dtype=np.uint8)
        
        mask = cv2.imread(sample['mask'], cv2.IMREAD_GRAYSCALE) if sample['mask'] is not None else np.zeros(image.shape[:2], dtype=np.uint8)
        mask = (mask > 127).astype(np.float32)

        if self.transform:
            augmented = self.transform(image=image, mask=mask)
            image, mask = augmented['image'], augmented['mask'].unsqueeze(0)
            
        return image, mask

def get_train_transforms():
    return A.Compose([
        A.Resize(256, 256),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.RandomRotate90(p=0.5),
        A.Affine(scale=(0.9, 1.1), translate_percent=(-0.0625, 0.0625), rotate=(-45, 45), p=0.5),
        A.ElasticTransform(p=0.3), # Cleaned up warning
        A.HueSaturationValue(hue_shift_limit=20, sat_shift_limit=30, val_shift_limit=20, p=0.3),
        A.CLAHE(p=0.3),
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ToTensorV2()
    ])

def get_val_transforms():
    return A.Compose([
        A.Resize(256, 256),
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ToTensorV2()
    ])

# ==========================================
# 2. ADVANCED METRICS
# ==========================================
def calculate_confusion_matrix(pred, target, threshold=0.5):
    pred = (torch.sigmoid(pred.float()) > threshold).float()
    tp = (pred * target).sum().item()
    fp = (pred * (1 - target)).sum().item()
    fn = ((1 - pred) * target).sum().item()
    return tp, fp, fn

def calculate_batch_hausdorff(pred, target, threshold=0.5):
    pred = (torch.sigmoid(pred.float()) > threshold).cpu().numpy().squeeze()
    target = target.cpu().numpy().squeeze()
    
    batch_hd = []
    if pred.ndim == 2:
        pred, target = np.expand_dims(pred, 0), np.expand_dims(target, 0)
        
    for p, t in zip(pred, target):
        p_coords = np.argwhere(p)
        t_coords = np.argwhere(t)
        
        if len(p_coords) == 0 or len(t_coords) == 0:
            batch_hd.append(256.0) 
            continue
            
        hd1 = directed_hausdorff(p_coords, t_coords)[0]
        hd2 = directed_hausdorff(t_coords, p_coords)[0]
        batch_hd.append(max(hd1, hd2))
        
    return sum(batch_hd) / len(batch_hd)

# ==========================================
# 3. LOSS FUNCTIONS
# ==========================================
class FocalTverskyLoss(nn.Module):
    def __init__(self, alpha=0.3, beta=0.7, gamma=0.75, smooth=1.0): # Increased smooth for stability
        super().__init__()
        self.alpha, self.beta, self.gamma, self.smooth = alpha, beta, gamma, smooth

    def forward(self, inputs, targets):
        inputs = torch.sigmoid(inputs.float()).view(-1)
        targets = targets.float().view(-1)
        
        tp = (inputs * targets).sum()
        fp = (inputs * (1 - targets)).sum()
        fn = ((1 - inputs) * targets).sum()
        
        tversky = (tp + self.smooth) / (tp + self.alpha * fp + self.beta * fn + self.smooth)
        tversky = torch.clamp(tversky, min=1e-6, max=1.0 - 1e-6)
        return (1 - tversky) ** self.gamma

class CombinedLoss(nn.Module):
    def __init__(self):
        super().__init__()
        self.bce = nn.BCEWithLogitsLoss()
        self.tversky = FocalTverskyLoss()

    def forward(self, inputs, targets):
        # Strict float32 enforcement
        return 0.5 * self.bce(inputs.float(), targets.float()) + 0.5 * self.tversky(inputs.float(), targets.float())

# ==========================================
# 4. ARCHITECTURE
# ==========================================
class FastMambaBlock(nn.Module):
    def __init__(self, d_model):
        super().__init__()
        self.mamba = Mamba(d_model=d_model, d_state=16, d_conv=4, expand=2)
        self.norm = nn.LayerNorm(d_model)

    def forward(self, x):
        B, C, H, W = x.shape
        x_fwd = x.view(B, C, -1).transpose(1, 2)
        x_bwd = torch.flip(x_fwd, dims=[1])
        out_fwd = self.mamba(self.norm(x_fwd))
        out_bwd = torch.flip(self.mamba(self.norm(x_bwd)), dims=[1])
        out_merged = (out_fwd + out_bwd) / 2.0
        out = out_merged.transpose(1, 2).view(B, C, H, W)
        return out + x

class PurePyTorchMambaBlock(nn.Module):
    def __init__(self, d_model, d_state=16, d_conv=4, expand=2):
        super().__init__()
        self.d_model = d_model
        self.expand = expand
        self.d_inner = int(self.expand * self.d_model)
        self.norm = nn.LayerNorm(self.d_model)
        self.in_proj = nn.Linear(self.d_model, self.d_inner * 2, bias=False)
        self.conv1d = nn.Conv1d(in_channels=self.d_inner, out_channels=self.d_inner, bias=True, kernel_size=d_conv, groups=self.d_inner, padding=d_conv - 1)
        self.activation = nn.SiLU()
        self.x_proj = nn.Linear(self.d_inner, d_state * 2 + 1, bias=False)
        self.dt_proj = nn.Linear(1, self.d_inner, bias=True)
        A = torch.arange(1, d_state + 1, dtype=torch.float32).repeat(self.d_inner, 1)
        self.A_log = nn.Parameter(torch.log(A))
        self.D = nn.Parameter(torch.ones(self.d_inner))
        self.out_proj = nn.Linear(self.d_inner, self.d_model, bias=False)

    def forward(self, x_in):
        with torch.amp.autocast('cuda', enabled=False):
            x = x_in.float() 
            B, C, H, W = x.shape
            x_flat = x.flatten(2).transpose(1, 2)
            L = x_flat.size(1)
            x_norm = self.norm(x_flat)
            xz = self.in_proj(x_norm)
            x_proj, z = xz.chunk(2, dim=-1)
            x_conv = self.activation(self.conv1d(x_proj.transpose(1, 2))[:, :, :L].transpose(1, 2))
            x_state_proj = self.x_proj(x_conv)
            dt, B_state, C_state = torch.split(x_state_proj, [1, 16, 16], dim=-1)
            dt = torch.clamp(F.softplus(self.dt_proj(dt)), min=0.001, max=0.1)
            A = torch.clamp(-torch.exp(self.A_log.float()), min=-100.0, max=-1e-4)
            dA = torch.exp(torch.einsum('bld,dn->bldn', dt, A))
            dB = torch.einsum('bld,bln->bldn', dt, B_state)
            
            h = torch.zeros(B, self.d_inner, 16, device=x.device, dtype=torch.float32)
            ys = []
            for i in range(L): 
                h = dA[:, i] * h + dB[:, i] * x_conv[:, i].unsqueeze(-1)
                ys.append(torch.einsum('bdn,bn->bd', h, C_state[:, i]))
            
            y = torch.stack(ys, dim=1) + x_conv * self.D.float()
            out = self.out_proj(y * self.activation(z)) + x_flat
            return out.transpose(1, 2).view(B, C, H, W).to(x_in.dtype)

class ConvBlock(nn.Module):
    def __init__(self, in_channels, out_channels):
        super(ConvBlock, self).__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.GELU(),
            nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.GELU()
        )
    def forward(self, x):
        return self.conv(x)

class MambaUNetPlusPlus(nn.Module):
    def __init__(self, in_channels=3, num_classes=1, deep_supervision=True):
        super(MambaUNetPlusPlus, self).__init__()
        self.deep_supervision = deep_supervision
        nb_filter = [32, 64, 128, 256, 512]
        self.pool = nn.MaxPool2d(2, 2)
        self.up = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
        
        MambaLayer = FastMambaBlock if USE_FAST_MAMBA else PurePyTorchMambaBlock
        
        self.conv0_0 = ConvBlock(in_channels, nb_filter[0])
        self.conv1_0 = ConvBlock(nb_filter[0], nb_filter[1])
        self.conv2_0 = ConvBlock(nb_filter[1], nb_filter[2])
        self.conv3_0 = nn.Sequential(ConvBlock(nb_filter[2], nb_filter[3]), MambaLayer(nb_filter[3]))
        self.conv4_0 = nn.Sequential(ConvBlock(nb_filter[3], nb_filter[4]), MambaLayer(nb_filter[4]))
        
        self.conv0_1 = ConvBlock(nb_filter[0] + nb_filter[1], nb_filter[0])
        self.conv1_1 = ConvBlock(nb_filter[1] + nb_filter[2], nb_filter[1])
        self.conv2_1 = ConvBlock(nb_filter[2] + nb_filter[3], nb_filter[2])
        self.conv3_1 = ConvBlock(nb_filter[3] + nb_filter[4], nb_filter[3])
        self.conv0_2 = ConvBlock(nb_filter[0]*2 + nb_filter[1], nb_filter[0])
        self.conv1_2 = ConvBlock(nb_filter[1]*2 + nb_filter[2], nb_filter[1])
        self.conv2_2 = ConvBlock(nb_filter[2]*2 + nb_filter[3], nb_filter[2])
        self.conv0_3 = ConvBlock(nb_filter[0]*3 + nb_filter[1], nb_filter[0])
        self.conv1_3 = ConvBlock(nb_filter[1]*3 + nb_filter[2], nb_filter[1])
        self.conv0_4 = ConvBlock(nb_filter[0]*4 + nb_filter[1], nb_filter[0])
        
        if self.deep_supervision:
            self.final1 = nn.Conv2d(nb_filter[0], num_classes, kernel_size=1)
            self.final2 = nn.Conv2d(nb_filter[0], num_classes, kernel_size=1)
            self.final3 = nn.Conv2d(nb_filter[0], num_classes, kernel_size=1)
            self.final4 = nn.Conv2d(nb_filter[0], num_classes, kernel_size=1)
        else:
            self.final = nn.Conv2d(nb_filter[0], num_classes, kernel_size=1)

    def forward(self, input):
        x0_0 = self.conv0_0(input)
        x1_0 = self.conv1_0(self.pool(x0_0))
        x0_1 = self.conv0_1(torch.cat([x0_0, self.up(x1_0)], 1))
        x2_0 = self.conv2_0(self.pool(x1_0))
        x1_1 = self.conv1_1(torch.cat([x1_0, self.up(x2_0)], 1))
        x0_2 = self.conv0_2(torch.cat([x0_0, x0_1, self.up(x1_1)], 1))
        x3_0 = self.conv3_0(self.pool(x2_0))
        x2_1 = self.conv2_1(torch.cat([x2_0, self.up(x3_0)], 1))
        x1_2 = self.conv1_2(torch.cat([x1_0, x1_1, self.up(x2_1)], 1))
        x0_3 = self.conv0_3(torch.cat([x0_0, x0_1, x0_2, self.up(x1_2)], 1))
        x4_0 = self.conv4_0(self.pool(x3_0))
        x3_1 = self.conv3_1(torch.cat([x3_0, self.up(x4_0)], 1))
        x2_2 = self.conv2_2(torch.cat([x2_0, x2_1, self.up(x3_1)], 1))
        x1_3 = self.conv1_3(torch.cat([x1_0, x1_1, x1_2, self.up(x2_2)], 1))
        x0_4 = self.conv0_4(torch.cat([x0_0, x0_1, x0_2, x0_3, self.up(x1_3)], 1))
        
        if self.deep_supervision:
            return [self.final1(x0_1), self.final2(x0_2), self.final3(x0_3), self.final4(x0_4)]
        return self.final(x0_4)

# ==========================================
# 5. TRAINING LOOP & PLOTTING
# ==========================================
def plot_metrics(history, filename='training_metrics.png'):
    epochs = range(1, len(history['train_loss']) + 1)
    fig, axs = plt.subplots(2, 2, figsize=(16, 10))
    
    axs[0, 0].plot(epochs, history['train_loss'], label='Train Loss', color='red', marker='o')
    axs[0, 0].set_title('Training Loss')
    axs[0, 0].set_xlabel('Epochs')
    axs[0, 0].set_ylabel('Loss')
    axs[0, 0].legend()
    axs[0, 0].grid(True)
    
    axs[0, 1].plot(epochs, history['val_dice'], label='Val Dice (F1)', color='blue', marker='o')
    axs[0, 1].plot(epochs, history['val_iou'], label='Val IoU', color='green', marker='s')
    axs[0, 1].set_title('Segmentation Overlap Metrics')
    axs[0, 1].set_xlabel('Epochs')
    axs[0, 1].set_ylabel('Score')
    axs[0, 1].legend()
    axs[0, 1].grid(True)
    
    axs[1, 0].plot(epochs, history['val_precision'], label='Precision', color='purple', marker='^')
    axs[1, 0].plot(epochs, history['val_recall'], label='Recall', color='orange', marker='v')
    axs[1, 0].set_title('Precision & Recall')
    axs[1, 0].set_xlabel('Epochs')
    axs[1, 0].set_ylabel('Score')
    axs[1, 0].legend()
    axs[1, 0].grid(True)
    
    axs[1, 1].plot(epochs, history['val_hd'], label='Hausdorff Distance', color='brown', marker='x')
    axs[1, 1].set_title('Hausdorff Distance (Lower is better)')
    axs[1, 1].set_xlabel('Epochs')
    axs[1, 1].set_ylabel('Distance (pixels)')
    axs[1, 1].legend()
    axs[1, 1].grid(True)
    
    plt.tight_layout()
    plt.savefig(filename)
    plt.close()

def train_model():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Training on device: {device}")
    
    base_dir = 'dataset/train'
    image_dir = os.path.join(base_dir, 'images')
    mask_dir = os.path.join(base_dir, 'labels')
    if not os.path.exists(image_dir): image_dir, mask_dir = base_dir, base_dir
        
    valid_ext = ('.png', '.jpg', '.jpeg')
    all_image_names = sorted([f for f in os.listdir(image_dir) if f.lower().endswith(valid_ext)])
    if not all_image_names: raise ValueError(f"No images found in {image_dir}")
        
    random.seed(42)
    random.shuffle(all_image_names)
    split_idx = int(0.8 * len(all_image_names))
    
    train_dataset = DFUDataset(image_dir, mask_dir, all_image_names[:split_idx], transform=get_train_transforms())
    val_dataset = DFUDataset(image_dir, mask_dir, all_image_names[split_idx:], transform=get_val_transforms())
    
    train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=4, pin_memory=True)
    val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=4, pin_memory=True)
    
    model = MambaUNetPlusPlus(deep_supervision=True).to(device)
    criterion = CombinedLoss()
    
    # Reverted to CosineAnnealing with a safe, stable base LR of 2e-4
    optimizer = torch.optim.AdamW(model.parameters(), lr=2e-4, weight_decay=1e-4) 
    epochs = 50
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs, eta_min=1e-6)
    scaler = torch.amp.GradScaler('cuda')
    
    best_val_dice = 0.0
    history = {'train_loss': [], 'val_dice': [], 'val_iou': [], 'val_precision': [], 'val_recall': [], 'val_hd': []}
    ds_weights = [0.1, 0.2, 0.3, 0.4] 
    
    for epoch in range(epochs):
        model.train()
        train_loss = 0
        pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs} [Train]")
        
        for images, masks in pbar:
            images, masks = images.to(device, non_blocking=True), masks.to(device, non_blocking=True)
            optimizer.zero_grad(set_to_none=True)
            
            with torch.amp.autocast('cuda'):
                outputs = model(images)
                
            # FORCE FLOAT32 before loss to prevent NaN collapse
            if isinstance(outputs, list):
                outputs = [out.float() for out in outputs]
                loss = sum(w * criterion(out, masks.float()) for w, out in zip(ds_weights, outputs))
            else:
                loss = criterion(outputs.float(), masks.float())
            
            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            
            # STRICTER GRADIENT CLIPPING
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=0.5)
            
            scaler.step(optimizer)
            scaler.update()
            
            train_loss += loss.item()
            pbar.set_postfix({'loss': f"{loss.item():.4f}", 'lr': f"{scheduler.get_last_lr()[0]:.6f}"})
            
        scheduler.step() # Step scheduler once per epoch, not per batch
        epoch_loss = train_loss / len(train_loader)
        
        # Validation Phase
        model.eval()
        total_tp, total_fp, total_fn = 0, 0, 0
        epoch_hd = 0.0
        
        with torch.no_grad():
            for images, masks in tqdm(val_loader, desc=f"Epoch {epoch+1}/{epochs} [Val]"):
                images, masks = images.to(device, non_blocking=True), masks.to(device, non_blocking=True)
                with torch.amp.autocast('cuda'):
                    outputs = model(images)
                    final_output = outputs[-1].float() if isinstance(outputs, list) else outputs.float()
                    
                tp, fp, fn = calculate_confusion_matrix(final_output, masks)
                total_tp += tp
                total_fp += fp
                total_fn += fn
                
                batch_hd = calculate_batch_hausdorff(final_output, masks)
                epoch_hd += batch_hd
                
        precision = (total_tp + 1e-6) / (total_tp + total_fp + 1e-6)
        recall = (total_tp + 1e-6) / (total_tp + total_fn + 1e-6)
        val_dice = (2. * total_tp + 1e-6) / (2. * total_tp + total_fp + total_fn + 1e-6)
        val_iou = (total_tp + 1e-6) / (total_tp + total_fp + total_fn + 1e-6)
        avg_hd = epoch_hd / len(val_loader)
        
        history['train_loss'].append(epoch_loss)
        history['val_dice'].append(val_dice)
        history['val_iou'].append(val_iou)
        history['val_precision'].append(precision)
        history['val_recall'].append(recall)
        history['val_hd'].append(avg_hd)
        
        print(f"Loss: {epoch_loss:.4f} | Dice: {val_dice:.4f} | IoU: {val_iou:.4f} | Prec: {precision:.4f} | Rec: {recall:.4f} | HD: {avg_hd:.2f}")
        
        if val_dice > best_val_dice:
            best_val_dice = val_dice
            torch.save(model.state_dict(), 'best_mamba_unetpp.pth')
            print("--> Saved Best Model!")
                
        plot_metrics(history)

if __name__ == "__main__":
    train_model()
```
